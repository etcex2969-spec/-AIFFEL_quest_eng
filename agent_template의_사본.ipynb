```
import os
import json
import shutil
import logging
import traceback
from datetime import datetime
import sys
import importlib

import requests # NewsAPI 연동을 위한 requests 라이브러리 추가
from google.colab import userdata # Colab Secrets에서 API 키를 가져오기 위해 추가

scripts_path = os.path.abspath('/content/agent_template/agent-template/scripts/')

# Ensure the scripts directory exists
os.makedirs(scripts_path, exist_ok=True)

# Generate verify_urls.py content directly within this cell to guarantee its presence
verify_script = """
import json
import sys

def verify_news(input_path, output_path):
    with open(input_path, 'r', encoding='utf-8') as f:
        data = json.load(f)

    unique_news = {}
    for item in data:
        # URL을 키로 사용하여 중복 제거
        unique_news[item['url']] = item

    # 결과 저장
    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(list(unique_news.values()), f, ensure_ascii=False, indent=2)
    print(f'정제 완료: {len(data)} -> {len(unique_news)}')

if __name__ == '__main__':
    # 실제 운영 시에는 인자로 경로를 받음
    # verify_news(sys.argv[1], sys.argv[2])
    pass
"""

# Write/re-write verify_urls.py to ensure it exists
verify_urls_file_path = os.path.join(scripts_path, 'verify_urls.py')
with open(verify_urls_file_path, 'w', encoding='utf-8') as f:
    f.write(verify_script)
print(f"DEBUG: Ensured {verify_urls_file_path} exists.")


# Ensure the scripts directory is at the beginning of sys.path, removing it first if it exists
if scripts_path in sys.path:
    sys.path.remove(scripts_path)
sys.path.insert(0, scripts_path)

# Invalidate import caches to ensure Python re-scans for modules
importlib.invalidate_caches()

# Remove the module from sys.modules if it was partially loaded or cached as missing
if 'verify_urls' in sys.modules:
    del sys.modules['verify_urls']
    print(f"DEBUG: Removed 'verify_urls' from sys.modules.")

# Now, perform a standard import.
try:
    from verify_urls import verify_news
    print("DEBUG: Standard import of verify_urls successful.")
except ModuleNotFoundError as e:
    print(f"DEBUG: Standard import of verify_urls failed: {e}")
    raise # Re-raise if it still fails

class NewsDigestAgent:
    def __init__(self, base_path='/content/agent_template/agent-template'):
        self.base_path = base_path
        self.run_id = None
        self.output_dir = None
        self.logger = None
        self.final_reports_dir = os.path.join(self.base_path, 'output/final_reports')
        os.makedirs(self.final_reports_dir, exist_ok=True)

    def _setup_logger(self):
        """실행별 로그 파일 설정"""
        log_path = os.path.join(self.output_dir, 'run.log')
        # Get a unique logger for each run to avoid duplicate handlers
        self.logger = logging.getLogger(self.run_id or 'default_run') # Use run_id if available, else default
        self.logger.setLevel(logging.INFO)

        # Clear existing handlers to prevent duplicate log entries if called multiple times
        if not self.logger.handlers:
            # 파일 핸들러 추가
            fh = logging.FileHandler(log_path, encoding='utf-8')
            # 상세한 로그 형식으로 수정 (파일명 및 라인 번호 포함)
            formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(filename)s:%(lineno)d - %(message)s')
            fh.setFormatter(formatter)
            self.logger.addHandler(fh)

            # 콘솔 출력 핸들러
            ch = logging.StreamHandler() # Changed to StreamHandler
            ch.setFormatter(formatter)
            self.logger.addHandler(ch)

    def _init_run(self):
        self.run_id = datetime.now().strftime('%Y%m%d_%H%M%S')
        self.output_dir = os.path.join(self.base_path, 'output', self.run_id)
        os.makedirs(self.output_dir, exist_ok=True)
        self._setup_logger()
        self.logger.info(f"Pipeline initialized. Run ID: {self.run_id}")

    def _send_error_notification(self, error_message):
        """
        Simulates sending an error notification.
        In a real scenario, this would send an email, Slack message, etc.
        """
        notification_message = f"[CRITICAL ERROR] NewsDigestAgent Pipeline Failed (Run ID: {self.run_id}): {error_message}"
        print(f"\n--- ERROR NOTIFICATION ---")
        print(notification_message)
        print(f"--------------------------")
        if self.logger:
            self.logger.critical(f"NOTIFICATION: {notification_message}")

    def collect(self):
        self.logger.info("Step 1: Starting collection (using NewsAPI)...")

        # NewsAPI 키를 Colab Secrets에서 가져옵니다.
        NEWS_API_KEY = userdata.get('NEWS_API_KEY')
        if not NEWS_API_KEY:
            raise ValueError("NEWS_API_KEY가 Colab Secrets에 설정되지 않았습니다.")

        # NewsAPI 설정
        # 최근 AI 트렌드에 대한 뉴스 검색 (예시)
        query = "artificial intelligence OR machine learning"
        # 언론사 필터링 (선택 사항, 예: techcrunch, wired)
        # sources = "techcrunch,wired"
        language = "en"
        page_size = 10

        url = f"https://newsapi.org/v2/everything?q={query}&language={language}&pageSize={page_size}&apiKey={NEWS_API_KEY}"

        try:
            response = requests.get(url)
            response.raise_for_status() # HTTP 오류가 발생하면 예외를 발생시킵니다.
            news_data = response.json()
        except requests.exceptions.RequestException as e:
            self.logger.error(f"NewsAPI 호출 중 오류 발생: {e}")
            raise

        collected_articles = []
        for article in news_data.get('articles', []) :
            # 필요한 정보만 추출하여 형식에 맞게 저장
            title = article.get('title')
            url = article.get('url')
            published_at = article.get('publishedAt')
            source_name = article.get('source', {}).get('name')

            # 필수 필드가 누락되지 않았는지 확인
            if all([title, url, published_at, source_name]):
                collected_articles.append({
                    "title": title,
                    "url": url,
                    "date": published_at.split('T')[0], # 날짜만 추출
                    "source": source_name
                })

        raw_path = os.path.join(self.output_dir, '01_raw_news.json')
        with open(raw_path, 'w', encoding='utf-8') as f:
            json.dump(collected_articles, f, ensure_ascii=False, indent=2)
        self.logger.info(f"Step 1: Collection finished. {len(collected_articles)} items found.")
        return raw_path

    def verify(self, input_path):
        self.logger.info("Step 2: Starting verification...")
        output_path = os.path.join(self.output_dir, '02_filtered_news.json')
        verify_news(input_path, output_path)
        self.logger.info("Step 2: Verification finished.")
        return output_path

    def summarize(self, input_path):
        self.logger.info("Step 3: Starting summarization...")
        output_path = os.path.join(self.output_dir, '03_summary.md')
        with open(output_path, 'w', encoding='utf-8') as f:
            f.write("# 주간 요약 리포트\n\n로깅 기능이 추가된 에이전트의 결과물입니다.")
        self.logger.info("Step 3: Summarization finished.")
        return output_path

    def publish(self, summary_path):
        self.logger.info("Step 4: Starting publication...")
        report_name = f"report_{self.run_id}.md"
        final_path = os.path.join(self.final_reports_dir, report_name)
        shutil.copy(summary_path, final_path)

        receipt = {
            "run_id": self.run_id,
            "status": "success",
            "timestamp": datetime.now().isoformat(),
            "file": final_path
        }
        with open(os.path.join(self.output_dir, 'publish_receipt.json'), 'w') as f:
            json.dump(receipt, f, indent=2)
        self.logger.info(f"Step 4: Publication successful. Report saved at {final_path}")
        return final_path

    def run(self):
        try:
            self._init_run()
            raw = self.collect()
            filtered = self.verify(raw)
            summary = self.summarize(filtered)
            final = self.publish(summary)
            self.logger.info("All pipeline stages completed successfully.")
        except Exception as e:
            error_message = str(e)
            if self.logger:
                self.logger.error(f"Pipeline failed: {error_message}")
                self.logger.error("Traceback:\n" + traceback.format_exc())
            else:
                print(f"Initialization failed or logger not available: {e}")
                print("Traceback:\n" + traceback.format_exc())
            self._send_error_notification(error_message)

# 실행 및 로그 확인
agent = NewsDigestAgent()
agent.run()
```

```
def filter_step_logs(log_file_path, step_name=None):
    """
    Filters log entries from a run.log file for a specific step.

    Args:
        log_file_path (str): The path to the run.log file.
        step_name (str, optional): The name of the step to filter (e.g., 'Step 1', 'Step 2').
                                   If None, all logs are returned.

    Returns:
        list: A list of log lines matching the filter criteria.
    """
    filtered_logs = []
    if not os.path.exists(log_file_path):
        print(f"Log file not found: {log_file_path}")
        return []

    with open(log_file_path, 'r', encoding='utf-8') as f:
        for line in f:
            if step_name is None or step_name in line:
                filtered_logs.append(line.strip())
    return filtered_logs
```

```
import re
from datetime import datetime

def summarize_pipeline_logs(log_file_path):
    """
    Summarizes the execution of a pipeline by parsing its run.log file.

    Args:
        log_file_path (str): The path to the run.log file.

    Returns:
        dict: A dictionary containing a summary of each step and the overall pipeline status.
    """
    summary = {
        "run_id": "N/A",
        "start_time": "N/A",
        "end_time": "N/A",
        "duration": "N/A",
        "status": "Unknown",
        "steps": {}
    }

    if not os.path.exists(log_file_path):
        summary["status"] = f"Error: Log file not found at {log_file_path}"
        return summary

    with open(log_file_path, 'r', encoding='utf-8') as f:
        logs = f.readlines()

    step_pattern = re.compile(r"Step (\d+): (Starting|Collection finished|Verification finished|Summarization finished|Publication successful)")
    timestamp_pattern = re.compile(r"^(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2},\d{3})")

    log_start_time = None
    log_end_time = None

    for line in logs:
        if "Pipeline initialized. Run ID:" in line and summary["run_id"] == "N/A":
            match = re.search(r"Run ID: (\d{8}_\d{6})", line)
            if match:
                summary["run_id"] = match.group(1)
            match_ts = timestamp_pattern.match(line)
            if match_ts:
                log_start_time = datetime.strptime(match_ts.group(1), '%Y-%m-%d %H:%M:%S,%f')
                summary["start_time"] = log_start_time.isoformat()

        if "All pipeline stages completed successfully." in line:
            summary["status"] = "Success"
            match_ts = timestamp_pattern.match(line)
            if match_ts:
                log_end_time = datetime.strptime(match_ts.group(1), '%Y-%m-%d %H:%M:%S,%f')
                summary["end_time"] = log_end_time.isoformat()

        if "Pipeline failed:" in line:
            summary["status"] = "Failed"
            match_ts = timestamp_pattern.match(line)
            if match_ts:
                log_end_time = datetime.strptime(match_ts.group(1), '%Y-%m-%d %H:%M:%S,%f')
                summary["end_time"] = log_end_time.isoformat()

        step_match = step_pattern.search(line)
        if step_match:
            step_num = step_match.group(1)
            step_action = step_match.group(2)
            if step_num not in summary["steps"]:
                summary["steps"][step_num] = {"start": "N/A", "end": "N/A", "detail": []}

            match_ts = timestamp_pattern.match(line)
            if match_ts:
                current_time = datetime.strptime(match_ts.group(1), '%Y-%m-%d %H:%M:%S,%f')
                if "Starting" in step_action:
                    summary["steps"][step_num]["start"] = current_time.isoformat()
                elif "finished" in step_action or "successful" in step_action:
                    summary["steps"][step_num]["end"] = current_time.isoformat()

            summary["steps"][step_num]["detail"].append(line.strip())

    # Calculate overall duration if possible
    if log_start_time and log_end_time:
        duration = log_end_time - log_start_time
        summary["duration"] = str(duration)

    return summary
```

```
def search_errors_in_logs(log_file_paths):
    """
    Searches for 'Error' or 'Exception' (case-insensitive) in a list of log files.

    Args:
        log_file_paths (list): A list of paths to log files.

    Returns:
        dict: A dictionary where keys are log file paths and values are lists of
              matching log lines. Only files with matching lines are included.
    """
    error_logs = {}
    keywords = ['error', 'exception']

    for log_path in log_file_paths:
        found_lines = []
        if not os.path.exists(log_path):
            found_lines.append(f"Log file not found: {log_path}")
            continue

        with open(log_path, 'r', encoding='utf-8', errors='ignore') as f:
            for line_num, line in enumerate(f, 1):
                if any(keyword in line.lower() for keyword in keywords):
                    found_lines.append(f"Line {line_num}: {line.strip()}")

        if found_lines:
            error_logs[log_path] = found_lines

    return error_logs
```

```
# 이전에 정렬된 로그 파일 목록을 사용하여 함수를 실행합니다.
# Kernel State에 'sorted_log_files' 변수가 있다고 가정합니다.
if 'sorted_log_files' in locals():
    found_errors = search_errors_in_logs(sorted_log_files)

    if found_errors:
        print("--- Found 'Error' or 'Exception' in the following log files ---")
        for file_path, lines in found_errors.items():
            print(f"\nFile: {file_path}")
            for line in lines:
                print(f"  {line}")
    else:
        print("No 'Error' or 'Exception' keywords found in any log files.")
else:
    print("Variable 'sorted_log_files' not found. Please ensure the previous cell was run.")
```
