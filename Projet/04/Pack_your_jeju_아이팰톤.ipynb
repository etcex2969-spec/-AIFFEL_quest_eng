```markdown
## "제주를 담다" 최종 발표 스크립트

---

### 1. 문제 인식: 정보가 너무 많아도 문제, '신뢰할 수 있는 판단 기준'이 없어요. (1분)

안녕하십니까, 저희는 "제주를 담다" 팀입니다.

제주 여행, 설레는 만큼 준비할 것도 많죠? 블로그, SNS, 각종 포털… 정보는 정말 넘쳐납니다. 하지만 저희는 여기서 근본적인 질문을 던졌습니다. **"지금, 나에게, 이 장소가 믿을 수 있는가?"** 과연 이 많은 정보 속에서 여행자들은 믿고 실행할 판단 기준을 가지고 있을까요?

많은 정보 속에는 광고성 콘텐츠, 오래된 정보, 심지어는 부정확한 내용도 많습니다. 특히 운영 시간, 휴무, 가격 같은 핵심 정보가 최신이 아니거나, 내 상황(날씨, 동반자, 이동수단)에 맞지 않는다면 소중한 여행을 망칠 수도 있습니다.

저희는 단순히 '좋은 모델을 찾는 것'보다, **'같은 모델(gpt-5-mini)이라도 믿을 만하게 작동하는 시스템 구조를 만드는 것'**이 제주 여행 정보의 신뢰도를 높이는 핵심이라고 보았습니다.

### 2. 서비스 핵심 정의: '제주를 담다', 계속 좋아지는 AI 여행 플랜 서비스 (1분 30초)

그래서 저희 "제주를 담다"는 추천에서 끝나지 않고, **공공데이터와 사용자 방문 기록이 서로 연결되어 계속 좋아지는 '순환형 AI 여행 플랜 서비스'**를 제안합니다.

저희 서비스의 목표는 단순히 유명한 곳을 알려주는 것을 넘어, **사용자들이 '믿고 실행할 수 있는 여행 판단을 할 수 있도록 돕는 것'**입니다.

서비스의 핵심 작동 방식은 다음과 같습니다. 먼저 공공데이터로 여행 후보를 추천하고, 사용자는 후보의 신뢰도와 출처를 확인한 뒤 플랜에 추가합니다. 그리고 실제 여행을 다녀와서 방문 기록을 남기면, 이 기록이 검증을 거쳐 다음 여행자를 위한 '신뢰 신호'로 다시 시스템에 반영되는 거죠. 이처럼 계속 순환하면서 서비스는 점점 더 좋아집니다.

### 3. 데이터 전략: '판단과 검증'에 딱 필요한 데이터만 모았어요. (1분 30초)

저희는 데이터를 무작정 많이 모으기보다 **'여행자가 실제로 판단할 때 쓸 수 있는 데이터'만 선별**했습니다.

핵심 데이터는 크게 다섯 가지입니다.

1.  **비짓제주 관광지 데이터**: 장소의 기본 정보죠.
2.  **비짓제주 수정요청 데이터**: 이 데이터가 정말 중요합니다. 공식 데이터도 사실 계속 바뀝니다. 비짓제주에서만 **1,686건의 수정요청**을 발견했어요! 이는 단순한 관리 기록이 아니라, 여행자 입장에서는 정보의 **최신성 위험을 판단하는 신호**입니다.
3.  **이동과 편의 데이터**: 주차장, 정류장, 접근성 정보 등 **실제 여행 가능성을 판단**하는 데 필수적인 데이터입니다.
4.  **기상청 날씨 API**: 비, 바람, 강수 확률 등 오름이나 해변 같은 장소의 **적합도를 변화시키는 정보**입니다.
5.  **사용자 피드백과 검증 신호**: 방문 여부, 정보 다름, 만족도 등 저희 서비스의 신뢰도 순환에 핵심적인 역할을 합니다.

결론적으로, 저희 서비스는 출처, 최신성, 수정 이력, '확인 필요' 여부까지 **함께 보면서** 지금 이 정보가 유효한지 판단합니다.

### 4. 사용자 흐름 (데모 요약): 판단하고, 담고, 기록하면서 신뢰를 만들어가요. (2분)

사용자는 단순히 추천을 소비하는 것이 아니라, 판단하고 실행하며 기록을 남기면서 정보의 신뢰도를 함께 만들어갑니다.

1.  **고르기**: 제주 지도에서 지역과 순간(예: 한림 지역, 혼자 여행, 힐링, 하루 일정)을 선택합니다. 이때, 단순히 추천 목록만 보여주는 게 아니라 **'이 지역에서 무엇을 얼마나 믿고 고를 수 있는지'**를 보여주는 **'지역 신뢰 대시보드'**를 제공합니다. 데이터가 부족하면 억지로 추천하지 않고 '확인 필요'로 솔직하게 표시합니다.
2.  **근거 확인**: 후보 카드에서는 단순 추천이 아닌 **신뢰도 점수, 공공데이터 출처, 수정요청 이력, 이용/주차 여부, '확인 필요' 항목** 등 근거 정보를 먼저 꼼꼼히 확인합니다. 사용자는 이러한 근거를 바탕으로 '이 장소가 내 조건에 맞는지', '확인해야 할 위험은 무엇인지'를 판단하여 선택합니다.
3.  **플랜 담기**: 선택한 후보는 단순 북마크가 아니라 '실제로 실행 가능한 계획'인 **'여행팩'**에 담깁니다.
4.  **방문 기록**: 실제 방문 후 기록을 남기면, 이 기록은 **즉시 반영되는 게 아니라 내부 검증 과정을 거쳐** 신뢰도 업데이트와 공공데이터 개선 신호로 연결됩니다. 악성 입력이나 단순 착오를 방지하기 위함입니다.

이 모든 과정에서 **'하루방 에이전트'**는 무엇이든 다 아는 챗봇이 아닙니다. 조회된 근거 안에서만 정보를 조율하고 설명하는 **'여행 판단 에이전트'**예요. 모르는 내용은 '확인 필요'로 명확히 알려주며 사용자의 합리적 판단을 돕습니다.

### 5. 시스템 구조: RAG + Trust Engine + gpt-5-mini (1분 30초)

저희 시스템의 핵심은 LLM이 먼저 대답하는 구조가 아니라는 점입니다. **데이터와 규칙(Trust Engine)이 먼저 판단하고, LLM(gpt-5-mini)은 마지막에 이를 설명하는 구조**입니다.

1.  **사용자 입력 구조화**: 사용자 입력이 들어오면 바로 LLM으로 보내지 않고, 지역, 동행자, 목적 등 **조건으로 구조화**합니다.
2.  **RAG (Retrieval-Augmented Generation) 작동**: 구조화된 입력을 바탕으로 RAG가 공공데이터와 API에서 **후보 장소, 출처, 근거 문서, 이동 및 주차 정보, 수정 요청 이력, 방문 피드백 신호** 등 다각적인 근거를 찾아옵니다.
3.  **Trust Engine 판단**: Trust Engine은 찾아온 정보를 기반으로 **신뢰도 점수, '확인 필요' 여부, 아예 배제할 조건** 등을 정교하게 계산하여 후보를 걸러냅니다. `public_data_quality`, `last_verified_date`, `feedback_summary`, `suitability_tags`, `discrepancy_status`와 같은 다차원적인 지표를 종합적으로 평가합니다. 이는 '가도 될 만한 곳'을 선별하고 '가서는 안 될 곳'을 미리 걸러내는 의사결정의 핵심입니다.
4.  **gpt-5-mini 역할 제한**: 마지막으로 gpt-5-mini는 새로운 후보를 만들지 않습니다. 이미 조회되고 계산된 근거를 바탕으로 사용자에게 **이해하기 쉬운 말로 '설명'해주고, 여행 계획 방향을 '조율'하는 역할**만 수행합니다. 근거 범위 밖의 답변은 가드레일이 차단하며, '확인 필요' 정보는 투명하게 고지합니다.

이러한 **RAG → Trust Engine → gpt-5-mini**의 순차적이고 상호 보완적인 구조 덕분에, 저희는 같은 `gpt-5-mini` 모델을 쓰더라도 **답변의 일관성과 정확성, 검증 가능성을 획기적으로 높일 수 있었습니다.**

### 6. 신뢰성 검증 및 시뮬레이션 결과 (1분 30초)

저희 "제주를 담다"가 제공하는 신뢰도는 단순히 '별점'이 아닙니다. **공식성, 최신성, 출처, 조건 적합성, 피드백 신호 등 각 평가 항목별 점수를 통해 왜 그런 점수가 나왔는지 설명 가능한 구조**입니다.

저희는 이 신뢰도 향상 전략의 효과를 정량적으로 입증하기 위해 시뮬레이션을 수행했습니다.

*   **'성산일출봉 (야외)' 플랜**: 초기 **0점**이었던 이 플랜은 '정보 불일치 탐지 및 경고' 해결, '정보 최신성' 해결, '부정적 피드백 해소 및 긍정 피드백 누적', '공공데이터 품질 개선', '날씨 적합성 해결', '운영시간 일치 해결' 등 단계적인 개선을 통해 최종 **100점**에 도달하는 것을 증명했습니다.
*   **'협재 해변' 플랜**: 초기 65점이었던 이 플랜 역시 '공공데이터 품질 개선' 및 '날씨 조건 반영'이라는 핵심 문제 해결을 통해 최종 **100점**에 도달했습니다.

이러한 시뮬레이션 결과는 저희의 신뢰도 평가 모델과 개선 전략이 **실제로 효과적이며, 어떠한 플랜이라도 체계적인 관리를 통해 높은 신뢰도를 확보할 수 있음**을 명확히 보여줍니다.

### 7. Q&A 준비 안내

저희는 심사위원 여러분의 날카로운 질문들을 기대하고 있습니다. 특히 LLM의 한계, 사용자 피드백 검증의 견고성, 그리고 저희 MVP의 실제 구현 범위와 로드맵에 대한 질문에 철저히 대비하고 있습니다. 기술적 깊이와 서비스 철학을 결합하여 명확한 답변을 드릴 준비가 되어 있습니다.

### 8. 최종 메시지: '제주를 담다', 신뢰로 그리는 여행의 미래 (1분)

저희 "제주를 담다"는 **억지 추천을 지양하고, 확인 필요를 분리하며, 검증된 피드백을 통해 지속적으로 개선되는 서비스**입니다.

외부 API 호출 실패 시에도 대체 UI와 안내 문구로 흐름을 유지하고, 정보가 부족하면 '확인 필요'로 솔직하게 표시합니다. API 키는 Secrets로 안전하게 관리하며, 민감 정보는 최소화하여 내부 신호로만 처리하는 등 서비스의 견고성과 보안에도 만전을 기하고 있습니다.

결론적으로, "제주를 담다"는 더 많이 추천하는 서비스가 아닙니다. **공공데이터로 계획하고, 사용자의 방문 기록으로 다시 좋아지는 '신뢰 기반 여행 서비스'** 입니다. 저희 "제주를 담다"에 귀 기울여 주셔서 정말 감사합니다. 함께 더 신뢰할 수 있는 여행의 미래를 만들어가고 싶습니다!

---
```

## 제주를 담다 (Jeju Damda) 핵심 요약

**서비스 명**: 제주를 담다

**핵심 메시지**:
*   공공데이터를 활용해 여행을 계획하고, 사용자의 실제 방문 기록과 피드백을 통해 정보의 신뢰도를 지속적으로 높이는 순환형 AI 여행 플랜 서비스입니다.

**문제 인식**:
*   제주 여행 정보는 많지만, 사용자들은 '오늘 이 장소를 믿고 선택해도 되는지'에 대한 명확한 근거를 필요로 합니다.

**해결 방안**:
*   공공데이터 기반의 여행 플랜을 제공하고, 사용자의 실제 방문 기록과 피드백을 '신뢰 신호'로 시스템에 반영하여 서비스의 데이터와 판단 기준을 끊임없이 개선합니다.

**차별점**:
*   단순한 추천을 넘어, 사용할수록 정보의 신뢰도와 정확성이 향상되는 시스템을 구축하여 사용자에게 더 나은 여행 경험을 제공합니다.

| 3. gpt-5-mini 활용 방식 및 역할 제한에 대해 설명해 주세요. | LLM의 오용 방지 및 서비스의 핵심 가치(신뢰성) 유지를 위한 LLM 활용 전략 확인 | (문서 내용 기반으로) LLM이 사실을 창작하지 않도록 역할이 제한되며, RAG와 Trust Engine 후 최종 설명 역할 수행 |
| 4. RAG/Trust Engine은 어떤 역할을 하며, 어떻게 작동하나요? | 서비스의 핵심 기술인 RAG와 Trust Engine의 기능 및 신뢰도 판단 메커니즘 이해 | (문서 내용 기반으로) RAG는 근거 데이터 조회, Trust Engine은 신뢰도 점수 계산 및 필터링 담당 |
| 5. 서비스의 검증 가능성은 어떻게 확보되나요? | 시스템의 판단 기준과 결과에 대한 투명성 및 검증 로직 확인 | (문서 내용 기반으로) 신뢰도 점수는 산식과 로그로 설명 가능하며, 사용자 피드백은 검증 과정을 거쳐 반영 |
| 6. 데모의 견고성은 어느 정도로 구현되었나요? | 실제 서비스 환경에서의 안정성과 오류 처리 능력 확인 | (문서 내용 기반으로) 외부 API 오류 시에도 대체 UI로 흐름 유지, 정보 부족 시 '확인 필요'로 명시 |
| 7. 개인 정보 및 데이터 보안은 어떻게 관리되고 있나요? | 사용자 데이터 및 시스템의 보안 취약점 방지 대책 확인 | (문서 내용 기반으로) API 키 Secrets 관리, 민감 정보 최소화 및 내부 신호 처리 |
| 8. 향후 서비스 확장성 계획은 무엇인가요? | 서비스의 잠재력 및 추가 발전 가능성 파악 | (문서 내용 기반으로) 제주 내 다른 카테고리 확장 및 다른 지역으로의 모델 적용 가능성 |
| 9. 신뢰 점수 산정의 실제 산식과 로그가 명확하게 설명될 수 있나요? | 신뢰도 점수의 객관성과 투명성 검증 | (문서 내용 기반으로) 신뢰도는 단순 별점이 아닌 공식성, 최신성 등 평가 항목별 점수로 설명 가능 |
| 10. 공공데이터에 오류가 발생했을 때의 보정 구조는 마련되어 있나요? | 데이터 오류 발생 시의 대응 체계 및 신뢰도 유지 방안 확인 | (문서 내용 기반으로) 사용자 피드백을 통해 정보 불일치를 감지하고 검증 후 보정 구조로 연결 |

## 반박 질문에 대한 모범 답안

### 1. GPT-5-mini의 역할 제한에도 불구하고 LLM의 본질적인 한계(예: 환각 현상)가 발생할 가능성은 없습니까? 이를 기술적으로 어떻게 완벽하게 방어하고 있으며, 구체적인 비상 대응 메커니즘은 무엇인가요?

**모범 답안:**
LLM의 환각 현상과 같은 본질적인 한계는 저희 'Pack Your Jeju' 팀이 가장 중요하게 고려한 부분입니다. 저희는 이를 **구조적인 설계**로 방어하고 있습니다. gpt-5-mini가 직접 정보를 '생성'하거나 '추천'하는 것이 아니라, 오직 **Trust Engine에서 검증된 근거 데이터**를 바탕으로 사용자에게 '설명'하고 '조율'하는 역할만 하도록 엄격하게 제한했습니다.

구체적으로:
*   **RAG 엔진 선행**: 사용자의 쿼리가 들어오면, RAG 엔진이 가장 먼저 공공데이터와 외부 API에서 **신뢰할 수 있는 근거 문서**를 조회하고 필터링합니다. LLM은 이 조회된 근거 텍스트를 벗어나는 답변을 생성할 수 없습니다.
*   **Trust Engine의 판단**: 조회된 근거들은 Trust Engine을 거쳐 신뢰도 점수를 산정하고, '확인 필요' 또는 '배제'와 같은 판단을 내립니다. LLM은 Trust Engine이 '확인 필요'로 분류한 정보에 대해서는 이를 사용자에게 솔직하게 알리고, 추측성 답변을 배제하도록 학습되었습니다.
*   **비상 대응 메커니즘**: 만약 LLM이 근거를 벗어난 답변을 생성하려 시도하는 경우, 내부적으로 **가드레일(Guardrails)**을 두어 답변을 차단하고, 사용자에게는 '현재 정보로는 정확한 답변을 드릴 수 없습니다'와 같은 중립적인 메시지를 제공합니다. 또한, 지속적인 **Ragas 평가 지표** 모니터링을 통해 환각 가능성을 탐지하고 모델의 동작을 재조정합니다. 저희는 LLM의 완벽한 방어보다는, **오류의 가능성을 최소화하고 불확실성을 사용자에게 투명하게 전달**하는 것을 목표로 합니다.

### 2. 사용자 피드백 검증 과정의 효율성과 공정성을 어떻게 담보할 수 있으며, 만약 대량의 의도적인 허위 피드백이 발생할 경우 이 '검증 레이어'가 얼마나 견고하게 작동할 수 있습니까?

**모범 답안:**
사용자 피드백의 품질 관리는 'Pack Your Jeju'의 신뢰 순환 구조에서 가장 핵심적인 부분입니다. 저희는 이를 위해 **다단계 검증 프로세스**와 **이상 탐지 메커니즘**을 설계했습니다.

*   **내부 피드백 대시보드 운영**: 모든 사용자 피드백은 즉시 서비스에 반영되지 않고, '검증 완료', '검증 필요', '반영 제외'로 분류되는 내부 대시보드로 유입됩니다. 이는 1차적인 필터 역할을 합니다.
*   **다각적 검증**: '검증 필요'로 분류된 피드백은 AI 기반의 자동 검증 시스템(예: 입력된 텍스트와 기존 데이터의 키워드/패턴 분석)과 필요시 운영자의 수동 검토를 통해 교차 확인됩니다. 이때, **사진, GPS 위치 정보, 시간 스탬프** 등 다른 데이터 소스와 대조하여 피드백의 신뢰도를 높입니다. 예를 들어, '영업 종료'라는 피드백이 들어오면 해당 장소의 최신 운영 시간 데이터나 유사한 다른 피드백과 비교하여 진위를 판단합니다.
*   **허위 피드백 방어**: 대량의 의도적인 허위 피드백 발생 시, 저희 시스템은 다음과 같은 방식으로 대응합니다.
    *   **패턴 분석**: 단시간 내 유사하거나 비정상적인 패턴의 피드백이 감지될 경우, 해당 계정 또는 피드백 소스를 **자동으로 '반영 제외' 처리하고 경고**합니다.
    *   **가중치 조절**: 신뢰도 점수 산정 시, 피드백을 제공한 사용자의 **활동 이력과 신뢰도**에 따라 피드백에 부여하는 가중치를 차등화합니다. 악의적인 의도를 가진 사용자의 피드백은 점수에 미치는 영향이 최소화됩니다.
    *   **커뮤니티 검증 (장기적)**: 향후 고도화 단계에서는 특정 피드백에 대해 다수 사용자의 검증을 요청하는 커뮤니티 기반의 크라우드소싱 검증 방안도 고려하고 있습니다. 이처럼 저희는 피드백의 효율성과 공정성을 담보하며, 허위 정보로 인한 데이터 오염을 적극적으로 방어하는 견고한 구조를 갖추고 있습니다.

### 3. 제시된 아키텍처(RAG, Trust Engine, 하루방 에이전트 등)와 기능들 중 현재 MVP 단계에서 실질적으로 구현되어 작동하는 범위는 어디까지이며, 나머지 고도화될 기능들은 향후 어떤 구체적인 로드맵과 마일스톤으로 발전시켜 나갈 계획이신가요?

**모범 답안:**
저희 'Pack Your Jeju' 프로젝트는 **핵심 신뢰도 순환 구조의 MVP 구현**에 집중했습니다. 현재 MVP 단계에서 실질적으로 구현되어 작동하는 주요 범위는 다음과 같습니다:

*   **공공데이터 기반 RAG 엔진**: 비짓제주 관광지 데이터와 수정 요청 데이터, 그리고 기본적인 이동/편의 데이터를 기반으로 한 RAG 엔진이 구현되어 LLM에 근거를 제공합니다.
*   **Trust Engine (핵심 로직)**: 신뢰도 점수 산정의 **기본 로직**이 구현되어 있으며, 공공데이터 품질, 최신성, 일부 상황 적합성(예: 날씨) 및 사용자 피드백 신호(긍정/부정 여부)를 반영하여 점수를 계산하고 '확인 필요' 상태를 분류합니다. (세부 평가 항목별 가중치는 1차적으로 적용)
*   **하루방 에이전트 (근거 기반 설명)**: gpt-5-mini를 활용하여 Trust Engine에서 필터링되고 점수화된 정보를 바탕으로 사용자에게 '설명'하는 기능이 구현되어 있습니다. 즉, 근거 안에서만 조율하여 설명하고, 모르는 것은 '확인 필요'로 명시합니다.
*   **사용자 피드백 수집 및 내부 대시보드 (수동 검증)**: 사용자가 방문 기록을 남길 수 있으며, 이 피드백이 내부 대시보드에 기록되어 운영자가 수동으로 검토할 수 있는 초기 단계의 검증 프로세스가 작동합니다.

나머지 고도화될 기능들과 로드맵은 다음과 같습니다:

*   **단기 로드맵 (MVP 이후 3개월)**:
    *   **Trust Engine 고도화**: 더 많은 `suitability_tags` (동반자, 이동수단 등) 및 `discrepancy_status` (불일치 유형별)를 신뢰도 점수 산정에 정교하게 반영하고, 가중치를 동적으로 조절하는 시스템을 구축합니다.
    *   **하루방 에이전트 개인화**: 사용자 프로필 데이터(예: 이전 여행 스타일, 선호도)를 Trust Engine에 연동하여 하루방 에이전트의 설명을 더욱 개인화하고 맞춤형 대안을 제시할 수 있도록 발전시킵니다.
    *   **자동 검증 시스템 도입**: 머신러닝 기반의 자동 검증 모듈을 개발하여 대량의 사용자 피드백을 효율적으로 분류하고, 허위 피드백을 사전에 탐지하는 정확도를 높입니다.

*   **중장기 로드맵 (MVP 이후 6개월 이상)**:
    *   **데이터 확장**: 숙소, 음식점, 교통, 축제 등 제주 내 다른 카테고리의 공공데이터를 연동하고 Trust Engine에 통합합니다. 이를 통해 통합적인 여행 플랜을 제공할 수 있도록 확장합니다.
    *   **실시간 데이터 연동 강화**: 지도 API를 활용한 실시간 교통 정보, 이벤트 API를 통한 지역 축제 정보 등 실시간 변동성이 큰 데이터를 Trust Engine에 실시간으로 반영하여 '실행 가능성' 점수를 더욱 정교화합니다.
    *   **타 지역 확장**: 제주도에서의 성공적인 모델을 바탕으로 전국 단위의 공공데이터 및 지역 특성을 반영한 Trust Engine을 구축하여 서비스 지역을 확대합니다.

저희는 MVP에서 핵심적인 신뢰 순환 구조를 증명하는 데 주력했으며, 앞으로 이러한 로드맵을 통해 서비스의 기능과 신뢰도를 지속적으로 고도화해 나갈 계획입니다.

## Pack Your Jeju 프로젝트 아키텍처 (개념 다이어그램 설명)

**1. 사용자 인터페이스 (Frontend Layer)**
*   **웹 애플리케이션 (Web Application)**: `React.js` 또는 `Next.js`로 개발된 웹 앱으로, 사용자가 여행 플랜을 생성하고, 추천 근거를 확인하며, 방문 기록을 입력하는 등의 모든 상호작용이 이루어지는 부분입니다.
    *   **주요 기능**: 여행 플랜 조회, 장소 상세 정보 표시, 지도 연동, 사용자 피드백 입력 UI (방문/미방문/변경/정보 불일치).

**2. 백엔드 서비스 (Backend & Application Logic Layer)**
*   **API 서버 (API Server)**: `Python (FastAPI/Flask)` 기반으로 개발됩니다. 프론트엔드와 AI/DB 및 외부 서비스 간의 모든 통신을 처리하는 핵심 로직을 담당합니다.
    *   **주요 역할**: 사용자 요청 처리, 데이터베이스 쿼리, LLM 호출, 외부 API 연동, 인증/인가 관리.
*   **RAG (Retrieval-Augmented Generation) 엔진**: `LangChain` 또는 `LlamaIndex`를 활용하여 구현됩니다.
    *   **주요 역할**: 사용자 쿼리 및 여행 플랜 컨텍스트를 분석하여 벡터 데이터베이스에서 관련 공공데이터 문서를 검색하고, 이를 LLM에 전달하여 정확하고 근거 있는 답변 및 추천을 생성합니다.
*   **하루방 에이전트 모듈**: LLM (`Google Gemini API`)을 활용하여 사용자의 특정 조건(날씨, 동반자, 렌터카 등)에 맞춰 공공데이터를 해석하고 개인화된 설명을 생성하는 역할을 합니다.

**3. 데이터 계층 (Data Layer)**
*   **주요 관계형 데이터베이스 (Primary Relational DB)**: `PostgreSQL (with PostGIS)`
    *   **저장 데이터**: 제주 공공데이터 (장소, POI 정보), 사용자 프로필, 생성된 여행 플랜, 사용자 방문 기록, 장소별 신뢰도 점수, 운영 시간, 이동 거리 등 정형 데이터.
    *   **PostGIS 활용**: 장소 간의 거리 계산, 특정 반경 내 장소 검색 등 지리 공간 데이터 처리.
*   **벡터 데이터베이스 (Vector Database)**: `ChromaDB`, `Pinecone` 또는 `Weaviate`
    *   **저장 데이터**: 공공데이터 문서(텍스트 형태)를 임베딩 모델(`Google Embeddings API`)을 통해 변환한 벡터 데이터.
    *   **활용**: RAG 엔진이 사용자 쿼리와 유사한 공공데이터를 빠르게 검색하는 데 사용됩니다.

**4. AI/ML 서비스 (AI/ML Services)**
*   **대규모 언어 모델 (LLM)**: `Google Gemini API (gpt-5-mini)`
    *   **활용**: 여행 플랜 생성, 하루방 에이전트의 개인화된 설명 생성, RAG를 통한 근거 기반 답변 생성, 사용자 피드백 분석 및 신뢰도 업데이트 로직.
*   **임베딩 모델 (Embedding Model)**: `Google Embeddings API` 또는 `Hugging Face Transformers`
    *   **활용**: 텍스트 데이터를 벡터 공간으로 변환하여 벡터 데이터베이스에 저장하고, RAG 검색의 효율성을 높입니다.

**5. 외부 서비스 연동 (External Services)**
*   **지도 및 위치 서비스**: `Google Maps Platform API` (또는 Naver/Kakao 지도 API)
    *   **활용**: 장소 검색, 지도 표시, 경로 안내, 이동 시간/거리 계산 (실행 가능성 반영에 필수적).
*   **날씨 정보 API**: 외부 날씨 API
    *   **활용**: 실시간 날씨 정보를 가져와 여행 플랜의 '실행 가능성'을 판단하고 신뢰도 점수에 반영합니다.

**6. 클라우드 인프라 (Cloud Infrastructure)**
*   **클라우드 플랫폼**: `Google Cloud Platform (GCP)`
    *   **컴퓨팅**: `Cloud Run` (서버리스 컨테이너) 또는 `App Engine` (PaaS)으로 API 서버 및 RAG 엔진 배포.
    *   **데이터베이스**: `Cloud SQL` (PostgreSQL 관리) 및 벡터 DB 호스팅 (GCP 마켓플레이스 또는 자체 배포).
    *   **AI/ML**: `Vertex AI` (LLM 및 임베딩 모델 관리, 미세 조정).
    *   **스토리지**: `Cloud Storage` (PDF 문서, 이미지 등 비정형 데이터 저장).
    *   **네트워킹**: `VPC`, `Cloud Load Balancing`, `Cloud CDN`.

**데이터 흐름:**
1.  **사용자 (프론트엔드)**: 여행 계획 요청 또는 피드백 입력.
2.  **API 서버 (백엔드)**: 요청 수신 및 처리.
3.  **RAG 엔진**: 사용자 요청을 바탕으로 벡터 DB에서 관련 임베딩 검색.
4.  **벡터 DB**: 임베딩된 공공데이터 제공.
5.  **API 서버**: 검색된 임베딩과 사용자 요청을 `Google Gemini API` (LLM)로 전송하여 여행 플랜/답변 생성.
6.  **PostgreSQL**: 생성된 플랜 저장, 사용자 방문 기록 업데이트, 신뢰도 점수 계산 및 반영.
7.  **외부 API (지도/날씨 등)**: 실시간 정보 조회하여 플랜의 실행 가능성 판단.
8.  **API 서버**: 최종 플랜 및 설명을 프론트엔드로 전달.
9.  **사용자 (프론트엔드)**: 생성된 여행 플랜 및 설명을 시각적으로 확인하고 상호작용.

### RAG (Retrieval-Augmented Generation) 샘플 데이터 구조 정의

RAG 시스템에서 LLM이 응답을 생성하기 위한 근거 자료로 활용될 수 있도록, 원본 PDF 문서의 텍스트를 작은 단위(청크)로 분할하고 각 청크에 유용한 메타데이터를 추가하는 것이 중요합니다.

여기서는 PDF에서 추출한 텍스트(`text` 변수)를 바탕으로 몇 가지 예시 청크와 메타데이터를 포함하는 Python 리스트 오브 딕셔너리(List of Dictionaries) 형태의 데이터 구조를 제안합니다.

각 청크는 `content` (실제 텍스트 내용)와 `metadata` (청크의 출처, 페이지, 섹션 등 추가 정보)를 포함하도록 구성할 수 있습니다.

### 텍스트 스플리터 (Text Splitter) 전략 제안

RAG 시스템에서 검색 품질을 높이려면 원본 문서를 적절한 크기의 '청크(Chunk)'로 나누는 것이 중요합니다. 너무 작으면 문맥이 손실되고, 너무 크면 검색 정밀도가 떨어지기 때문입니다.

`Pack Your Jeju` 제안서와 같이 구조화된 PDF 문서의 경우, 다음을 고려한 텍스트 스플리터 전략을 제안합니다.

1.  **재귀적 문자 분할 (RecursiveCharacterTextSplitter)**:
    *   가장 일반적이고 유연한 방법으로, 다양한 구분자(`\n\n`, `\n`, ` ` 등)를 사용하여 텍스트를 재귀적으로 분할합니다.
    *   큰 단위로 먼저 분할하고, 만약 청크가 너무 크면 더 작은 단위로 분할하여 문맥이 보존되도록 노력합니다.
    *   `chunk_size` (청크의 최대 크기)와 `chunk_overlap` (청크 간 중복되는 부분) 파라미터를 조절하여 적절한 청크 크기를 찾습니다.

2.  **메타데이터 포함**: 각 청크에 '원본 파일', '페이지 번호', '섹션 제목' 등의 메타데이터를 함께 저장하여 검색된 청크가 어느 부분에서 왔는지 LLM이 정확하게 참조할 수 있도록 합니다.

**적절한 `chunk_size` 및 `chunk_overlap` 선택 기준:**
*   `chunk_size`: LLM의 컨텍스트 윈도우 크기와 검색하려는 정보의 밀도에 따라 결정합니다. 제안서와 같은 기술 문서는 문맥을 유지하기 위해 500~1000자 내외가 적절할 수 있습니다.
*   `chunk_overlap`: 청크 경계에서 정보 손실을 방지하기 위해 겹치는 부분을 둡니다. 보통 `chunk_size`의 10~20% 정도를 설정합니다.

아래에서는 `LangChain` 라이브러리의 `RecursiveCharacterTextSplitter`를 사용하여 PDF에서 추출된 `text` 변수를 청크로 나누는 예시를 보여드립니다.

### 페이지 메타데이터를 포함한 텍스트 청킹

이전 단계에서 전체 텍스트를 청크로 나누었지만, 검색된 청크가 어느 페이지에서 왔는지 아는 것이 RAG 시스템의 신뢰도를 높이는 데 중요합니다.

`pypdf` 라이브러리를 사용하면 PDF를 페이지별로 읽어올 수 있습니다. 각 페이지의 텍스트를 추출하고, 해당 페이지 번호를 메타데이터로 추가하여 청크를 생성하는 방법을 보여드리겠습니다.

위 결과는 'Pack Your Jeju' 제안서에서 '제주 공공데이터'가 어떻게 활용되는지에 대한 정보를 담고 있습니다. 예를 들어, 서비스의 핵심 주제로 제주 공공데이터가 언급되며, RAG 엔진과 PostgreSQL 데이터베이스에서 어떻게 다루어지는지 설명하고 있습니다.

만약 특정 유형의 제주 공공데이터 목록이나 그 예시를 찾으신다면, 이 문서는 프로젝트의 개요를 다루므로 직접적인 목록을 제공하지는 않습니다. 다른 형태의 문서나 데이터를 제공해 주시면 해당 정보 추출을 시도할 수 있습니다.

## Pack Your Jeju 추천 기술 스택

### 1. 프론트엔드 (사용자 인터페이스)
*   **웹 (Web)**: **React.js / Next.js**
    *   **선택 이유**: 사용자 피드백 루프, 실시간 정보 표시, 복잡한 여행 플랜 UI를 구현하기에 적합한 현대적인 JavaScript 프레임워크입니다. Next.js는 SEO 및 성능 최적화에 유리합니다.
*   **모바일 (Mobile - 선택 사항)**: **React Native / Flutter**
    *   **선택 이유**: 단일 코드베이스로 iOS 및 Android 앱을 개발하여 개발 효율성을 높일 수 있습니다. (MVP 단계에서는 웹 앱에 집중하는 것이 효율적일 수 있습니다.)

### 2. 백엔드 (애플리케이션 로직 및 API)
*   **언어 및 프레임워크**: **Python (FastAPI / Flask)**
    *   **선택 이유**: AI/ML 모델(RAG, LLM)과의 연동이 가장 용이하며, 경량 API 개발에 적합합니다. 비동기 처리가 필요한 경우 FastAPI가 좋은 선택입니다.
*   **데이터 처리 및 오케스트레이션**: **LangChain / LlamaIndex** (RAG 구현 시)
    *   **선택 이유**: RAG 파이프라인 구축 및 LLM과의 상호작용을 관리하는 데 도움을 주는 프레임워크입니다.

### 3. 데이터베이스
*   **주요 데이터**: **PostgreSQL (with PostGIS extension)**
    *   **선택 이유**: 공공데이터(장소, 위치 정보), 사용자 프로필, 방문 기록, 신뢰도 점수 등 구조화된 데이터를 안정적으로 저장하고 관리합니다. PostGIS를 통해 복잡한 지리 공간 쿼리를 효율적으로 처리할 수 있습니다.
*   **벡터 데이터베이스 (RAG용)**: **ChromaDB / Pinecone / Weaviate**
    *   **선택 이유**: RAG 구현 시, 공공데이터 문서의 임베딩을 저장하고 검색하여 LLM에 전달할 관련 정보를 빠르게 찾는 데 사용됩니다.

### 4. AI/ML 모델 및 서비스
*   **대규모 언어 모델 (LLM)**: **Google Gemini API (gpt-5-mini)**
    *   **선택 이유**: 제안서에 언급된 `gpt-5-mini` 모델을 활용하여 여행 플랜 생성, 개인화된 설명 제공(하루방 에이전트), RAG 기반 답변 생성 등을 수행합니다. Google Cloud Platform과의 연동이 자연스럽습니다.
*   **임베딩 모델**: **Google Embeddings API / Hugging Face Transformers**
    *   **선택 이유**: 공공데이터 및 사용자 피드백 텍스트를 벡터 공간으로 변환하여 벡터 데이터베이스에 저장하고, RAG 검색에 활용합니다.
*   **외부 API 통합**: **Google Maps Platform API (또는 Naver/Kakao 지도 API)**
    *   **선택 이유**: 이동 거리 계산, 경로 추천, 장소 정보(운영 시간 등) 확인 등 '실행 가능성 반영' 기능을 구현하는 데 필수적입니다.

### 5. 클라우드 인프라
*   **플랫폼**: **Google Cloud Platform (GCP)**
    *   **선택 이유**: Gemini API와의 긴밀한 통합, Vertex AI를 통한 ML 모델 배포 및 관리, Cloud Run/App Engine을 통한 백엔드 배포, Cloud SQL을 통한 PostgreSQL 관리, Cloud Storage를 통한 데이터 저장 등 필요한 모든 서비스를 제공합니다.

위 결과는 'Pack Your Jeju' 제안서에서 '신뢰도 루브릭'이 언급된 모든 섹션을 보여줍니다. 이 정보는 '신뢰도 루브릭'이 프로젝트의 핵심 기술 중 하나이며, 서비스의 신뢰도 판단 및 업데이트 과정에서 중요한 역할을 함을 시사합니다.

### 신뢰도 점수 계산 로직 주요 평가 항목 제안

'Pack Your Jeju' 프로젝트의 목표와 '신뢰도 루브릭'의 역할을 고려할 때, 다음 항목들을 신뢰도 점수 계산 로직의 주요 평가 항목으로 제안합니다.

1.  **공공데이터 기반 근거 제시율**:
    *   **내용**: 추천된 여행 정보(장소, 활동 등)가 얼마나 충실하게 제주 공공데이터를 근거로 제시하는지 여부.
    *   **설명**: RAG 엔진을 통해 검색된 공공데이터의 양, 질, 그리고 추천과의 관련성.

2.  **사용자 피드백 신호 반영**:
    *   **내용**: 실제 사용자의 방문 기록(방문함, 방문하지 않음, 변경함, 정보 불일치)이 얼마나 신속하고 정확하게 점수에 반영되는지.
    *   **설명**: 특히 '정보 불일치' 피드백은 신뢰도를 크게 하락시키는 요인으로 작용.

3.  **실행 가능성 조건 적합성**:
    *   **내용**: 추천된 일정이 사용자의 현재 조건(날씨, 동행자, 이동수단, 운영시간, 이동 거리 등)에 얼마나 적합한지.
    *   **세부 항목**:
        *   **날씨 적합성**: 비, 바람 등 기상 조건에 따른 실내/실외 활동 적합도.
        *   **운영시간 일치**: 방문 예정 시간과 장소의 실제 운영 시간 일치 여부.
        *   **이동수단/거리 적합성**: 선택된 이동수단과 여행자 유형(뚜벅이, 렌터카 등)에 따른 이동 거리 및 접근성.
        *   **동행자 조건 충족**: 아이 동반, 부모님 동반 등 동행자 유형에 맞는 장소/활동의 적합도.

4.  **정보 불일치 탐지 및 경고**:
    *   **내용**: 공공데이터와 실시간 정보(예: 날씨, 지도 API) 간의 잠재적 불일치 발생 시 이를 탐지하고 사용자에게 경고하는 기능.
    *   **설명**: 운영시간, 날씨, 장소 유형 등에서 불일치가 감지될 경우 신뢰도 점수 하락 및 '확인 필요' 표시.

5.  **정보 최신성**:
    *   **내용**: 제공되는 공공데이터 및 기타 정보의 업데이트 주기 및 최신 상태 유지 여부.
    *   **설명**: 오래된 정보는 신뢰도에 부정적인 영향을 미치므로, 주기적인 데이터 업데이트가 중요.

이러한 평가 항목들을 종합하여 각 추천 장소 또는 여행 플랜에 대한 '신뢰도 점수'를 산정하고, 이를 통해 사용자에게 더욱 신뢰할 수 있는 여행 정보를 제공할 수 있을 것입니다.

### 신뢰도 점수 계산 로직 설명 및 예시 해석

위에서 정의한 `calculate_trust_score` 함수는 `trustworthiness_evaluation_items`에 기반하여 각 여행 플랜의 신뢰도 점수를 산정합니다. 각 평가 항목에 따라 미리 설정된 점수 감점 로직이 적용되며, 최종 점수는 0점 미만으로 내려가지 않도록 보정됩니다.

#### 함수의 작동 방식:
1.  **초기 점수**: 모든 플랜은 100점에서 시작합니다.
2.  **평가 항목별 감점**: `travel_plan` 딕셔너리에 포함된 정보들을 기반으로 각 평가 항목에 해당하는 조건에 따라 점수가 감점됩니다.
    *   **공공데이터 기반 근거 제시율**: 공공데이터의 품질(고/중/저)에 따라 감점.
    *   **사용자 피드백 신호 반영**: '정보 불일치'는 가장 큰 감점 요인이며, '방문하지 않음'도 감점 요인입니다. 긍정적 피드백은 감점 없음.
    *   **실행 가능성 조건 적합성**: 날씨, 운영시간, 이동수단, 동행자 조건 등이 플랜과 맞지 않을 경우 각각 감점.
    *   **정보 불일치 탐지 및 경고**: 실시간 정보에서 불일치가 탐지되면 크게 감점.
    *   **정보 최신성**: 정보가 '오래됨'으로 판단되면 감점.
3.  **상세 평가**: 점수 변화와 그 이유가 `reasons` 리스트에 기록되어 최종 점수와 함께 제공됩니다.

#### 예시 해석:

*   **제주 동백수목원 (실내 일부) - 최종 신뢰도 점수: 100점**
    *   이 플랜은 모든 평가 항목에서 최적의 조건을 만족하여 감점 없이 만점을 유지했습니다. 공공데이터 품질 '고', 긍정적 사용자 피드백, 맑은 날씨에 적합한 실내 활동, 운영시간 일치, 렌터카 및 아이 동반에 적합한 이동/동행 조건, 최신 정보 및 불일치 없음 등 모든 면에서 신뢰할 수 있는 플랜으로 평가되었습니다.

*   **성산일출봉 (야외) - 최종 신뢰도 점수: 0점**
    *   이 플랜은 여러 부정적인 요인으로 인해 신뢰도 점수가 크게 하락하여 최종 0점을 받았습니다. 낮은 공공데이터 품질, '정보 불일치' 사용자 피드백, 강풍에 부적합한 야외 장소, 운영시간 불일치, 뚜벅이/부모님 동반 부적합, 실시간 정보 불일치 탐지, 오래된 정보 등 모든 감점 요인이 적용되어 매우 신뢰하기 어려운 플랜으로 판단되었습니다.

*   **협재 해변 - 최종 신뢰도 점수: 65점**
    *   이 플랜은 긍정적인 요소와 부정적인 요소가 혼합되어 65점의 중간 점수를 받았습니다. 공공데이터 품질이 '보통'이어서 10점 감점되었고, 비 오는 날씨에 야외 해변이 부적합하여 25점이 추가 감점되었습니다. 하지만 사용자 피드백은 긍정적이었고, 운영시간, 이동수단, 동행자 조건, 정보 최신성 등은 양호했습니다. 이처럼 `calculate_trust_score` 함수는 다양한 조건에 따라 여행 플랜의 신뢰도를 세밀하게 평가할 수 있습니다.

### 신뢰도 평가 항목별 점수 상세 해석

위 표는 각 여행 플랜에 대해 개별 신뢰도 평가 항목들이 총점에 어떻게 기여했는지(감점되었는지)를 보여줍니다. 양의 값(0점)은 해당 항목에서 감점이 없었음을, 음의 값은 감점된 점수를 의미합니다.

*   **제주 동백수목원 (실내 일부)**: 모든 항목에서 0점을 기록하여 감점이 전혀 없었음을 명확히 보여줍니다. 이는 이 플랜이 모든 평가 기준을 완벽하게 충족했음을 의미합니다.

*   **협재 해변**: '공공데이터 기반 근거 제시율'에서 -10점, '날씨 적합성'에서 -25점 감점되어 총점 65점을 만들었습니다. 나머지 항목에서는 감점이 없었습니다.

*   **성산일출봉 (야외)**: 거의 모든 항목에서 감점되어 총점 0점을 기록했습니다. 특히 '정보 불일치 탐지 및 경고'에서 가장 큰 감점(-35점)이 있었고, '사용자 피드백 신호 반영'(-30점), '날씨 적합성'(-25점), '운영시간 일치'(-20점) 등 여러 부정적인 요인들이 복합적으로 작용했음을 알 수 있습니다.

이러한 상세 점수 분석은 각 여행 플랜의 강점과 약점을 파악하고, 신뢰도가 낮은 경우 어떤 항목이 주요 원인인지 심층적으로 이해하는 데 도움을 줍니다. 이제 이 데이터를 시각화하여 한눈에 비교해 보겠습니다.

### 신뢰도 점수 기반 여행 플랜 추천 결과 상세 보고서

**1. 추천 로직 개요**

본 추천 시스템은 각 여행 플랜에 대해 미리 정의된 `trustworthiness_evaluation_items`를 기반으로 신뢰도 점수(`calculate_trust_score` 함수)를 계산합니다. 이 점수는 100점 만점으로 시작하여 공공데이터 근거 제시율, 사용자 피드백, 실행 가능성 조건(날씨, 운영 시간, 이동 수단, 동행자 적합성), 정보 불일치 탐지, 정보 최신성 등의 항목에서 감점 요인이 발생할 경우 점수가 차감됩니다. 최종 신뢰도 점수는 0점 미만이 되지 않도록 보정됩니다.

이후, 계산된 신뢰도 점수를 기준으로 여행 플랜을 내림차순으로 정렬하여 사용자에게 추천합니다. 이는 가장 신뢰할 수 있는 정보를 제공하는 플랜을 우선적으로 제시함으로써 사용자가 합리적인 의사결정을 내릴 수 있도록 돕기 위함입니다.

**2. 추천 결과**

샘플 여행 플랜 3가지에 대한 신뢰도 점수 기반 추천 결과는 다음과 같습니다:

1.  **제주 동백수목원 (실내 일부)**: **100점**
2.  **협재 해변**: **65점**
3.  **성산일출봉 (야외)**: **0점**

**3. 개별 플랜 점수 해석 및 추천 이유**

*   **제주 동백수목원 (실내 일부) - 추천 순위 1위 (100점)**
    *   **점수 분석**: 이 플랜은 모든 평가 항목에서 감점 요인이 전혀 발견되지 않았습니다. 공공데이터 품질 '고', 긍정적 사용자 피드백, 맑은 날씨에 적합한 실내 활동, 운영 시간 일치, 렌터카 및 아이 동반에 적합한 이동 및 동행 조건, 최신 정보 유지 및 정보 불일치 없음 등 모든 면에서 높은 신뢰도를 보였습니다.
    *   **추천 이유**: 제시된 조건 하에 가장 완벽하고 신뢰할 수 있는 정보를 제공하므로 최우선으로 추천됩니다.

*   **협재 해변 - 추천 순위 2위 (65점)**
    *   **점수 분석**: 이 플랜은 '공공데이터 기반 근거 제시율'이 '보통'(-10점)이었으며, '날씨 적합성'에서 '비'가 오는 상황에 '야외' 활동으로 부적합하여 -25점 감점되었습니다. 하지만 사용자 피드백은 긍정적이었고, 운영 시간, 이동 수단, 동행자 조건, 정보 최신성 등 다른 항목에서는 문제가 없었습니다.
    *   **추천 이유**: 일부 감점 요인이 있지만, 전반적으로 신뢰할 만한 정보를 포함하고 있어 두 번째로 추천됩니다. 단, 날씨와 같은 외부 조건에 따라 유의할 점이 있음을 시사합니다.

*   **성산일출봉 (야외) - 추천 순위 3위 (0점)**
    *   **점수 분석**: 이 플랜은 거의 모든 항목에서 감점을 받았습니다. 낮은 공공데이터 품질(-20점), '정보 불일치' 사용자 피드백(-30점), '강풍'에 부적합한 '야외' 활동(-25점), 운영 시간 불일치(-20점), 실시간 정보 불일치 탐지(-35점), 오래된 정보(-15점) 등 복합적인 요인으로 인해 최종 점수가 0점이 되었습니다.
    *   **추천 이유**: 신뢰도가 매우 낮아 추천 순위에서 가장 아래에 위치합니다. 이 플랜은 사용자에게 상당한 불편을 초래하거나 잘못된 정보를 제공할 위험이 크므로, 이대로는 추천되기 어렵습니다.

**4. 결론**

신뢰도 점수 기반 추천 로직은 단순히 여행지를 제시하는 것을 넘어, 각 플랜의 정보 신뢰도를 정량적으로 평가하고 감점 요인을 명확히 제시함으로써 사용자가 더욱 현명하게 여행 계획을 세울 수 있도록 돕습니다. 이를 통해 'Pack Your Jeju' 프로젝트의 핵심 가치인 '신뢰할 수 있는 제주 여행 정보'를 제공하는 데 기여할 것입니다.

### 제주 여행 플랜 신뢰도 개선을 위한 제안 전략 요약

'Pack Your Jeju' 프로젝트 제안서에 따르면, 제주 여행 정보의 신뢰도를 개선하기 위한 핵심 전략은 다음과 같이 요약할 수 있습니다.

**1. 공공데이터 기반의 근거 제시 및 RAG 엔진 활용:**
*   **전략**: 제주 공공데이터를 기반으로 여행 플랜을 생성하고, AI가 추천하는 모든 장소 및 활동에 대한 **명확한 근거와 출처를 제시**합니다.
*   **기술**: `RAG (Retrieval-Augmented Generation)` 엔진을 활용하여 사용자 쿼리 및 여행 플랜 컨텍스트에 맞춰 벡터 데이터베이스에서 관련 공공데이터 문서를 검색하고, 이를 LLM에 전달하여 근거 기반 답변 및 추천을 생성합니다.
*   **목표**: 'AI가 왜 이 장소를 추천했는지 알기 어려움'이라는 기존 문제점을 해결하고, 추천의 설명 가능성을 높입니다.

**2. 사용자 피드백 루프를 통한 신뢰도 업데이트:**
*   **전략**: 사용자의 **실제 방문 기록(방문함, 방문하지 않음, 변경함, 정보 불일치)**을 '신뢰 신호'로 시스템에 반영하여 데이터의 신뢰도를 지속적으로 개선합니다.
*   **기술**: '신뢰도 루브릭'의 핵심 평가 항목에 사용자 피드백 신호를 포함시켜, 특히 '정보 불일치'와 같은 부정적 피드백은 신뢰도를 크게 하락시키는 요인으로 작용하게 합니다.
*   **목표**: '다녀온 뒤의 경험이 다음 추천에 반영되지 않음' 문제를 해결하고, 공공데이터의 빈틈을 실제 경험으로 보완하는 **선순환 구조**를 구축합니다.

**3. '하루방 에이전트'를 통한 개인화된 정보 해석 및 실행 가능성 반영:**
*   **전략**: LLM (`Google Gemini API`) 기반의 '하루방 에이전트'가 사용자의 특정 조건(날씨, 동반자, 렌터카 등)에 맞춰 공공데이터를 해석하고, 개인화된 설명을 생성합니다.
*   **기술**: 여행 플랜의 '실행 가능성 조건 적합성'을 평가 항목으로 두어, 날씨, 운영 시간, 이동 수단/거리, 동행자 조건 등이 플랜에 얼마나 적합한지를 점수에 반영합니다. 외부 날씨 API 및 지도 API 연동을 통해 실시간 정보를 활용합니다.
*   **목표**: '비·바람·운영시간·이동거리 때문에 일정 실패 가능성' 및 '공공데이터의 맥락 부족' 문제를 해결하여, 사용자에게 맥락에 맞는 신뢰도 높은 정보를 제공합니다.

**4. 정보 불일치 탐지 및 경고:**
*   **전략**: 공공데이터와 실시간 정보(날씨, 지도 API 등) 간에 잠재적인 불일치가 발생할 경우 이를 **탐지하고 사용자에게 경고**합니다.
*   **기술**: '정보 불일치 탐지 및 경고'를 신뢰도 루브릭의 주요 감점 요인으로 설정하여, 정보의 정확성을 확보합니다.
*   **목표**: 사용자에게 정확하지 않은 정보로 인한 불편을 최소화하고, '이 정보가 실제 현장과 맞는지'에 대한 불확실성을 줄입니다.

**5. 정보 최신성 유지:**
*   **전략**: 제공되는 공공데이터 및 기타 정보의 **업데이트 주기와 최신 상태를 유지**하는 것을 중요한 신뢰도 요소로 관리합니다.
*   **기술**: '정보 최신성'을 신뢰도 루브릭의 감점 요인으로 두어, 오래된 정보가 추천 신뢰도를 하락시키는 것을 방지합니다.
*   **목표**: 항상 최신화된 정보를 제공하여 정보의 유효성을 보장합니다.

이러한 전략들은 각 여행 플랜의 신뢰도를 정량적으로 평가하고 개선함으로써, 'Pack Your Jeju' 프로젝트가 궁극적으로 **신뢰할 수 있는 제주 여행 정보의 순환형 생태계**를 구축하는 데 기여할 것입니다.

### 신뢰도 점수 향상을 위한 시뮬레이션 시나리오

다음은 `Pack Your Jeju` 프로젝트에서 각 신뢰도 평가 항목에 대한 개선 전략이 여행 플랜의 신뢰도 점수에 어떻게 긍정적인 영향을 미칠 수 있는지 보여주는 시뮬레이션 시나리오입니다. 기존의 `성산일출봉 (야외)` (저신뢰도 플랜)와 `협재 해변` (중간 신뢰도 플랜)을 대상으로 개선 효과를 측정합니다.

#### 시나리오 1: '성산일출봉 (야외)'의 정보 최신성 및 불일치 해결

*   **목표**: '정보 불일치 탐지 및 경고'와 '정보 최신성' 문제를 해결하여 신뢰도를 향상시킵니다.
*   **가정**: 시스템이 실시간 정보를 업데이트하고(정보 최신성 개선), 감지된 불일치 문제를 해결했다고 가정합니다.
*   **기대 효과**: 최종 점수가 크게 상승할 것입니다.

#### 시나리오 2: '협재 해변'의 공공데이터 품질 개선 및 날씨 조건 반영

*   **목표**: '공공데이터 기반 근거 제시율'을 높이고, 날씨 조건에 대한 적합성을 확보하여 신뢰도를 향상시킵니다.
*   **가정**: 공공데이터의 품질이 '고'로 향상되고, 방문 시 날씨가 '맑음'으로 변경되거나, 시스템이 날씨에 적합한 대안을 제시하여 감점 요인이 해소되었다고 가정합니다.
*   **기대 효과**: 점수가 크게 상승하여 '만점'에 가까워질 것입니다.

#### 시나리오 3: '성산일출봉 (야외)'에 대한 부정적 사용자 피드백 해소 및 긍정 피드백 누적

*   **목표**: 부정적인 사용자 피드백('정보 불일치')이 해소되고, 긍정적인 피드백이 누적될 경우의 신뢰도 변화를 시뮬레이션합니다.
*   **가정**: 시나리오 1에서 정보 최신성 및 불일치 문제가 해결된 후, 사용자들이 해당 장소에 대해 긍정적인 피드백을 제공하기 시작했다고 가정합니다.
*   **기대 효과**: 사용자 피드백 관련 감점 요인이 사라지면서 점수가 추가로 향상될 것입니다.

### **Pack Your Jeju: 신뢰도 향상을 위한 핵심 전략 제안**

'Pack Your Jeju' 프로젝트는 제주 여행 정보의 신뢰도를 높여 사용자에게 더욱 만족스러운 경험을 제공하는 것을 목표로 합니다. 이를 위해 다음과 같은 핵심 전략들을 제안하며, 각 전략은 신뢰도 점수 향상에 직접적으로 기여합니다.

#### **1. 공공데이터 기반 근거 제시 및 RAG 엔진 활용 강화**
*   **전략**: 모든 추천 콘텐츠에 대해 제주 공공데이터를 명확한 근거와 출처로 제시하고, RAG(Retrieval-Augmented Generation) 엔진을 통해 관련성 높은 정보를 검색하여 LLM의 응답 품질을 높입니다.
*   **기대 효과**: AI 추천의 '설명 가능성'을 극대화하여 사용자의 신뢰를 확보하고, 정보의 투명성을 제공합니다.

#### **2. 사용자 피드백 루프를 통한 신뢰도 순환 시스템 구축**
*   **전략**: 사용자의 실제 방문 기록(방문함, 방문하지 않음, 변경함, 정보 불일치)을 '신뢰 신호'로 적극 반영하여 시스템 내 데이터의 신뢰도를 지속적으로 업데이트합니다. 특히 '정보 불일치'와 같은 부정적 피드백은 중요한 감점 요인으로 작용하여 정보 개선을 유도합니다.
*   **기대 효과**: 사용자의 경험이 다음 여행자에게 도움이 되는 선순환 구조를 만들고, 공공데이터의 잠재적 빈틈을 실제 경험 데이터로 보완합니다.

#### **3. '하루방 에이전트'의 개인화된 정보 해석 및 실행 가능성 반영**
*   **전략**: LLM 기반의 '하루방 에이전트'가 사용자의 특정 조건(날씨, 동반자, 이동수단 등)에 맞춰 공공데이터를 해석하고, 개인화된 설명을 제공합니다. 실시간 날씨 및 지도 API 연동을 통해 여행 플랜의 '실행 가능성'을 신뢰도 점수에 반영합니다.
*   **기대 효과**: '날씨, 운영시간, 이동거리 등으로 인한 일정 실패'와 '공공데이터의 맥락 부족' 문제를 해결하여, 사용자 맞춤형의 실현 가능한 정보를 제공합니다.

#### **4. 정보 불일치 탐지 및 선제적 경고 시스템 운영**
*   **전략**: 공공데이터와 실시간 정보(예: 외부 API) 간의 잠재적 불일치를 지속적으로 탐지하고, 사용자에게 '확인 필요' 등의 경고 메시지를 제공합니다.
*   **기대 효과**: 정보의 정확성을 보장하고, 사용자에게 불확실한 정보를 강요하지 않음으로써 서비스의 신뢰도를 높입니다.

#### **5. 정보 최신성 유지 관리**
*   **전략**: 제공되는 공공데이터 및 모든 정보의 업데이트 주기를 관리하고, 항상 최신 상태를 유지하도록 노력합니다. 오래된 정보는 신뢰도 점수에 부정적인 영향을 미치도록 설계합니다.
*   **기대 효과**: 정보의 유효성을 지속적으로 보장하여, 사용자가 시의적절하고 정확한 정보를 얻을 수 있도록 합니다.


### **신뢰도 향상 시뮬레이션 결과 요약**

이러한 전략들이 실제로 여행 플랜의 신뢰도 점수에 미치는 영향을 확인하기 위해 세 가지 시뮬레이션을 수행했습니다. 아래 결과는 전략 적용 전후의 점수 변화를 명확히 보여줍니다.

1.  **시나리오 1: '성산일출봉 (야외)'의 정보 최신성 및 불일치 해결**
    *   **이전 점수**: 0점
    *   **개선 후 점수**: 5점
    *   **효과**: 정보 불일치 탐지 및 경고(-35점)와 정보 최신성(-15점) 문제를 해결함으로써 점수가 크게 상승하는 것을 확인했습니다. 여전히 낮은 점수이나, 핵심 감점 요인 해소의 중요성을 보여줍니다.

2.  **시나리오 2: '협재 해변'의 공공데이터 품질 개선 및 날씨 조건 반영**
    *   **이전 점수**: 65점
    *   **개선 후 점수**: 100점
    *   **효과**: 공공데이터 기반 근거 제시율(-10점)과 날씨 적합성(-25점)을 개선하자 만점으로 점프하는 것을 확인했습니다. 이는 데이터 품질 개선 및 조건 적합성 판단의 중요성을 시사합니다.

3.  **시나리오 3: '성산일출봉 (야외)'에 대한 부정적 사용자 피드백 해소 및 긍정 피드백 누적**
    *   **이전 점수 (시나리오 1 개선 후)**: 5점
    *   **개선 후 점수**: 35점
    *   **효과**: 사용자 피드백 신호 반영(-30점)이 긍정적으로 전환되면서 점수가 5점에서 35점으로 추가 상승했습니다. 이는 사용자 피드백이 신뢰도 점수에 미치는 강력한 영향을 보여줍니다.


### **결론**

위 전략들과 시뮬레이션 결과는 'Pack Your Jeju' 프로젝트가 단순히 여행지를 추천하는 것을 넘어, **정보의 신뢰도를 핵심 가치로 삼아 지속적으로 개선하고 증명할 수 있는 시스템**임을 보여줍니다. 각 감점 요인을 명확히 식별하고 해결함으로써, 사용자가 제주 여행을 계획하고 실행하는 데 있어 최고의 신뢰도를 제공할 수 있을 것입니다.

### **신뢰도 점수 향상 전략에 따른 정량적 기대 효과**

'Pack Your Jeju' 프로젝트의 신뢰도 향상 전략은 다음과 같은 정량적인 점수 변화를 통해 그 효과를 입증할 수 있습니다. 각 시나리오별 개선 전후 점수와 주요 효과를 요약했습니다.

이러한 시뮬레이션 결과는 각 핵심 전략이 여행 플랜의 신뢰도 점수를 얼마나 효과적으로 끌어올릴 수 있는지 보여주며, 'Pack Your Jeju' 프로젝트가 제공하는 정보의 신뢰도를 구체적인 수치로 증명할 수 있음을 시사합니다.

### '성산일출봉 (야외)' 및 '협재 해변' 여행 플랜 신뢰도 개선 종합 보고서

'Pack Your Jeju' 프로젝트의 핵심 목표는 여행 정보의 신뢰도를 극대화하는 것입니다. 이 목표를 달성하기 위해 '성산일출봉 (야외)'이라는 초기 신뢰도 0점 플랜과 '협재 해변'이라는 중간 신뢰도 65점 플랜을 대상으로 일련의 시뮬레이션을 수행하여, 체계적인 문제 해결 전략이 신뢰도 점수에 미치는 긍정적인 영향을 정량적으로 입증했습니다.

#### **1. '성산일출봉 (야외)' 신뢰도 개선 여정 (0점 -> 100점)**

'성산일출봉 (야외)' 플랜은 초기 상태에서 **0점**이라는 매우 낮은 신뢰도를 기록했습니다. 이는 공공데이터 품질 부족, '정보 불일치' 사용자 피드백, 강풍에 부적합한 야외 활동, 운영시간 불일치, 실시간 정보 불일치 탐지, 오래된 정보 등 거의 모든 평가 항목에서 감점 요인이 존재했기 때문입니다. 하지만 다음 단계별 시뮬레이션을 통해 모든 문제를 해결하고 최종적으로 100점 만점을 달성했습니다.

*   **시나리오 1: 정보 최신성 및 불일치 해결 (+5점, 0점 -> 5점)**
    *   가장 큰 감점 요인이었던 '정보 불일치 탐지 및 경고'(-35점)와 '정보 최신성'(-15점) 문제가 해결되자, 신뢰도 점수는 5점으로 상승했습니다. 이는 즉각적인 정보의 정확성과 최신성 유지가 신뢰도 향상의 첫걸음임을 보여줍니다.

*   **시나리오 3: 부정적 사용자 피드백 해소 및 긍정 피드백 누적 (+30점, 5점 -> 35점)**
    *   '정보 불일치'(-30점) 사용자 피드백이 긍정적인('방문함', '긍정') 피드백으로 전환되면서, 점수는 5점에서 35점으로 크게 상승했습니다. 사용자 피드백 루프를 통한 신뢰도 순환 시스템의 중요성을 강조하는 결과입니다.

*   **시나리오 4: 공공데이터 품질 개선 (+20점, 35점 -> 55점)**
    *   '공공데이터 기반 근거 제시율'이 '저'(-20점)에서 '고'로 개선되자 점수는 55점으로 향상되었습니다. 이는 제공되는 공공데이터 자체의 품질 확보가 서비스의 근본적인 신뢰도를 결정하는 요소임을 보여줍니다.

*   **시나리오 5: 날씨 적합성 해결 (+25점, 55점 -> 80점)**
    *   강풍에 부적합했던 야외 활동이 '맑음' 날씨 조건으로 변경되면서 '날씨 적합성'(-25점) 감점이 해소되었고, 점수는 80점에 도달했습니다. '하루방 에이전트'와 같은 개인화된 조건 해석 및 실행 가능성 반영의 중요성을 입증합니다.

*   **시나리오 6: 운영시간 일치 해결 (+20점, 80점 -> 100점)**
    *   마지막 남은 감점 요인이었던 '운영시간 불일치'(-20점) 문제가 해결되면서, '성산일출봉 (야외)' 플랜은 **100점 만점**을 달성했습니다. 이는 모든 잠재적 불확실성 요소를 제거하여 사용자에게 완벽하게 신뢰할 수 있는 정보를 제공할 수 있음을 의미합니다.

#### **2. '협재 해변' 신뢰도 개선 여정 (65점 -> 100점)**

'협재 해변' 플랜은 초기 **65점**으로 시작했으며, '공공데이터 기반 근거 제시율'이 '보통'(-10점)이었고, 비 오는 날씨에 야외 활동으로 부적합하여 '날씨 적합성'(-25점)에서 감점이 있었습니다.

*   **시나리오 2: 공공데이터 품질 개선 및 날씨 조건 반영 (+35점, 65점 -> 100점)**
    *   공공데이터 품질을 '고'로 향상시키고, 날씨 조건을 '맑음'으로 가정하여 감점 요인을 해소하자, '협재 해변' 플랜 역시 **100점 만점**을 기록했습니다. 이는 중간 수준의 신뢰도를 가진 플랜도 핵심 감점 요인을 해결함으로써 최상의 신뢰도를 얻을 수 있음을 보여줍니다.

#### **3. 종합 결론**

이러한 다단계 시뮬레이션 결과는 'Pack Your Jeju' 프로젝트의 신뢰도 향상 전략들이 매우 효과적임을 정량적으로 입증합니다. 초기 0점이었던 플랜이 모든 감점 요인을 체계적으로 해결함으로써 100점 만점을 달성하고, 65점 플랜 역시 핵심 문제를 해결하여 100점을 얻는 과정을 통해, 우리는 다음을 확인할 수 있습니다.

*   **시스템적 문제 해결의 중요성**: 각 신뢰도 평가 항목에 대한 명확한 진단과 해결책 적용은 점진적이고 궁극적인 신뢰도 향상을 가져옵니다.
*   **사용자 피드백의 강력한 영향**: '정보 불일치'와 같은 사용자 피드백은 가장 큰 감점 요인 중 하나이며, 이를 긍정적으로 전환하는 것이 신뢰도 순환 구조의 핵심입니다.
*   **데이터 품질 및 실시간 반영**: 공공데이터의 품질, 정보 최신성, 그리고 날씨와 운영 시간 등 실시간 환경 요인의 반영은 플랜의 실행 가능성과 신뢰도를 결정하는 필수적인 요소입니다.

'Pack Your Jeju'는 단순히 여행 코스를 추천하는 것을 넘어, 정보의 신뢰성을 보장하고 지속적으로 개선하는 체계를 갖춤으로써 사용자에게 가장 신뢰할 수 있는 제주 여행 경험을 제공할 것입니다.

### Pack Your Jeju: 신뢰도 향상 시뮬레이션 최종 보고서

'Pack Your Jeju' 프로젝트는 제주 여행 정보의 신뢰도를 체계적으로 향상시키기 위한 전략들을 시뮬레이션을 통해 성공적으로 검증했습니다. 특히, 초기 신뢰도 점수가 0점에 불과했던 '성산일출봉 (야외)' 플랜이 단계적인 개선을 통해 100점 만점에 도달하는 과정을 입증했으며, '협재 해변' 플랜 역시 단일 시나리오에서 100점으로 향상되었습니다.

#### 1. '성산일출봉 (야외)' 플랜 신뢰도 개선 과정 (0점 → 100점)

'성산일출봉 (야외)' 플랜은 초기 상태에서 '공공데이터 기반 근거 제시율 낮음', '정보 불일치 사용자 피드백', '강풍에 부적합한 야외 장소', '운영시간 불일치', '실시간 정보 불일치 탐지', '정보 최신성 오래됨' 등 여러 감점 요인으로 인해 0점이라는 낮은 신뢰도를 기록했습니다. 하지만 다음의 단계적인 시뮬레이션을 통해 모든 감점 요인을 해소하고 만점을 달성했습니다.

*   **시나리오 1: 정보 최신성 및 불일치 해결 (+5점, 0점 → 5점)**
    *   **개선**: '실시간 정보 불일치 탐지됨'(-35점) 및 '정보 최신성 오래됨'(-15점) 문제를 해결하여, 총 50점의 잠재적 감점 중 50점을 회복했습니다. (최종 점수 5점은 다른 감점 요인이 남아있었기 때문).
    *   **의미**: 정보의 정확성과 최신성 유지가 신뢰도 향상의 첫걸음임을 보여줍니다.

*   **시나리오 3: 부정적 사용자 피드백 해소 및 긍정 피드백 누적 (+30점, 5점 → 35점)**
    *   **개선**: '정보 불일치' 사용자 피드백(-30점)이 '긍정적 피드백'으로 전환되면서 점수가 30점 상승했습니다.
    *   **의미**: 사용자 피드백 루프가 신뢰도에 미치는 강력한 긍정적 영향을 입증했습니다.

*   **시나리오 4: 공공데이터 품질 개선 (+20점, 35점 → 55점)**
    *   **개선**: '공공데이터 기반 근거 제시율'이 '저'(-20점)에서 '고'(0점)로 개선되면서 20점 상승했습니다.
    *   **의미**: 공공데이터의 품질 확보가 서비스의 근본적인 신뢰도를 높이는 데 필수적임을 확인했습니다.

*   **시나리오 5: 날씨 적합성 해결 (+25점, 55점 → 80점)**
    *   **개선**: '강풍에 부적합한 야외 장소'(-25점) 문제가 '맑은 날씨에 적합한 야외 장소'(0점)로 해결되면서 25점 상승했습니다.
    *   **의미**: 실시간 환경 조건(날씨)을 반영한 적합성 판단이 사용자 경험과 신뢰도에 중요하게 작용합니다.

*   **시나리오 6: 운영시간 일치 해결 (+20점, 80점 → 100점)**
    *   **개선**: '운영시간 불일치'(-20점) 문제가 해결되어 만점을 달성했습니다.
    *   **의미**: 사용자에게 정확한 운영시간 정보를 제공하는 것이 플랜의 실행 가능성과 신뢰도를 완성하는 요소임을 보여줍니다.

이로써 '성산일출봉 (야외)' 플랜은 체계적인 개선 전략 적용을 통해 초기 0점에서 최종 100점 만점으로 신뢰도를 완벽하게 회복할 수 있음을 증명했습니다.

#### 2. '협재 해변' 플랜 신뢰도 개선 결과 (65점 → 100점)

'협재 해변' 플랜은 초기 65점에서 시작하여 '공공데이터 기반 근거 제시율'이 '보통'(-10점)이고 '비' 오는 날씨에 '야외' 활동으로 부적합(-25점)하다는 감점 요인이 있었습니다.

*   **시나리오 2: 공공데이터 품질 개선 및 날씨 조건 반영 (+35점, 65점 → 100점)**
    *   **개선**: 공공데이터 품질을 '고'로 개선하고 날씨 조건을 '맑음'으로 가정하여 두 가지 주요 감점 요인(-10점, -25점)을 동시에 해결함으로써 최종 100점 만점을 달성했습니다.
    *   **의미**: 비교적 적은 수의 핵심 감점 요인이라도 효과적으로 해결하면 신뢰도를 크게 향상시킬 수 있음을 보여줍니다.

#### 3. 결론

이러한 시뮬레이션 결과는 'Pack Your Jeju' 프로젝트가 제안하는 신뢰도 루브릭 및 개선 전략이 여행 플랜의 신뢰도를 정량적으로 평가하고 실질적으로 향상시킬 수 있는 강력한 프레임워크임을 명확히 보여줍니다. 각 감점 요인을 정확히 식별하고 대응함으로써, 사용자에게 '설렘부터 근거 있는' 제주 여행 정보를 제공하고, 궁극적으로는 제주 여행 정보의 신뢰도 높은 선순환 생태계를 구축하는 데 기여할 것입니다.

### 10. 데모1: 지역신뢰대시보드

**핵심 메시지:** 첫 화면은 단순히 많은 여행지를 추천하는 것이 아니라, **지역별로 사용자가 무엇을 얼마나 믿고 선택할 수 있는지**를 보여주는 데 중점을 둡니다.

**문제:**
*   데이터가 부족한 지역임에도 불구하고 억지로 추천 목록을 채우는 경향이 있습니다.
*   사용자는 많은 정보 속에서 어떤 정보를 신뢰해야 할지 판단하기 어렵습니다.

**해결 방안:**
*   **지역별 신뢰 대시보드**: 제주 지도 기반으로 사용자가 지역을 선택하면, 해당 지역에서 확인 가능한 공공데이터 후보 수, 특정 순간에 담을 수 있는 후보 수, 추천 가능한 활동 유형을 함께 표시합니다.
*   **신뢰도 기반 정보 분리**: 확인 가능한 후보와 확인이 필요한 후보를 명확하게 분리하여 보여줍니다.
*   **의사결정 단위로 재구성**: 공공데이터를 단순 출력하는 것이 아니라, 사용자가 '무엇을 믿고 선택할 수 있는지'를 판단할 수 있는 의사결정 단위로 정보를 재구성합니다.

**주요 내용:**
*   사용자는 지역을 선택하고, 그 안에서 **어떤 선택을 얼마나 신뢰할 수 있는지**를 우선적으로 파악합니다.
*   데이터가 부족한 조합에 대해 억지로 추천하지 않고, 투명하게 정보를 제공하여 사용자의 합리적인 판단을 돕습니다.
*   이 화면은 여행지 리스트의 양보다는 **정보의 신뢰도**에 초점을 맞추어 사용자 경험을 개선합니다.

### 11. 데모2: 후보 카드와 여행팩

**핵심 메시지:** 사용자는 후보를 여행 계획에 추가하기 전에, **해당 정보의 신뢰 근거와 확인 필요 사항**을 우선적으로 확인합니다.

**문제:**
*   일반적인 여행 서비스는 단순히 시각적으로 매력적인 사진과 추천 문구로 후보를 제시합니다.
*   사용자는 비슷한 후보들(예: 오름 종류) 사이에서 단순히 이름이나 사진만으로 선택하기 어렵습니다.
*   정보의 신뢰성, 최신성, 개인 조건과의 적합성에 대한 판단 근거가 부족합니다.

**해결 방안:**
*   **순서 변경**: 예쁜 사진과 추천 문구보다 **신뢰도 정보**를 먼저 보여줍니다.
*   **후보 카드 상세 정보**: 각 후보 카드에 **신뢰도 점수, 공공데이터 출처, 수정 요청 이력, 이용 및 주차 확인 여부, 확인 필요 항목**을 함께 표시합니다.
*   **사용자 판단 지원**: 사용자는 이러한 근거 정보를 바탕으로 '이 장소가 현재 내 조건에 맞는지', '확인해야 할 위험은 무엇인지', '어떤 근거로 신뢰도가 산정되었는지'를 종합적으로 판단하여 선택합니다.
*   **여행팩 (실행 가능한 계획)**: 선택된 후보는 단순 북마크가 아닌, **근거를 확인한 후보만 저장되는 '실행 가능한 계획'인 여행팩**에 담깁니다.

**주요 내용:**
*   사용자는 **정보의 신뢰성**을 최우선으로 고려하여 여행 후보를 선택합니다.
*   여행팩에 담긴 계획은 단순한 저장 목록이 아니라, **향후 방문 기록을 통해 피드백 대시보드로 이어지는 신뢰 순환 구조**의 시작점이 됩니다.
*   이는 사용자가 **수동적인 정보 소비자가 아닌, 정보의 신뢰도를 함께 만들어가는 참여자**로 거듭나게 합니다.

### 12. 데모3: 방문 기록과 신뢰도 업데이트

**핵심 메시지:** 사용자의 방문 기록은 즉시 신뢰도에 반영되지 않고, **내부 검증 과정을 거쳐 신뢰도 업데이트 및 공공데이터 개선 신호로 연결**됩니다.

**문제:**
*   사용자가 남긴 방문 기록(예: 방문 완료, 정보 다름, 만족/불만족, 변경 사항)을 즉시 공공데이터나 신뢰도에 반영할 경우, 악의적인 입력이나 단순 착오로 인해 데이터 품질이 저하될 수 있습니다.

**해결 방안:**
*   **내부 피드백 대시보드 운영**: 사용자의 방문 기록은 먼저 내부 피드백 대시보드에서 **'검증 완료', '검증 필요', '반영 제외'**로 분류됩니다.
*   **다각적 검증**: 분류된 기록은 기존 공공데이터, 사진, 위치 정보 등과 대조하여 **정확성을 검증**합니다.
*   **검증된 신호만 반영**: **검증이 완료된 신호에 한해서만** 신뢰도 점수와 공공데이터 반영 큐(queue)로 연결됩니다.

**주요 내용:**
*   사용자 참여를 적극적으로 받되, 모든 입력을 맹신하지 않고 **철저한 검증 과정을 필수적으로 거칩니다.**
*   이러한 구조를 통해 서비스는 사용할수록 **더욱 믿을 수 있는 판단 구조**를 갖게 됩니다.
*   결론적으로, 이 시스템은 사용자 참여를 통해 **지속적으로 신뢰도를 향상시키는 선순환 메커니즘**을 구현합니다.

### 13. 하루방 에이전트

**핵심 메시지:** 하루방 에이전트는 무엇이든 답하는 **챗봇이 아니라, 명확한 근거 안에서만 정보를 조율하고 설명하는 '여행 판단 에이전트'**입니다.

**문제:**
*   기존 챗봇들은 자유롭게 상상하여 답변을 생성할 수 있어 정보의 신뢰도가 떨어질 수 있습니다.
*   사용자는 단순히 답변을 얻는 것뿐만 아니라, 그 답변이 '왜' 나왔는지에 대한 근거와 신뢰할 수 있는 판단을 필요로 합니다.

**해결 방안:**
*   **근거 기반 답변:** 사용자의 질문(예: "한림에서 혼자 힐링 여행 가능한가요?")에 대해, 하루방은 **공공데이터 후보, 후보 카드, 날씨, 이동 조건, 수정 요청 이력, 방문 피드백** 등 확인된 근거 데이터들을 먼저 조회합니다.
*   **조율 및 설명:** 조회된 근거 안에서만 답변하며, 좋은 점, 보완점, 이미 플랜에 포함된 활동, 비어있는 활동 등을 설명합니다.
*   **불확실성 명시:** 모르는 내용은 그럴듯하게 채우지 않고 **'확인 필요'**로 명확히 분리하여, 사용자가 정보의 한계를 인지할 수 있도록 합니다.

**주요 내용:**
*   하루방 에이전트의 역할은 단순히 많은 답을 제공하는 것이 아니라, 사용자의 **특정 조건과 서비스가 가진 신뢰할 수 있는 근거를 조합하여, 믿고 실행할 수 있는 여행 판단을 돕는 것**입니다.
*   이를 통해 사용자는 정보의 양보다 **정보의 품질과 신뢰도**에 기반한 의사결정을 내릴 수 있게 됩니다.

### 14. 시스템 구조: RAG + Trust Engine + gpt-5-mini

**핵심 메시지:** 이 시스템은 LLM이 먼저 답변을 생성하는 구조가 아니라, **데이터와 규칙(Trust Engine)이 먼저 판단하고, LLM(gpt-5-mini)은 마지막에 이를 설명하는 구조**입니다.

**문제:**
*   기존 LLM 중심의 시스템은 생성 모델이 모든 것을 결정하여 답변의 일관성과 검증 가능성이 떨어질 수 있습니다.
*   작은 모델(gpt-5-mini)만으로는 신뢰도 높고 근거 있는 답변을 생성하기 어렵습니다.

**해결 방안:**
*   **사용자 입력 구조화**: 사용자 입력이 들어오면 바로 생성 모델로 보내지 않고, 지역, 동행자, 목적, 순간, 특별 요청 등 **조건으로 구조화**합니다.
*   **RAG (Retrieval-Augmented Generation) 작동**: 구조화된 입력을 바탕으로 RAG가 공공데이터와 API에서 **후보 장소, 출처, 근거 문서, 이동 및 주차 정보, 수정 요청 이력, 방문 피드백 신호**를 조회합니다.
*   **Trust Engine 판단**: Trust Engine은 조회된 정보를 기반으로 **신뢰도 점수, 확인 필요 여부, 배제 조건**을 계산하여 후보를 걸러냅니다.
*   **gpt-5-mini 역할 제한**: 마지막으로 gpt-5-mini는 새로운 후보를 만들어내는 역할이 아니라, **이미 조회되고 계산된 근거를 사용자에게 이해하기 쉬운 말로 설명하고, 여행팩 방향을 조율**하는 역할을 수행합니다.

**주요 내용:**
*   시스템의 핵심은 **LLM이 모든 것을 결정하지 않도록 하는 것**입니다. 추천 후보는 데이터가 찾고, 신뢰도는 Trust Engine이 계산하며, 모델은 이 **근거 안에서만 설명**합니다.
*   이러한 구조를 통해 같은 작은 모델(gpt-5-mini)을 사용하더라도 답변의 **일관성과 검증 가능성을 높일 수** 있습니다.

### 시스템 구조 보완: RAG + Trust Engine + gpt-5-mini 상세 설명

`Pack Your Jeju`의 핵심 강점은 `gpt-5-mini`와 같은 LLM의 한계를 극복하고 신뢰도를 높이기 위한 구조적 접근 방식에 있습니다. 단순히 LLM을 활용하는 것을 넘어, **데이터와 규칙 기반의 판단 엔진이 선행**되고, LLM은 그 결과를 **사용자 친화적으로 설명하는 보조적인 역할**을 수행합니다.

*   **RAG (Retrieval-Augmented Generation)의 역할 확장**:
    RAG는 사용자 쿼리 및 여행 플랜 컨텍스트를 분석하여 단순한 정보 검색을 넘어, `후보 장소, 출처, 근거 문서, 이동 및 주차 정보, 수정 요청 이력, 방문 피드백 신호` 등 **여행 판단에 필요한 다각적인 근거 데이터**를 `공공데이터`와 다양한 `외부 API` (날씨, 지도 등)로부터 조회합니다. 이는 LLM이 '상상'이 아닌 '팩트'에 기반하여 작동하도록 하는 1차적인 안전장치입니다.

*   **Trust Engine (신뢰 판단 엔진)의 핵심 기능**:
    RAG를 통해 조회된 방대한 정보를 기반으로, Trust Engine은 `신뢰도 점수, 확인 필요 여부, 배제 조건`을 **정교하게 계산**합니다. 이 과정에서 `공공데이터의 품질, 정보 최신성, 사용자 피드백(긍정/부정), 실시간 날씨 및 운영시간과의 적합성, 정보 불일치 여부` 등 다양한 신뢰도 루브릭 평가 항목들이 종합적으로 고려됩니다. Trust Engine은 **'가도 될 만한 장소'를 선별하고, '가서는 안 될 장소'를 미리 걸러내는 의사결정의 핵심**입니다.

*   **gpt-5-mini의 제한적이고 전략적인 활용**:
    Trust Engine이 1차적으로 필터링하고 신뢰도를 평가한 `근거 데이터`와 `후보 플랜`만이 `gpt-5-mini`에 전달됩니다. `gpt-5-mini`는 **새로운 정보를 만들어내거나 추천을 임의로 생성하지 않습니다.** 대신, 이미 조회되고 계산된 근거를 바탕으로 사용자에게 `이해하기 쉬운 말로 추천의 이유와 제한 사항을 설명`하고, `여행팩의 방향을 조율`하는 역할을 담당합니다. 특히, Trust Engine이 '확인 필요'로 분류한 정보에 대해서는 LLM 또한 이를 명확히 인지하고 사용자에게 불확실성을 솔직하게 전달하도록 설계되어 있습니다.

이러한 **RAG → Trust Engine → gpt-5-mini**의 순차적이고 상호 보완적인 구조를 통해, `Pack Your Jeju`는 같은 `gpt-5-mini` 모델을 사용하더라도 **답변의 일관성, 정확성, 검증 가능성을 획기적으로 높여, 궁극적으로 사용자가 믿고 실행할 수 있는 여행 정보를 제공**합니다.

```markdown
## 🌟 'Trust Engine' 구조도: 신뢰 기반 여행 플랜의 핵심 🌟

### **문제 인식:** LLM의 환각 위험 ⚠️ & 정보 과부하

---

### **핵심 솔루션: 구조적 신뢰 시스템 구축**

```mermaid
graph TD
    A[사용자 쿼리 및 조건] --> B{RAG 엔진: 근거 데이터 조회}
    B --> C{Trust Engine: 신뢰도 판단 및 필터링}
    C --> D{gpt-5-mini: 검증된 근거 기반 설명/조율}
    D --> E[사용자에게 신뢰 가능한 여행 플랜 제공]
```

---

### **단계별 역할 설명:**

#### **1. RAG 엔진: '팩트' 수집의 최전선**
*   **입력**: 사용자 쿼리, 여행 플랜 컨텍스트
*   **역할**: 공공데이터, 외부 API(날씨, 지도)에서 `후보 장소`, `출처`, `근거 문서`, `이동/주차 정보`, `수정 요청 이력`, `방문 피드백 신호` 등 **다각적인 근거 데이터**를 조회합니다.
*   **주요 메타데이터 확보**: `public_data_quality`, `last_verified_date`

#### **2. Trust Engine: '신뢰'를 계산하고 '정직함'을 판단**
*   **입력**: RAG가 조회한 방대한 근거 데이터
*   **역할**: 찾아온 정보를 기반으로 `신뢰도 점수`, `'확인 필요' 여부`, `배제 조건`을 **정교하게 계산하고 필터링**합니다.
*   **다차원 신뢰도 평가 지표**:
    *   **`public_data_quality`**: 공공데이터 품질 수준 (최상/상/중/하) 평가
    *   **`last_verified_date`**: 정보의 최신성 검증 (오래된 정보 감점)
    *   **`feedback_summary`**: 사용자 피드백(긍정/부정, 정보 불일치) 분석 및 반영
    *   **`suitability_tags`**: 실시간 조건(날씨, 운영시간, 동반자, 이동수단) 적합성 평가
    *   **`discrepancy_status`**: 정보 불일치 여부 탐지 및 경고
*   **결정**: '가도 될 만한 곳' 선별, '가서는 안 될 곳' 원천 차단. `확인 필요` 정보 명확히 분류.

#### **3. gpt-5-mini: '설명'하고 '조율'하는 책임감 있는 AI**
*   **입력**: Trust Engine에서 **최종 검증 및 필터링된 근거 데이터**
*   **역할**: **새로운 정보를 생성하거나 임의로 추천하지 않습니다.** 오직 검증된 근거를 바탕으로 사용자에게 `이해하기 쉬운 말로 설명`하고, `여행 계획 방향을 조율`합니다.
*   **가드레일**: 근거 범위를 벗어나는 답변 시도 시 즉시 차단하고, 불확실성은 '확인 필요'로 투명하게 고지합니다.

---

### **결론:**
`RAG → Trust Engine → gpt-5-mini`의 **순차적, 상호 보완적 구조**는 LLM의 한계를 극복하고 **답변의 일관성, 정확성, 검증 가능성**을 획기적으로 높여, 사용자가 **'믿고 실행할 수 있는 여행 판단'**을 할 수 있도록 돕습니다.
```

### 15. 기본 GPT 와의 차이 및 신뢰성 검증

**핵심 메시지:** 같은 `gpt-5-mini` 모델이라도 **공공데이터, RAG, Trust Engine, 피드백 검증 구조가 결합**되면 신뢰할 수 있는 여행 답변이 됩니다.

**문제 (기본 GPT의 약점):**
*   그럴듯한 답변을 빠르게 만들 수 있으나, 지역 여행 정보의 **운영 시간이나 주소 변경을 실시간으로 보장하기 어렵습니다.**
*   추천의 **출처가 명확하게 분리되어 보이지 않아 신뢰도가 낮습니다.**
*   사용자 조건이나 방문 피드백이 다음 추천 구조로 이어지기 어려워 **개인화 및 개선이 어렵습니다.**

**해결 방안 (Pack Your Jeju의 차별점):**
*   **후보 선 조회**: 답변을 생성하기 전에 후보를 먼저 조회합니다.
*   **근거 기반 확보**: RAG(Retrieval-Augmented Generation)를 통해 공공데이터에서 근거를 찾습니다.
*   **Trust Engine 통한 점수화**: Trust Engine이 공식성, 출처 확인, 사용자 조건 적합성, 이동 및 주차, 날씨 적합성, 피드백 신호 등을 **구체적인 평가 항목으로 나누어 점수화**합니다.
*   **설명 가능한 신뢰도**: 최종 신뢰도 점수는 단순한 숫자가 아니라, **각 평가 항목별 점수를 통해 왜 그런 점수가 나왔는지 설명 가능한 구조**로 제공됩니다.
*   **제한된 LLM 역할**: `gpt-5-mini`는 조회되고 계산된 근거 안에서만 답변을 설명하고 조율하는 역할을 수행합니다. 부족한 정보는 분리하고, 검증된 신호만 답변에 반영합니다.

**주요 내용:**
*   `Pack Your Jeju`의 차이는 **모델 자체가 아닌, 모델을 둘러싼 구조**에 있습니다.
*   이 시스템은 근거를 먼저 확인하고, 정보 부족 시 판단을 보류하며, 검증된 신호만을 답변에 반영하여 **같은 LLM을 사용하더라도 답변의 신뢰성과 검증 가능성을 획기적으로 높입니다.**

### 16. 서비스 견고성·보안·마무리

**핵심 메시지:** `제주를 담다`는 **억지 추천을 지양하고, 확인 필요를 분리하며, 검증된 피드백을 통해 지속적으로 개선되는 서비스**입니다.

**문제:**
*   외부 API나 데이터 부족 시 서비스 흐름이 끊기거나, 부정확한 정보로 억지 추천이 이루어질 수 있습니다.
*   보안 취약점(API 키 노출, 민감 정보 처리 등)이 발생할 수 있습니다.

**해결 방안:**
*   **견고성 설계**: 지도, 날씨 등 외부 호출이 실패하더라도 **대체 UI와 안내 문구로 흐름을 유지**합니다. 정보가 부족한 후보는 억지로 추천하지 않고 **'확인 필요'로 명확히 분리**하여 보여줍니다.
*   **보안 관리**: API 키는 **환경변수와 Secrets로 관리**하고 코드에 직접 노출하지 않습니다. 사용자 입력은 구조화하여 안전하게 반영하고, 방문 피드백에서 민감 정보는 최소화하여 **내부 신호로만 처리**합니다.

**주요 내용:**
*   이 서비스는 '더 많이 추천하는 서비스'가 아닌, **'모르는 것은 확인 필요로 남기고, 확인된 정보와 검증된 피드백으로 여행자가 더 믿고 선택할 수 있게 돕는 서비스'**를 지향합니다.
*   **공공데이터**는 시작점이며, **사용자의 실제 방문 기록**이 다음 여행자를 위한 업데이트 신호가 되어 `제주를 담다`를 **공공데이터로 계획하고, 방문 기록으로 다시 좋아지는 순환형 AI 여행 플랜 서비스**로 만듭니다.

### 마지막 요약

**심사위원에게 남길 최종 메시지 3개:**
1.  **제주 여행 정보의 문제는 정보 부족이 아니라, 믿고 실행할 판단 기준의 부족입니다.**
2.  **`제주를 담다`는 공공데이터, RAG, Trust Engine, 방문 피드백을 연결한 순환형 AI 여행 플랜 서비스입니다.**
3.  **`gpt-5-mini`를 크게 쓰는 것이 아니라, 근거 안에서만 설명하게 만들어 신뢰도 분석이라는 주제를 서비스 경험으로 풀었습니다.**

**Q&A에서 강조할 답변 포인트 5개:**
1.  **신뢰도는 단순 별점이 아니라 공식성, 최신성, 출처, 조건 적합성, 이동성, 피드백 신호를 분리해 설명하는 구조입니다.**
2.  **사용자 피드백은 바로 반영하지 않고, 내부 대시보드에서 분류·검증한 뒤 검증된 신호만 반영합니다.**
3.  **RAG는 GPT가 모르는 지역 정보를 아는 척하지 않도록, 수집 DB와 근거 문서를 먼저 조회하게 하는 장치입니다.**
4.  **데이터가 부족하면 억지 추천하지 않고 확인 필요로 표시합니다. 이 정직함이 신뢰도 설계의 핵심입니다.**
5.  **확장성은 제주 관광지에서 숙소, 음식점, 교통, 축제, 지역 상권 데이터로 넓힐 수 있고, 방문 기록 기반 업데이트 구조는 다른 지역에도 적용 가능합니다.**

**최종 한 문장:**
`제주를 담다`는 더 많이 추천하는 서비스가 아니라, **공공데이터로 계획하고 방문 기록으로 다시 좋아지는 신뢰 기반 여행 서비스**입니다.

## 발표 클로징 멘트 (구어체로 다듬기)

---

"저희 '제주를 담다'는 단순히 여행 정보를 나열하는 서비스를 넘어섭니다.

제주 여행의 본질적인 문제, 바로 '정보의 신뢰성 부족'을 저희는 제대로 봤습니다. 그리고 이 문제를 해결하기 위해 **공공데이터, RAG, Trust Engine, 그리고 사용자 피드백**이라는 아주 튼튼한 기술적 순환 시스템을 만들었어요.

저희는 GPT-5-mini 같은 AI 모델이 가진 한계를 누구보다 잘 알고 있습니다. 그래서 저희는 **'완벽한 답'보다는 '검증된 근거를 기반으로 투명하게 설명'**하는 쪽으로 방향을 잡았고, 사용자들이 **'믿고 실행할 수 있는 판단 기준'**을 제공하려고 합니다.

이제 제주 여행자들은 더 이상 정보의 홍수 속에서 헤매지 않고, **'설렘부터 근거 있게'** 여행을 계획하고 실제로 떠날 수 있게 될 겁니다.

'제주를 담다'는 앞으로도 계속 발전하면서, 제주를 넘어 모든 여행 정보의 신뢰도를 높이는 데 크게 기여할 거예요.

저희 '제주를 담다'에 귀 기울여 주셔서 정말 감사합니다. 함께 더 신뢰할 수 있는 여행의 미래를 만들어가고 싶습니다!"

---

## 시뮬레이션 결과 주요 인사이트 및 향후 개선 방안 요약

'Pack Your Jeju' 프로젝트의 신뢰도 향상 시뮬레이션은 제주 여행 정보의 신뢰도를 정량적으로 평가하고 개선할 수 있음을 명확하게 입증했습니다. 초기 신뢰도가 매우 낮았던 플랜('성산일출봉 (야외)', 0점)과 중간 수준의 플랜('협재 해변', 65점) 모두 체계적인 전략 적용을 통해 최종 100점 만점에 도달했습니다.

### 주요 인사이트:

1.  **데이터 품질 및 최신성의 근본적 중요성**: '정보 불일치 탐지 및 경고' 및 '정보 최신성' 문제 해결은 신뢰도 향상의 첫걸음이며, '공공데이터 기반 근거 제시율' 개선은 근본적인 정보 품질을 높이는 핵심 요소입니다. 시뮬레이션에서 이러한 문제 해결을 통해 점수가 크게 상승하는 것을 확인했습니다.

2.  **사용자 피드백의 강력한 영향력**: '정보 불일치'와 같은 부정적 사용자 피드백을 긍정적인 피드백으로 전환하는 것은 신뢰도 점수를 극적으로 향상시키는 강력한 동인이었습니다. 이는 사용자 경험이 신뢰도 순환 구조의 핵심임을 보여줍니다.

3.  **실시간 환경 요소 반영의 필수성**: '날씨 적합성'과 '운영시간 일치'와 같은 실행 가능성 조건의 개선은 여행 플랜의 실용성과 신뢰도를 크게 높였습니다. 이는 외부 API 연동을 통한 실시간 정보 반영의 중요성을 강조합니다.

4.  **체계적인 문제 해결의 효과**: 각 감점 요인을 정확하게 식별하고, 이에 대한 개선 전략을 순차적으로 적용함으로써 신뢰도 0점 플랜도 완벽한 신뢰도를 달성할 수 있음을 증명했습니다.

### 향후 개선 방안:

이러한 인사이트를 바탕으로 'Pack Your Jeju' 프로젝트의 신뢰도 높은 서비스 제공을 위해 다음과 같은 개선 방안을 제안합니다.

1.  **지속적인 공공데이터 확보 및 RAG 엔진 고도화**: 최신 제주 공공데이터를 지속적으로 확보하고, RAG 엔진의 검색 및 응답 생성 품질을 고도화하여 AI 추천의 근거 제시율과 정확성을 더욱 높여야 합니다.

2.  **사용자 피드백 시스템 강화**: 사용자가 쉽게 피드백을 제공하고, 그 피드백이 신뢰도 점수에 반영되는 과정을 투명하게 보여주는 UI/UX를 개발합니다. 특히 '정보 불일치' 피드백을 신속하게 처리하고, 긍정적 피드백을 유도할 수 있는 기능을 강화하여 신뢰도 순환 시스템을 활성화합니다.

3.  **'하루방 에이전트'의 실시간 정보 연동 확장**: 날씨, 운영 시간, 교통 정보 등 실시간으로 변하는 외부 데이터를 지도 및 기타 API와 연동하여 '하루방 에이전트'가 제공하는 정보의 정확성과 최신성을 극대화합니다. 이를 통해 사용자 조건에 더욱 적합한 개인화된 플랜을 제공합니다.

4.  **정보 불일치 탐지 및 경고 시스템 정교화**: 공공데이터와 실시간 정보 간의 불일치를 예측하고, 사용자에게 사전 경고하는 시스템의 정확도를 높여 잠재적 불편을 최소화합니다.

5.  **정보 최신성 관리 자동화**: 모든 여행 정보의 업데이트 주기를 시스템적으로 관리하고, 오래된 정보에 대한 자동 갱신 또는 경고 로직을 구현하여 정보의 유효성을 상시 보장합니다.

이러한 개선 방안들을 통해 'Pack Your Jeju'는 사용자에게 단순히 제주 여행 코스를 넘어, '설렘부터 근거 있게' 여행을 계획하고 실행할 수 있도록 돕는 **가장 신뢰할 수 있는 제주 여행 정보 플랫폼**으로 발전할 것입니다.

### 시뮬레이션 결과: 신뢰도 향상 효과 요약

'Pack Your Jeju' 프로젝트의 신뢰도 향상 전략들이 각 여행 플랜에 미치는 영향을 정량적으로 분석한 결과입니다. 각 시나리오별 초기 점수 대비 개선 후 점수 변화를 통해 전략의 효과를 확인할 수 있습니다.

| 시나리오                                    | 대상 플랜            | 이전 점수 | 개선 후 점수 | 점수 변화 | 주요 개선 효과                                                                         |
| :------------------------------------------ | :------------------- | :-------- | :----------- | :-------- | :------------------------------------------------------------------------------------- |
| **Initial State**                           | 성산일출봉 (야외)    | 0점       | 0점          | 0점       | Baseline for 성산일출봉 (야외)                                                       |
| **1. 정보 최신성 및 불일치 해결**           | 성산일출봉 (야외)    | 0점       | 5점          | +5점      | 정보 불일치 탐지 및 경고(-35점)와 정보 최신성(-15점) 문제 해결                       |
| **3. 부정적 피드백 해소 및 긍정 피드백 누적** | 성산일출봉 (야외)    | 5점       | 35점         | +30점     | 사용자 피드백 신호 반영(-30점)이 긍정적으로 전환 (시나리오 1 개선 후 기준)         |
| **4. 공공데이터 품질 개선**                 | 성산일출봉 (야외)    | 35점      | 55점         | +20점     | 공공데이터 기반 근거 제시율(-20점) 개선                                              |
| **5. 날씨 적합성 해결**                     | 성산일출봉 (야외)    | 55점      | 80점         | +25점     | 날씨 적합성(-25점) 개선 (강풍/야외 -> 맑음/야외)                                     |
| **6. 운영시간 일치 해결**                   | 성산일출봉 (야외)    | 80점      | 100점        | +20점     | 운영시간 일치(-20점) 해결                                                            |
| **Initial State**                           | 협재 해변            | 65점      | 65점         | 0점       | Baseline for 협재 해변                                                               |
| **2. 공공데이터 품질 개선 및 날씨 조건 반영** | 협재 해변            | 65점      | 100점        | +35점     | 공공데이터 기반 근거 제시율(-10점) 및 날씨 적합성(-25점) 개선                         |

## 발표 전략 다듬기: 핵심 내용과 RAG 메타데이터 가중치 결합

'제주를 담다' 발표의 핵심은 **정보의 신뢰성**과 이를 **기술적으로 어떻게 구현하고 검증하는지**를 명확하게 전달하는 것입니다. 발표 스크립트의 주요 메시지와 RAG 메타데이터 가중치 제안을 결합하여 더욱 설득력 있는 발표 전략을 수립할 수 있습니다.

### 1. 문제 정의 강화: '믿을 수 있는 판단 기준'의 부재

*   **발표 포인트**: 제주 여행 정보의 문제는 정보의 양이 아니라, '지금, 나에게, 이 장소가 믿을 수 있는가?'에 대한 판단 기준의 부재임을 강조합니다.
*   **RAG 연결**: RAG 시스템이 단순히 정보를 검색하는 것을 넘어, **`public_data_quality`** 메타데이터를 통해 원본 데이터의 신뢰성을 먼저 확인하고, **`last_verified_date`**를 통해 최신성을 보장함으로써, 이 '판단 기준'을 제공하는 서비스임을 어필합니다.

### 2. 서비스 핵심: '신뢰 순환'과 '책임감 있는 AI' 강조

*   **발표 포인트**: '제주를 담다'는 공공데이터, RAG, Trust Engine, 방문 피드백이 연결된 **순환형 AI 여행 플랜 서비스**임을 명확히 합니다. 특히 gpt-5-mini가 '모든 것을 생성'하는 것이 아니라, **'근거 안에서 설명'**하도록 설계된 **책임감 있는 AI**임을 강조해야 합니다.
*   **RAG 연결**: Trust Engine이 **`discrepancy_status`** 메타데이터를 통해 정보 불일치 여부를 판단하고, **`suitability_tags`**를 통해 사용자 조건에 맞춰 적합성을 조율하는 과정을 설명합니다. LLM은 이러한 검증된 메타데이터를 기반으로 답변하며, 이는 '억지 추천'이 아닌 '신뢰 기반 조율'임을 보여줍니다.

### 3. 사용자 피드백의 중요성: '데이터 오염 방지'와 '신뢰 신호' 확보

*   **발표 포인트**: 사용자의 방문 기록이 단순한 리뷰가 아니라, 내부 검증을 거쳐 다음 여행자를 위한 **'신뢰 신호'**로 재활용되는 선순환 구조의 핵심임을 강조합니다. 대량의 허위 피드백에 대한 방어 전략도 언급합니다.
*   **RAG 연결**: **`feedback_summary`** 메타데이터가 RAG 검색의 가장 높은 가중치를 갖는 이유를 설명하며, 사용자 피드백이 서비스의 신뢰도 점수에 결정적인 영향을 미치는 과정을 기술적으로 뒷받침합니다. 이는 데이터 오염을 방지하고 신뢰할 수 있는 피드백만 반영하는 시스템의 견고함을 보여줍니다.

### 4. 시뮬레이션 결과: 정량적 증명으로 신뢰도 입증

*   **발표 포인트**: 초기 0점이었던 플랜이 단계적인 개선 전략(정보 최신성, 피드백, 데이터 품질, 날씨 적합성, 운영시간 일치 해결)을 통해 100점에 도달하는 과정을 **정량적인 시뮬레이션 결과**로 제시하여, 서비스의 신뢰도 향상 메커니즘이 실제로 작동함을 입증합니다.
*   **RAG 연결**: 각 시뮬레이션 시나리오에서 해결된 문제들(정보 불일치, 최신성, 공공데이터 품질, 날씨/운영시간 적합성, 피드백)이 곧 RAG 메타데이터의 핵심 필드들과 직결되어 있음을 보여줍니다. 즉, RAG 메타데이터의 강화가 이 시뮬레이션 결과를 가능하게 하는 **핵심 기술적 토대**임을 강조합니다.

### 5. 핵심 질문 및 방어 전략

*   **GPT-5-mini의 한계**: RAG가 **`public_data_quality`**, **`last_verified_date`**, **`discrepancy_status`** 등의 메타데이터로 근거를 확보하고, Trust Engine이 판단하며, LLM은 그 안에서 설명하는 구조임을 재차 강조합니다.
*   **사용자 피드백 검증**: **`feedback_summary`** 메타데이터를 통한 신뢰도 가중치 부여, 다단계 검증 프로세스, 이상 탐지 메커니즘 등이 허위 피드백에 대한 견고한 방어 장치임을 설명합니다.
*   **MVP 범위**: 현재 구현된 RAG 및 Trust Engine의 핵심 기능을 설명하고, 메타데이터 강화 로드맵(예: `suitability_tags` 확장)이 서비스 확장성 계획과 어떻게 연결되는지 제시합니다.

이러한 방식으로 발표 스크립트의 서사적 흐름에 RAG 메타데이터의 기술적 중요성을 유기적으로 연결한다면, 심사위원들에게 '제주를 담다'가 단순한 아이디어를 넘어 **견고한 기술적 기반 위에서 신뢰를 구축하는 서비스**임을 강력하게 어필할 수 있을 것입니다.

## 심사위원 압박 질문 대비 방어 스크립트

'제주를 담다'의 핵심 가치인 '신뢰성'을 심사위원들에게 명확하게 각인시키기 위해, 예상되는 압박 질문에 대한 구체적이고 논리적인 방어 스크립트를 준비했습니다. 기술적인 깊이와 서비스 철학을 결합하여 강력한 인상을 남길 수 있도록 구성했습니다.

### 1. GPT-5-mini의 환각 현상, 기술적으로 어떻게 완벽 방어하고 있으며 비상 대응 메커니즘은?

"저희는 GPT-5-mini의 본질적 한계를 누구보다 깊이 이해하고 있으며, 이를 완벽히 '방어'한다기보다는 **'근거 없는 답변 생성을 원천 차단하고, 모든 정보의 신뢰성을 기술적으로 검증하는 구조'**를 핵심 전략으로 삼았습니다. 이는 단순히 가능성을 최소화하는 것을 넘어, **공고한 아키텍처를 통해 환각 현상 자체가 발생할 여지를 차단**하는 데 주력합니다.

1.  **'생성'이 아닌 '검증 기반 설명' 역할의 하드코딩**: GPT-5-mini는 새로운 정보를 창작하거나 추천을 임의로 생성하도록 설계되지 않았습니다. 오직 **Trust Engine에서 '사전에 엄격히 검증되고 필터링된 근거 데이터'**만을 입력받아 사용자에게 '설명'하고 '조율'하는 역할을 수행하도록 **코드 레벨에서 그 역할이 명확히 제한**되어 있습니다. RAG 시스템은 `public_data_quality`, `last_verified_date`, `discrepancy_status` 등의 핵심 메타데이터를 활용해 신뢰성 높은 근거를 선제적으로 확보합니다.
2.  **RAG-Trust Engine 다단계 필터링 아키텍처**: 사용자의 쿼리는 즉시 LLM으로 전달되지 않고, **'하드 테크니컬 배리어(Hard Technical Barrier)'** 역할을 하는 **RAG 엔진**을 통해 신뢰할 수 있는 공공데이터와 외부 API 근거를 먼저 조회합니다. 이후 **Trust Engine이 이 조회된 근거들의 신뢰도를 다각도로 판단하고 필터링**합니다. LLM은 오직 이 '필터링되고 신뢰성이 입증된 근거'만을 최종 입력으로 받아 답변을 구성합니다. 근거가 부족하면 답변을 생성하지 않습니다.

    특히, Trust Engine은 RAG를 통해 조회된 방대한 근거 데이터를 기반으로, **'어떤 정보가 사용자에게 신뢰할 수 있고, 실행 가능한가'**를 판단하는 핵심 기술입니다. 이는 단순히 정보를 걸러내는 것을 넘어, **다차원적인 신뢰도 평가 지표들을 종합적으로 계산하여 후보 장소들을 정교하게 필터링**합니다. 주요 기술적 로직은 다음과 같습니다:

    *   **다차원 신뢰도 루브릭 기반 점수 산정**: Trust Engine은 각 후보 장소 및 플랜에 대해 100점 만점의 **'신뢰도 점수(Trust Score)'**를 산정합니다. 이 점수는 다음의 핵심 평가 항목들을 기반으로 계산되며, 각 항목별로 미리 정의된 감점 로직이 적용됩니다.
        *   **`public_data_quality` (공공데이터 품질)**: 참조하는 공공데이터의 품질 수준(최상, 상, 중, 하)에 따라 신뢰도에 가중치를 부여하거나 감점합니다. 품질이 낮은 데이터는 필터링 우선순위가 낮아집니다.
        *   **`last_verified_date` (최종 검증일)**: 정보의 최신성을 평가하여, 오래된 정보일수록 신뢰도 점수를 감점합니다. 최신 정보가 우선적으로 사용자에게 제공되도록 합니다.
        *   **`feedback_summary` (사용자 피드백 요약)**: 사용자 방문 기록(긍정/부정 경험, 정보 불일치 등)을 분석하여 신뢰도에 크게 반영합니다. 특히 '정보 불일치' 피드백은 가장 큰 감점 요인으로 작용하며, 악의적인 허위 피드백은 패턴 분석 및 가중치 조절을 통해 반영을 최소화합니다.
        *   **`suitability_tags` (적합성 태그)**: 실시간 날씨, 운영 시간, 동반자 유형, 이동 수단 등 사용자의 현재 조건에 대한 적합성을 평가합니다. 조건에 부적합한 장소는 신뢰도 점수를 감점하며, '확인 필요' 상태로 분류될 수 있습니다.
        *   **`discrepancy_status` (불일치 상태)**: 공공데이터와 외부 API(지도, 날씨 등) 또는 사용자 피드백 간의 정보 불일치 여부를 탐지합니다. 불일치가 감지되면 신뢰도를 크게 감점하고, LLM이 '확인 필요'로 사용자에게 투명하게 고지하도록 지시합니다.

    *   **'확인 필요 여부' 및 '배제 조건' 결정**: 특정 임계점 이하의 신뢰도 점수이거나, `discrepancy_status`가 '불일치 감지'와 같은 중요한 플래그를 포함하는 경우, 해당 정보를 **'확인 필요'**로 분류하여 LLM이 사용자에게 불확실성을 명확히 전달하도록 지시합니다. 또한, 사용자의 특정 조건(예: '폐업', '영업시간 아님')에 명백히 맞지 않거나, 신뢰도 점수가 극도로 낮은 후보는 아예 **'배제 조건'**으로 분류하여 LLM에게 전달되지 않도록 원천 차단합니다.

3.  **코드 레벨 가드레일 및 투명한 고지 프로토콜**: 만약 LLM이 할당된 근거 범위를 벗어나는 답변을 시도할 경우, 시스템 내부에 구현된 **'가드레일(Guardrails)'이 즉시 해당 답변 생성을 차단**합니다. 사용자에게는 '현재 정보로는 정확한 답변 제공이 어렵습니다'와 같은 **명확하고 중립적인 메시지를 제공**합니다. Trust Engine이 `discrepancy_status` 메타데이터를 통해 '확인 필요'로 분류한 정보는 LLM 또한 이를 답변에 명확히 명시하여 정보의 불확실성을 투명하게 전달하는 프로토콜을 따릅니다.

저희는 **Ragas 평가 지표**를 통한 지속적인 모니터링을 넘어, 이러한 **구조적, 코드 레벨의 방어 메커니즘**을 통해 LLM의 한계를 가장 견고하게 관리하고 있으며, 이는 **'제주를 담다'의 핵심 기술적 신뢰성**을 담보하는 핵심 요소입니다."

### 2. 사용자 피드백 검증의 효율성과 공정성, 대량의 허위 피드백 발생 시 견고성은?

"사용자 피드백은 저희 서비스의 '신뢰 순환'에서 가장 중요한 축이지만, 동시에 데이터 오염의 위험이 가장 큰 영역임을 명확히 인지하고 있습니다. 저희는 **다단계 검증 프로세스**와 **이상 탐지 메커니즘**을 통해 효율성, 공정성, 그리고 허위 피드백에 대한 견고성을 확보합니다.

1.  **내부 피드백 대시보드와 교차 검증**: 모든 피드백은 '검증 완료', '검증 필요', '반영 제외'로 분류되는 내부 대시보드를 거칩니다. '검증 필요' 피드백은 AI 기반 자동 검증 시스템과 필요시 운영자의 수동 검토를 통해 **사진, GPS 위치 정보, 시간 스탬프** 등 다른 데이터 소스와 교차 대조하여 진위를 판단합니다. 특히 `discrepancy_status` 메타데이터와 피드백을 비교하여 정보 불일치 여부를 정밀하게 확인합니다.
2.  **허위 피드백 방어 시스템**:
    *   **패턴 분석 및 자동 차단**: 비정상적인 패턴의 피드백이 감지될 경우, 해당 계정이나 피드백 소스를 **자동으로 '반영 제외' 및 경고 조치**합니다.
    *   **가중치 차등 부여**: 피드백 제공자의 **활동 이력과 신뢰도**에 따라 피드백에 부여하는 가중치를 차등화합니다. 악의적 의도의 피드백은 RAG 검색 시 `feedback_summary` 메타데이터가 미치는 영향이 최소화되도록 설계되어 있습니다.

이러한 체계적인 과정을 통해 저희는 RAG 검색 시 `feedback_summary` 메타데이터가 가장 높은 가중치를 가질 수 있도록, **가장 신뢰할 수 있는 사용자 피드백만을 시스템에 반영하며 데이터 오염을 적극적으로 방어**합니다."

### 3. MVP 구현 범위와 고도화 로드맵은?

"저희 'Pack Your Jeju'는 **'핵심 신뢰도 순환 구조'의 MVP 구현**에 집중했습니다. 현재 MVP 단계에서 실질적으로 작동하는 범위는 다음과 같습니다.

1.  **공공데이터 기반 RAG 엔진**: 비짓제주 관광지 데이터 등을 기반으로 `public_data_quality`, `last_verified_date` 메타데이터를 활용하여 LLM에 근거를 제공하는 RAG 엔진이 구현되어 있습니다.
2.  **Trust Engine (핵심 로직)**: 공공데이터 품질, 최신성, 사용자 피드백 신호(`feedback_summary`) 등을 반영하여 신뢰도를 계산하고 '확인 필요' 상태(`discrepancy_status`)를 분류하는 **기본 로직**이 구현되었습니다.
3.  **하루방 에이전트 (근거 기반 설명)**: Trust Engine에서 필터링된 정보를 바탕으로 사용자에게 '설명'하는 `gpt-5-mini` 기반의 기능이 구현되어, 근거 안에서만 조율하고 모르는 것은 '확인 필요'로 명시합니다.
4.  **사용자 피드백 수집 및 내부 대시보드 (수동 검증)**: 사용자의 방문 기록을 수집하고, 내부 대시보드를 통해 운영자가 수동으로 검토할 수 있는 초기 검증 프로세스가 작동합니다.

**저희의 고도화 로드맵은 다음과 같습니다.**

*   **단기 (MVP 이후 3개월)**:
    *   **Trust Engine 고도화**: `suitability_tags` (동반자, 이동수단 등) 및 `discrepancy_status` (불일치 유형별)를 신뢰도 점수 산정에 정교하게 반영하고, 가중치를 동적으로 조절하는 시스템을 구축할 것입니다.
    *   **자동 검증 시스템 도입**: 머신러닝 기반 모듈을 개발하여 허위 피드백 탐지 정확도를 높일 계획입니다.
*   **중장기 (MVP 이후 6개월 이상)**:
    *   **데이터 확장**: 숙소, 음식점, 교통 등 다른 카테고리의 공공데이터를 통합하고, **실시간 데이터 연동을 강화**하여 '실행 가능성'을 더욱 정교화할 것입니다.
    *   **타 지역 확장**: 제주도 모델을 기반으로 전국 단위 서비스 확장을 목표하고 있습니다.

저희는 MVP에서 **신뢰 순환 구조를 기술적으로 증명**하는 데 집중했으며, 이 로드맵을 통해 서비스의 기능과 신뢰도를 지속적으로 고도화하여 **가장 믿을 수 있는 AI 여행 플랜 서비스**로 발전시켜 나갈 것입니다."

## Q&A: 기술적 신뢰도를 강조할 수 있는 유려한 답변 멘트

'제주를 담다'의 기술적 신뢰성을 효과적으로 전달하기 위한 답변 멘트들을 다듬었습니다. 질문의 핵심을 꿰뚫는 동시에, 저희 시스템의 구조적 강점을 명확하게 부각하는 데 초점을 맞췄습니다.

### 1. GPT-5-mini의 환각 현상, 기술적으로 어떻게 완벽 방어하고 있으며 비상 대응 메커니즘은?

"저희는 GPT-5-mini의 본질적 한계를 누구보다 깊이 이해하고 있습니다. 저희는 **'환각 현상을 완벽히 방어한다'기보다, '발생 가능성을 최소화하고 불확실성을 사용자에게 투명하게 전달'하는 구조**를 핵심 전략으로 삼고 있습니다.

1.  **'생성'이 아닌 '검증 기반 설명'**: GPT-5-mini는 새로운 정보를 창작하거나 추천을 임의로 생성하지 않습니다. 오직 **Trust Engine에서 엄격히 검증된 근거 데이터**를 바탕으로 사용자에게 '설명'하고 '조율'하는 역할을 수행합니다. RAG 시스템이 `public_data_quality`, `last_verified_date` 등의 메타데이터로 신뢰성 높은 근거를 선제적으로 확보합니다.
2.  **RAG-Trust Engine 다단계 필터링**: 사용자의 쿼리는 즉시 LLM으로 전달되지 않습니다. **RAG 엔진이 신뢰할 수 있는 공공데이터와 외부 API 근거를 먼저 조회하고, Trust Engine이 이 근거들의 신뢰도를 판단하고 필터링**합니다. LLM은 오직 이 필터링된, 신뢰성이 입증된 근거만을 입력받아 답변합니다.
3.  **가드레일과 투명한 고지**: 만약 LLM이 근거 범위를 벗어나는 답변을 시도할 경우, 내부 **가드레일이 즉시 차단**하고, 사용자에게는 '현재 정보로는 정확한 답변이 어렵습니다'와 같은 **중립적인 메시지를 제공**합니다. Trust Engine이 '확인 필요'로 분류한 정보는 LLM도 이를 명확히 명시하여 불확실성을 투명하게 알립니다.

저희는 **Ragas 평가 지표**를 통해 환각 가능성을 지속적으로 모니터링하며 모델의 동작을 재조정하고 있습니다. 이러한 구조적 접근이야말로 LLM의 한계를 가장 견고하게 관리하는 방법입니다."

### 2. 사용자 피드백 검증의 효율성과 공정성, 대량의 허위 피드백 발생 시 견고성은?

"사용자 피드백은 저희 서비스의 '신뢰 순환'에 필수적이지만, 동시에 데이터 오염의 위험이 가장 큰 영역임을 명확히 인지하고 있습니다. 저희는 **다단계 검증 프로세스**와 **이상 탐지 메커니즘**을 통해 효율성, 공정성, 그리고 허위 피드백에 대한 견고성을 확보합니다.

1.  **내부 피드백 대시보드와 교차 검증**: 모든 피드백은 '검증 완료', '검증 필요', '반영 제외'로 분류되는 내부 대시보드를 거칩니다. '검증 필요' 피드백은 AI 기반 자동 검증 시스템과 필요시 운영자의 수동 검토를 통해 **사진, GPS 위치 정보, 시간 스탬프** 등 다른 데이터 소스와 교차 대조하여 진위를 판단합니다. 특히 `discrepancy_status` 메타데이터와 피드백을 비교하여 정보 불일치 여부를 정밀하게 확인합니다.
2.  **허위 피드백 방어 시스템**:
    *   **패턴 분석 및 자동 차단**: 비정상적인 패턴의 피드백이 감지될 경우, 해당 계정이나 피드백 소스를 **자동으로 '반영 제외' 및 경고 조치**합니다.
    *   **가중치 차등 부여**: 피드백 제공자의 **활동 이력과 신뢰도**에 따라 피드백에 부여하는 가중치를 차등화합니다. 악의적 의도의 피드백은 RAG 검색 시 `feedback_summary` 메타데이터가 미치는 영향이 최소화되도록 설계되어 있습니다.

이러한 체계적인 과정을 통해 저희는 RAG 검색 시 `feedback_summary` 메타데이터가 가장 높은 가중치를 가질 수 있도록, **가장 신뢰할 수 있는 사용자 피드백만을 시스템에 반영하며 데이터 오염을 적극적으로 방어**합니다."

### 3. MVP 구현 범위와 고도화 로드맵은?

"저희 'Pack Your Jeju'는 **'핵심 신뢰도 순환 구조'의 MVP 구현**에 집중했습니다. 현재 MVP 단계에서 실질적으로 작동하는 범위는 다음과 같습니다.

1.  **공공데이터 기반 RAG 엔진**: 비짓제주 관광지 데이터 등을 기반으로 `public_data_quality`, `last_verified_date` 메타데이터를 활용하여 LLM에 근거를 제공하는 RAG 엔진이 구현되어 있습니다.
2.  **Trust Engine (핵심 로직)**: 공공데이터 품질, 최신성, 사용자 피드백 신호(`feedback_summary`) 등을 반영하여 신뢰도를 계산하고 '확인 필요' 상태(`discrepancy_status`)를 분류하는 **기본 로직**이 구현되었습니다.
3.  **하루방 에이전트 (근거 기반 설명)**: Trust Engine에서 필터링된 정보를 바탕으로 사용자에게 '설명'하는 `gpt-5-mini` 기반의 기능이 구현되어, 근거 안에서만 조율하고 모르는 것은 '확인 필요'로 명시합니다.
4.  **사용자 피드백 수집 및 내부 대시보드 (수동 검증)**: 사용자의 방문 기록을 수집하고, 내부 대시보드를 통해 운영자가 수동으로 검토할 수 있는 초기 검증 프로세스가 작동합니다.

**저희의 고도화 로드맵은 다음과 같습니다.**

*   **단기 (MVP 이후 3개월)**:
    *   **Trust Engine 고도화**: `suitability_tags` (동반자, 이동수단 등) 및 `discrepancy_status` (불일치 유형별)를 신뢰도 점수 산정에 정교하게 반영하고, 가중치를 동적으로 조절하는 시스템을 구축할 것입니다.
    *   **자동 검증 시스템 도입**: 머신러닝 기반 모듈을 개발하여 허위 피드백 탐지 정확도를 높일 계획입니다.
*   **중장기 (MVP 이후 6개월 이상)**:
    *   **데이터 확장**: 숙소, 음식점, 교통 등 다른 카테고리의 공공데이터를 통합하고, **실시간 데이터 연동을 강화**하여 '실행 가능성'을 더욱 정교화할 것입니다.
    *   **타 지역 확장**: 제주도 모델을 기반으로 전국 단위 서비스 확장을 목표하고 있습니다.

저희는 MVP에서 **신뢰 순환 구조를 기술적으로 증명**하는 데 집중했으며, 이 로드맵을 통해 서비스의 기능과 신뢰도를 지속적으로 고도화하여 **가장 믿을 수 있는 AI 여행 플랜 서비스**로 발전시켜 나갈 것입니다."

## [발표 레이아웃 제안 - Q&A 섹션]

**슬라이드 제목: Q&A - 심사위원 질문 & 답변**

### 레이아웃 근거:
*   **명확성 및 집중**: 각 질문과 해당 답변은 명확하게 제시되어 심사위원들이 논의를 쉽게 따라갈 수 있도록 합니다.
*   **가독성**: 복잡한 답변에는 명확한 제목과 글머리 기호를 사용합니다.
*   **Trust Engine 강조**: 답변이 'Trust Engine'의 핵심 원칙 및 기술 구현과 일관되게 연결되도록 합니다.

### 제안된 내용 흐름:

---

**슬라이드 1: Q&A - 일반 질문**

*   **전환 슬라이드**: 주요 메시지를 간략하게 다시 강조하거나 청중에게 질문을 요청합니다.

---

**슬라이드 2 (또는 필요에 따라 여러 슬라이드): 질문 1: 프로젝트의 문제 정의가 명확한가요?**

*   **질문**: 질문을 눈에 띄게 표시합니다.
*   **답변**: 핵심 구성 요소를 위해 글머리 기호를 사용하여 상세한 답변을 제시합니다.

---

**슬라이드 3 (또는 필요에 따라 여러 슬라이드): 질문 2: 공공데이터는 어떻게 활용되며, 신뢰성 확보 방안은 무엇인가요?**

*   **질문**: 질문을 눈에 띄게 표시합니다.
*   **답변**: RAG, 데이터 선택, 사용자 피드백의 역할을 강조하여 상세한 답변을 제시합니다.

...

(예상되는 다른 심사위원 질문들을 각각 별도의 논리적 섹션/슬라이드에 이어서 진행합니다)

---

**슬라이드 X: Q&A - 압박/반박 질문**

*   **전환 슬라이드**: 더 어려운 질문을 위한 섹션을 소개합니다.

---

**슬라이드 Y (또는 필요에 따라 여러 슬라이드): 압박 질문 1: GPT-5-mini의 환각 현상, 기술적으로 어떻게 방어하고 있으며 비상 대응 메커니즘은?**

*   **질문**: 반박 질문을 눈에 띄게 표시합니다.
*   **답변**: 아키텍처적 안전장치 (RAG, Trust Engine), 가드레일, 투명성을 강조하며 상세한 방어 스크립트를 제시합니다.

---

**슬라이드 Z (또는 필요에 따라 여러 슬라이드): 압박 질문 2: 사용자 피드백 검증의 효율성과 공정성, 허위 피드백 발생 시 견고성은?**

*   **질문**: 반박 질문을 눈에 띄게 표시합니다.
*   **답변**: 다단계 검증, 이상 탐지, 가중치 부여 피드백에 초점을 맞춰 상세한 방어 스크립트를 제시합니다.

...

(다른 반박 질문들을 이어서 진행합니다)

### 시각적 요소 및 모범 사례:
*   **슬라이드당 최소 텍스트**: 너무 많은 텍스트로 청중을 압도하지 않도록 합니다. 키워드를 사용하고 구두로 자세히 설명합니다.
*   **일관된 브랜딩**: 전체 프레젠테이션의 시각적 테마를 유지합니다.
*   **명확한 강조**: 주요 용어(예: 'Trust Engine', 'RAG', 'Guardrails', '신뢰도 점수')를 강조하기 위해 굵은 글씨나 다른 색상을 사용합니다.
*   **상호 작용을 위한 공간**: 슬라이드에 여백을 남겨두고, 특히 라이브 Q&A 중에 그림을 그리거나 주석을 달 것으로 예상되는 경우 더욱 그렇습니다.
*   **백업 슬라이드**: 더 심층적인 질문이 요청될 경우를 대비하여 추가 데이터, 기술 세부 정보 또는 복잡한 주제에 대한 확장된 설명을 담은 숨겨진 슬라이드를 몇 개 준비합니다.

### 시뮬레이션 결과를 활용한 RAG 메타데이터 필드 가중치 제안

`summary_df`에 나타난 신뢰도 점수 변화(`Score Change`)는 각 개선 방안이 여행 플랜의 신뢰도에 미치는 영향력을 보여줍니다. 이 분석을 통해 RAG 검색 시 다음과 같은 메타데이터 필드에 가중치를 두어 검색 정확도와 응답 품질을 높일 수 있습니다.

#### 1. 높은 가중치를 부여해야 할 메타데이터 필드:

*   **`feedback_summary` (사용자 피드백 요약)**
    *   **근거**: '성산일출봉 (야외)' 플랜의 시나리오 3에서 **부정적 피드백 해소 및 긍정 피드백 누적** 시 **+30점**이라는 가장 큰 단일 점수 변화를 보였습니다. 이는 사용자 피드백이 정보의 신뢰도에 미치는 영향이 매우 크다는 것을 의미합니다. RAG는 긍정적인 피드백을 받은 정보나, 최근 부정적 피드백이 해소된 정보를 우선적으로 검색하도록 가중치를 두어야 합니다.

*   **`suitability_tags` (적합성 태그 - 특히 날씨, 운영 시간 관련)**
    *   **근거**: '성산일출봉 (야외)' 플랜의 시나리오 5에서 **날씨 적합성 해결** 시 **+25점**, 시나리오 6에서 **운영시간 일치 해결** 시 **+20점**이 상승했습니다. '협재 해변' 플랜의 시나리오 2에서도 **날씨 조건 반영**이 점수 향상에 크게 기여했습니다. 날씨, 운영 시간과 같은 실시간 또는 상황적 적합성 정보는 사용자 만족도와 직결되므로, 쿼리 조건에 맞는 `suitability_tags`를 가진 청크에 높은 가중치를 주어 개인화된 추천을 강화해야 합니다.

*   **`public_data_quality` (공공데이터 품질)**
    *   **근거**: '성산일출봉 (야외)' 플랜의 시나리오 4에서 **공공데이터 품질 개선** 시 **+20점**이 상승했으며, '협재 해변' 플랜의 시나리오 2에서도 품질 개선이 점수 향상에 기여했습니다. 이는 원본 데이터의 신뢰성이 근본적으로 중요함을 나타냅니다. '최상', '상' 등 품질 수준이 높은 데이터를 포함하는 청크에 더 높은 가중치를 부여해야 합니다.

#### 2. 중요한 기본 가중치를 부여해야 할 메타데이터 필드:

*   **`discrepancy_status` (불일치 상태)**
    *   **근거**: '성산일출봉 (야외)' 플랜의 시나리오 1에서 **정보 불일치 탐지 및 경고 해결** 시 `Initial Score` 0점에서 `Improved Score` 5점으로 변화하는 데 기여했습니다. 비록 단일 점수 변화는 다른 항목보다 작아 보이나, 이는 가장 기초적인 신뢰도의 문제입니다. 불일치 상태가 '불일치 없음'인 청크에 우선적인 가중치를 부여하여 잘못된 정보가 전달되는 것을 방지해야 합니다.

*   **`last_verified_date` (최종 검증일)**
    *   **근거**: '성산일출봉 (야외)' 플랜의 시나리오 1에서 **정보 최신성 해결**에 기여했습니다. 정보의 최신성은 모든 정보의 기본 전제이므로, 최근에 검증되거나 업데이트된 정보를 포함하는 청크에 가중치를 부여하여 정보의 유효성을 보장해야 합니다.

이러한 가중치 적용을 통해 RAG 엔진은 단순한 키워드 매칭을 넘어, 정보의 신뢰도, 최신성, 상황 적합성 및 사용자 경험까지 고려한 지능적인 검색 및 답변 생성을 수행할 수 있을 것입니다.

### RAG 메타데이터 필드별 우선순위 리스트

'Pack Your Jeju' 신뢰도 향상 시뮬레이션 결과(특히 `summary_df`의 `Score Change`)를 바탕으로, RAG 검색 시 가중치를 두어 정보 검색 및 LLM 응답 품질을 최적화할 수 있는 메타데이터 필드별 우선순위는 다음과 같습니다.

---

#### **🌟 1순위: 핵심 가중치 (Highest Priority Weight)**

이 필드들은 신뢰도 점수 변화에 가장 큰 영향을 미치며, 사용자 만족도 및 정보의 유효성에 직결되므로 RAG 검색 시 최우선적으로 높은 가중치를 부여해야 합니다.

*   **`feedback_summary` (사용자 피드백 요약)**
    *   **근거**: 시나리오 3에서 **+30점** 상승을 기록하며 가장 큰 단일 점수 변화를 보였습니다. 부정적 피드백 해소 및 긍정 피드백 누적이 서비스 신뢰도에 결정적인 영향을 미칩니다. (LLM이 답변의 긍정적 사용자 경험을 우선적으로 참조하도록 유도)

*   **`suitability_tags` (적합성 태그 - 특히 날씨, 운영 시간 관련)**
    *   **근거**: 시나리오 5에서 날씨 적합성 해결 시 **+25점**, 시나리오 6에서 운영시간 일치 해결 시 **+20점**을 기록했습니다. '협재 해변'에서도 날씨 조건 반영이 크게 기여했습니다. 사용자 쿼리(예: '비 오는 날 실내', '아이 동반')에 가장 적합한 정보를 제공하는 데 필수적입니다. (사용자 조건에 맞는 정보 우선 검색)

---

#### **✨ 2순위: 중요 가중치 (High Priority Weight)**

이 필드들은 정보의 근본적인 품질과 신뢰성을 보장하며, 1순위 필드만큼은 아니지만 여전히 매우 중요한 영향을 미치므로 높은 수준의 가중치를 유지해야 합니다.

*   **`public_data_quality` (공공데이터 품질)**
    *   **근거**: 시나리오 4에서 공공데이터 품질 개선 시 **+20점** 상승했습니다. 정보의 원천적인 신뢰도를 나타내며, LLM이 불확실한 정보에 의존하지 않도록 돕습니다. (품질이 높은 공공데이터를 근거로 우선 참조)

*   **`discrepancy_status` (불일치 상태)**
    *   **근거**: 시나리오 1에서 정보 불일치 탐지 및 경고를 해결하여 초기 0점에서 5점으로 상승하는 데 기여했습니다. 비록 점수 변화는 작았으나, 불일치 정보는 서비스의 신뢰도를 근본적으로 저해하므로, '불일치 없음' 상태의 정보에 높은 가중치를 부여해야 합니다. (불확실하거나 오류 가능성이 있는 정보 배제)

---

#### **💡 3순위: 기본 가중치 (Standard Priority Weight)**

이 필드는 모든 정보의 기본 전제 조건이자 유효성 유지에 필수적이므로, 항상 적절한 기본 가중치를 적용해야 합니다.

*   **`last_verified_date` (최종 검증일)**
    *   **근거**: 시나리오 1에서 정보 최신성 해결에 기여했습니다. 정보의 최신성 유지는 기본적인 신뢰도 요소입니다. 오래된 정보보다는 최근에 검증된 정보를 우선적으로 제공하도록 가중치를 부여합니다. (최신 정보를 우선적으로 제공)

### 주요 메타데이터 필드별 RAG 검색 정확도 평가 지표 제안

앞서 제안된 RAG 메타데이터 필드별 우선순위를 바탕으로, 각 필드가 검색 결과의 정확도에 미치는 영향을 정량적으로 평가하기 위한 지표를 제안합니다. 이러한 지표들을 활용하여 RAG 시스템을 고도화하고, LLM의 응답 품질을 개선할 수 있습니다.

---

#### **🌟 1순위 메타데이터 평가 지표**

1.  **`feedback_summary` (사용자 피드백 요약)**
    *   **평가 지표**: **피드백 일치율 (Feedback Match Rate)**
    *   **설명**: 특정 사용자 피드백(예: '정보 불일치', '긍정적 경험 많음')이 태그된 문서를 검색했을 때, LLM의 답변이 해당 피드백을 적절히 반영하거나, 피드백 유형(긍정/부정)에 맞는 내용을 제공하는 비율. 또는 사용자 쿼리에 '정보 불일치'가 있는 장소는 배제하고 검색되는 비율.
    *   **측정 방법**: 사용자 쿼리와 함께 특정 `feedback_summary` 태그를 강제하여 검색된 문서의 LLM 답변이 의도한 방향(예: '정보 불일치' 태그 문서는 답변에서 해당 불일치 언급 또는 배제)으로 생성되는지를 전문가 평가 또는 정량적 방법(키워드 포함 여부)으로 측정.

2.  **`suitability_tags` (적합성 태그 - 특히 날씨, 운영 시간 관련)**
    *   **평가 지표**: **조건 적합성 검색 성공률 (Condition Suitability Retrieval Success Rate)**
    *   **설명**: 사용자 쿼리에 특정 조건(예: '비 오는 날 실내', '아이 동반', '현재 운영 중인 장소')이 명시되었을 때, `suitability_tags`가 해당 조건과 일치하는 문서 청크가 상위 검색 결과에 얼마나 많이 포함되는지, 그리고 LLM의 답변이 이 조건에 부합하는 장소를 추천하는 비율.
    *   **측정 방법**: 다양한 조건 쿼리에 대해 RAG가 검색한 상위 K개 문서에 해당 `suitability_tags`를 가진 문서의 비율을 측정. LLM 답변에서 조건 부합 여부를 전문가 평가.

---

#### **✨ 2순위 메타데이터 평가 지표**

3.  **`public_data_quality` (공공데이터 품질)**
    *   **평가 지표**: **고품질 데이터 검색 우선순위 (High-Quality Data Ranking Priority)**
    *   **설명**: 동일한 키워드에 대해 `public_data_quality`가 '최상' 또는 '상'으로 태그된 문서 청크가 '중' 또는 '하'로 태그된 문서보다 상위 검색 결과에 얼마나 우선적으로 노출되는지.
    *   **측정 방법**: 특정 키워드에 대해 다양한 품질 태그를 가진 문서들을 준비한 후, RAG 검색 시 반환되는 문서들의 `public_data_quality` 태그 분포를 분석하여 고품질 문서의 평균 순위를 측정.

4.  **`discrepancy_status` (불일치 상태)**
    *   **평가 지표**: **불일치 정보 회피율 (Discrepancy Avoidance Rate)**
    *   **설명**: RAG 검색 및 LLM 응답 과정에서 `discrepancy_status`가 '불일치 감지'로 태그된 정보가 사용자에게 제공되거나 답변의 근거로 활용되는 것을 얼마나 효과적으로 회피하는지.
    *   **측정 방법**: 의도적으로 불일치 정보가 포함된 쿼리를 생성하고, LLM 답변에서 해당 불일치 정보가 제시되는 빈도를 측정. 이상적으로는 불일치 정보는 답변에서 배제되거나, '확인 필요' 등의 경고와 함께 제시되어야 함.

---

#### **💡 3순위 메타데이터 평가 지표**

5.  **`last_verified_date` (최종 검증일)**
    *   **평가 지표**: **최신 정보 활용률 (Freshness Utilization Rate)**
    *   **설명**: RAG가 검색된 문서 청크 중 `last_verified_date`가 최신인 정보를 LLM 답변의 근거로 얼마나 자주 활용하는지. 또는 오래된 정보(특정 기준일 이전)는 검색에서 얼마나 효과적으로 낮은 가중치를 받거나 배제되는지.
    *   **측정 방법**: 시간 의존적인 쿼리(예: '최근에 업데이트된 맛집')를 사용하여 검색된 문서의 `last_verified_date` 분포를 분석. LLM 답변에서 최신 정보가 주로 인용되는 비율을 측정.

### Ragas 라이브러리를 활용한 RAG 검색 정확도 평가

Ragas는 Retrieval-Augmented Generation(RAG) 파이프라인의 성능을 평가하기 위한 오픈소스 프레임워크입니다. LLM이 생성한 답변의 **정확성(faithfulness)**, **관련성(relevancy)**, 그리고 검색된 문서의 **품질(context relevancy, context recall)** 등을 다양한 지표로 측정할 수 있습니다.

#### 1. Ragas 라이브러리 설치

#### 2. 평가 메트릭 정의 및 데이터 준비

Ragas를 사용하려면 평가할 `question`, LLM이 생성한 `answer`, RAG가 검색한 `contexts`, 그리고 사람이 작성한 `ground_truth` 답변이 포함된 데이터셋이 필요합니다. 여기서는 예시 데이터를 생성하여 사용합니다.

#### 3. Ragas 평가 실행

정의된 메트릭과 준비된 데이터셋을 사용하여 Ragas 평가를 실행합니다. 평가에는 다소 시간이 걸릴 수 있습니다.

#### 4. 평가 결과 분석

`score` 객체는 각 메트릭에 대한 평균 점수를 포함하고 있습니다. 이 점수들을 통해 RAG 시스템의 강점과 약점을 파악하고, 앞서 제안했던 메타데이터 필드별 가중치 조절을 통해 성능 개선 효과를 측정할 수 있습니다.

*   **Faithfulness (충실성)**: LLM 답변이 제공된 컨텍스트에 얼마나 부합하는가.
*   **Answer Relevancy (답변 관련성)**: LLM 답변이 질문에 얼마나 관련성이 있는가.
*   **Context Relevancy (컨텍스트 관련성)**: 검색된 컨텍스트가 질문에 얼마나 불필요한 정보를 포함하고 있는가 (낮을수록 좋다).
*   **Context Recall (컨텍스트 재현율)**: 정답(ground truth)이 검색된 컨텍스트에 얼마나 잘 포함되어 있는가.
*   **Context Precision (컨텍스트 정밀도)**: 검색된 컨텍스트 중 질문에 관련된 컨텍스트의 비율.

이러한 지표들을 주기적으로 모니터링하고, 특정 메타데이터 필드에 가중치를 부여한 후 재평가하여 RAG 시스템의 검색 정확도 향상을 정량적으로 확인할 수 있습니다.

### 향후 RAG 엔진에 추가할 메타데이터 항목 제안

'Pack Your Jeju' 프로젝트의 신뢰도 향상 개선 방안들을 고려할 때, RAG 엔진의 문서 청크에 다음과 같은 메타데이터 항목들을 추가하면 검색 및 답변 생성 품질을 더욱 높일 수 있습니다.

1.  **`public_data_quality` (공공데이터 품질)**
    *   **설명**: 해당 청크가 참조하는 공공데이터의 품질 수준 (예: '최상', '상', '중', '하'). 데이터의 신뢰성을 직접적으로 반영하며, LLM이 답변의 신뢰도를 판단하는 근거로 활용될 수 있습니다.
    *   **개선 방안 연관**: '지속적인 공공데이터 확보 및 RAG 엔진 고도화'와 연관하여, 품질이 높은 데이터를 우선적으로 검색하도록 가중치를 줄 수 있습니다.

2.  **`last_verified_date` (최종 검증일)**
    *   **설명**: 해당 정보가 마지막으로 검증되거나 업데이트된 날짜. 정보의 최신성을 나타내는 중요한 지표입니다.
    *   **개선 방안 연관**: '정보 최신성 관리 자동화'와 직접적으로 연결되며, 오래된 정보에 대한 경고 또는 낮은 가중치 부여에 활용됩니다.

3.  **`feedback_summary` (피드백 요약)**
    *   **설명**: 해당 장소 또는 정보에 대한 사용자 피드백을 요약한 태그 (예: '정보 불일치', '긍정적 경험 많음', '운영 시간 변동 잦음').
    *   **개선 방안 연관**: '사용자 피드백 시스템 강화'와 연관하여, 특정 피드백이 많은 정보에 대해 LLM이 더 신중하게 답변하거나, 관련 정보를 추가로 제공하도록 유도할 수 있습니다.

4.  **`suitability_tags` (적합성 태그)**
    *   **설명**: 특정 조건(날씨, 동반자, 이동수단)에 대한 적합성 정보를 미리 태그 형태로 저장 (예: '비 올 때 실내', '아이 동반 좋음', '뚜벅이 적합', '렌터카 필수').
    *   **개선 방안 연관**: '하루방 에이전트의 실시간 정보 연동 확장'과 연관하여, 사용자 쿼리의 조건에 맞는 정보를 더 빠르게 찾아내어 개인화된 추천을 강화합니다.

5.  **`discrepancy_status` (불일치 상태)**
    *   **설명**: 해당 정보에서 현재 감지된 불일치 여부 (예: '불일치 없음', '운영 시간 불일치 감지', '날씨 조건 부적합').
    *   **개선 방안 연관**: '정보 불일치 탐지 및 경고 시스템 정교화'와 직접적으로 연결되며, LLM이 불일치 정보를 사용자에게 경고하거나 대안을 제시하는 데 활용됩니다.

이러한 메타데이터를 RAG 청크에 포함함으로써, LLM은 단순한 텍스트 내용뿐만 아니라 정보의 '품질', '최신성', '사용자 경험', '상황 적합성', '정확성'까지 고려하여 더욱 신뢰할 수 있고 맥락에 맞는 여행 정보를 제공할 수 있게 될 것입니다.

### 향후 RAG 엔진에 추가할 메타데이터 항목 제안

'Pack Your Jeju' 프로젝트의 신뢰도 향상 개선 방안들을 고려할 때, RAG 엔진의 문서 청크에 다음과 같은 메타데이터 항목들을 추가하면 검색 및 답변 생성 품질을 더욱 높일 수 있습니다.

1.  **`public_data_quality` (공공데이터 품질)**
    *   **설명**: 해당 청크가 참조하는 공공데이터의 품질 수준 (예: '최상', '상', '중', '하'). 데이터의 신뢰성을 직접적으로 반영하며, LLM이 답변의 신뢰도를 판단하는 근거로 활용될 수 있습니다.
    *   **개선 방안 연관**: '지속적인 공공데이터 확보 및 RAG 엔진 고도화'와 연관하여, 품질이 높은 데이터를 우선적으로 검색하도록 가중치를 줄 수 있습니다.

2.  **`last_verified_date` (최종 검증일)**
    *   **설명**: 해당 정보가 마지막으로 검증되거나 업데이트된 날짜. 정보의 최신성을 나타내는 중요한 지표입니다.
    *   **개선 방안 연관**: '정보 최신성 관리 자동화'와 직접적으로 연결되며, 오래된 정보에 대한 경고 또는 낮은 가중치 부여에 활용됩니다.

3.  **`feedback_summary` (피드백 요약)**
    *   **설명**: 해당 장소 또는 정보에 대한 사용자 피드백을 요약한 태그 (예: '정보 불일치', '긍정적 경험 많음', '운영 시간 변동 잦음').
    *   **개선 방안 연관**: '사용자 피드백 시스템 강화'와 연관하여, 특정 피드백이 많은 정보에 대해 LLM이 더 신중하게 답변하거나, 관련 정보를 추가로 제공하도록 유도할 수 있습니다.

4.  **`suitability_tags` (적합성 태그)**
    *   **설명**: 특정 조건(날씨, 동반자, 이동수단)에 대한 적합성 정보를 미리 태그 형태로 저장 (예: '비 올 때 실내', '아이 동반 좋음', '뚜벅이 적합', '렌터카 필수').
    *   **개선 방안 연관**: '하루방 에이전트의 실시간 정보 연동 확장'과 연관하여, 사용자 쿼리의 조건에 맞는 정보를 더 빠르게 찾아내어 개인화된 추천을 강화합니다.

5.  **`discrepancy_status` (불일치 상태)**
    *   **설명**: 해당 정보에서 현재 감지된 불일치 여부 (예: '불일치 없음', '운영 시간 불일치 감지', '날씨 조건 부적합').
    *   **개선 방안 연관**: '정보 불일치 탐지 및 경고 시스템 정교화'와 직접적으로 연결되며, LLM이 불일치 정보를 사용자에게 경고하거나 대안을 제시하는 데 활용됩니다.

이러한 메타데이터를 RAG 청크에 포함함으로써, LLM은 단순한 텍스트 내용뿐만 아니라 정보의 '품질', '최신성', '사용자 경험', '상황 적합성', '정확성'까지 고려하여 더욱 신뢰할 수 있고 맥락에 맞는 여행 정보를 제공할 수 있게 될 것입니다.

## 신뢰도 향상 시뮬레이션 결과 요약 및 상세 분석

본 시뮬레이션은 'Pack Your Jeju' 프로젝트에서 제안된 신뢰도 향상 전략들이 여행 플랜의 신뢰도 점수에 어떻게 실질적인 영향을 미 미치는지 검증하고자 수행되었습니다. 특히 초기 신뢰도 점수가 낮았던 '성산일출봉 (야외)' 플랜이 체계적인 개선 과정을 통해 만점인 100점에 도달하는 과정을 상세히 분석하고, '협재 해변' 플랜의 개선 사례도 함께 제시합니다.

### '성산일출봉 (야외)' 플랜의 신뢰도 0점에서 100점까지의 여정

'성산일출봉 (야외)' 플랜은 초기 **0점**이라는 매우 낮은 신뢰도 점수를 기록했습니다. 이는 '정보 불일치 탐지 및 경고', '정보 최신성', '공공데이터 기반 근거 제시율', '사용자 피드백 신호 반영', '날씨 적합성', '운영시간 일치' 등 다양한 감점 요인들이 복합적으로 작용한 결과였습니다. 하지만 일련의 시뮬레이션 시나리오를 통해 이러한 문제점들을 하나씩 해결하면서 최종적으로 100점 만점에 도달할 수 있었습니다.

1.  **시나리오 1: 정보 최신성 및 불일치 해결 (+5점)**
    *   **개선 내용**: '정보 불일치 탐지 및 경고'(-35점) 및 '정보 최신성'(-15점) 문제를 해결했습니다. 즉, 시스템이 실시간 정보를 업데이트하고 감지된 불일치 문제를 해소한 상황을 가정했습니다.
    *   **결과**: 초기 0점에서 **5점**으로 점수가 상승했습니다. 이는 핵심 정보의 정확성과 시의적절한 제공이 신뢰도 향상의 첫걸음임을 보여줍니다.

2.  **시나리오 3: 부정적 사용자 피드백 해소 및 긍정 피드백 누적 (+30점)**
    *   **개선 내용**: 시나리오 1 개선 후(5점)를 기준으로, '정보 불일치' 사용자 피드백(-30점)이 긍정적 피드백으로 전환되었습니다. 즉, 사용자 경험 개선과 함께 긍정적인 피드백이 누적된 상황을 가정했습니다.
    *   **결과**: 5점에서 **35점**으로 점수가 크게 상승했습니다. 사용자 피드백이 신뢰도에 미치는 강력한 영향을 확인할 수 있었습니다.

3.  **시나리오 4: 공공데이터 품질 개선 (+20점)**
    *   **개선 내용**: 시나리오 3 개선 후(35점)를 기준으로, '공공데이터 기반 근거 제시율'을 '저'에서 '고'로 향상시켰습니다. 즉, RAG 엔진을 통해 제공되는 공공데이터의 품질과 근거 제시 수준을 높인 상황을 가정했습니다.
    *   **결과**: 35점에서 **55점**으로 점수가 상승했습니다. 데이터 자체의 품질 개선이 신뢰도에 미치는 중요성을 보여줍니다.

4.  **시나리오 5: 날씨 적합성 해결 (+25점)**
    *   **개선 내용**: 시나리오 4 개선 후(55점)를 기준으로, '날씨 적합성'(-25점) 문제를 해결했습니다. 즉, '강풍'이라는 날씨 조건에도 적합한 대안을 제시하거나, 날씨 조건 자체가 '맑음'으로 변화하여 야외 활동이 가능해진 상황을 가정했습니다.
    *   **결과**: 55점에서 **80점**으로 점수가 상승했습니다. 외부 환경 변화에 대한 유연한 대응 및 정보 제공이 사용자 신뢰도에 필수적임을 시사합니다.

5.  **시나리오 6: 운영시간 일치 해결 (+20점)**
    *   **개선 내용**: 시나리오 5 개선 후(80점)를 기준으로, '운영시간 일치'(-20점) 문제를 해결했습니다. 즉, 플랜과 장소의 실제 운영 시간이 일치하도록 정보를 업데이트하거나, 사용자의 방문 시간을 조정하도록 안내하는 등의 조치가 이루어진 상황을 가정했습니다.
    *   **결과**: 80점에서 최종 **100점**에 도달했습니다. 운영시간과 같은 기본적인 정보의 정확성이 만점 달성에 중요함을 확인했습니다.

### '협재 해변' 플랜의 신뢰도 65점에서 100점까지의 여정

'협재 해변' 플랜은 초기 **65점**으로 중간 수준의 신뢰도를 보였습니다. 주요 감점 요인은 '공공데이터 기반 근거 제시율'(보통)과 '날씨 적합성'(비 오는 날 야외 해변 부적합)이었습니다.

1.  **시나리오 2: 공공데이터 품질 개선 및 날씨 조건 반영 (+35점)**
    *   **개선 내용**: '공공데이터 기반 근거 제시율'을 '중'에서 '고'로 개선하고, '날씨 적합성' 문제를 해결했습니다. 즉, 공공데이터의 품질을 높이고, 비 오는 날에도 적합한 대안 제시 또는 날씨 변화를 가정했습니다.
    *   **결과**: 65점에서 **100점**으로 크게 상승했습니다. 이는 데이터 품질과 상황 적합성을 확보함으로써 신뢰도를 단번에 만점으로 끌어올릴 수 있음을 보여줍니다.

### 결론

이러한 시뮬레이션 결과는 'Pack Your Jeju' 프로젝트의 신뢰도 평가 모델과 개선 전략이 매우 효과적임을 입증합니다. 초기 낮은 신뢰도 점수를 가진 플랜이라 할지라도, '정보 불일치 탐지 및 경고', '정보 최신성', '사용자 피드백', '공공데이터 품질', '날씨 적합성', '운영시간 일치' 등 각 감점 요인을 체계적으로 식별하고 해결해 나간다면 최종적으로 100점 만점의 높은 신뢰도를 확보할 수 있습니다. 이는 사용자가 제주 여행을 계획하고 실행하는 데 있어 'Pack Your Jeju'가 가장 신뢰할 수 있는 정보를 제공할 수 있는 시스템이 될 것임을 시사합니다.

## 제주를 담다 (Jeju Damda) 핵심 요약

**서비스 명**: 제주를 담다

**핵심 메시지**:
*   공공데이터를 활용해 여행을 계획하고, 사용자의 실제 방문 기록과 피드백을 통해 정보의 신뢰도를 지속적으로 높이는 순환형 AI 여행 플랜 서비스입니다.

**문제 인식**:
*   제주 여행 정보는 많지만, 사용자들은 '오늘 이 장소를 믿고 선택해도 되는지'에 대한 명확한 근거를 필요로 합니다.

**해결 방안**:
*   공공데이터 기반의 여행 플랜을 제공하고, 사용자의 실제 방문 기록과 피드백을 '신뢰 신호'로 시스템에 반영하여 서비스의 데이터와 판단 기준을 끊임없이 개선합니다.

**차별점**:
*   단순한 추천을 넘어, 사용할수록 정보의 신뢰도와 정확성이 향상되는 시스템을 구축하여 사용자에게 더 나은 여행 경험을 제공합니다.

## 우리는 왜 이 주제를 선택했나

**핵심 질문**: "어디가 유명한가?"가 아니라 "**지금 내 일정에 넣어도 되는가?**"

**문제 인식**:
*   여행자는 단순히 유명한 장소 이름을 아는 것을 넘어, **실제 일정 안에서 움직일 수 있는지, 오늘 가도 괜찮은지, 후회하지 않을지**에 대한 실질적인 판단 근거를 필요로 합니다.

**선택 및 방향성**:
*   다양한 아이디어 중 '여행 플랜 서비스'를 최종 선택했습니다.
*   서비스의 목표를 "제주에서 좋은 곳을 알려주자"에서 "**사용자가 믿고 실행할 수 있는 여행 판단을 돕자**"로 좁혔습니다.

**결론**:
*   이러한 관점은 데이터 수집, RAG 구조, 신뢰도 계산, 방문 기록 설계 등 프로젝트의 모든 요소에 연결됩니다.
*   여행 플랜은 단순히 화면 기능이 아니라, **신뢰도 분석을 사용자 경험으로 풀어내는 방법**으로 정의했습니다.

## 제주 여행 정보의 문제

**핵심 메시지**: 제주 여행의 문제는 정보 부족이 아니라, **많은 정보 속에서 믿고 실행할 기준이 부족**하다는 점입니다.

**문제 인식**:
*   블로그, SNS, 지도 리뷰, 관광 포털, 공공데이터 등 제주 여행 정보는 이미 **넘쳐납니다.**
*   정보가 많을수록 여행자가 쉽게 결정하기보다, **광고성, 오래된 정보, 불분명한 출처 등**을 직접 판단해야 하는 부담이 커집니다.

**주요 이슈**:
*   **정보의 최신성**: 운영시간, 휴무, 주소, 연락처, 가격 등 핵심 정보가 변경되면 여행 실패로 이어질 수 있습니다.
*   **상황적합성 부족**: 같은 장소라도 동행자, 날씨, 이동수단, 여행 목적에 따라 적합성 여부가 달라집니다.

**결론**: 저희가 정의한 문제는 "정보가 없다"가 아니라, "**이 정보를 지금 믿고 실행해도 되는지 판단하기 어렵다**"였습니다. 이것이 서비스의 출발점입니다.

## 프리아이펠톤 규정에 대한 우리의 해석

**핵심 메시지**: 좋은 모델을 찾는 것이 아니라, **같은 모델(gpt-5-mini)도 믿을 만하게 작동하도록 구조를 만드는 것**이 과제라고 해석했습니다.

**제약 사항**: gpt-5-mini 모델 사용.

**해석**: 이 제약을 단순한 한계가 아닌, **'같은 모델이 더 믿을 만하게 답하도록 만드는 구조 설계'**의 기회로 보았습니다.

**접근 방식**:
*   gpt-5-mini를 자유로운 추천 모델로 두지 않았습니다.
*   대신, **공공데이터, 근거 문서, 수정 요청 이력, 날씨 및 이동 조건, 사용자 방문 피드백**을 먼저 확인하게 했습니다.
*   확인된 범위 안에서만 설명하고 조율하도록 LLM의 역할을 **제한**했습니다.

**결론**: 신뢰도 분석은 단순히 답변을 좋게 만드는 것이 아니라, **답변 이전에 어떤 근거를 봤고, 무엇을 확인했으며, 어떤 부분은 아직 모르는지 드러내는 일**이라는 관점으로 접근했습니다.

## 서비스 핵심 정의

**핵심 메시지**: 제주를 담다는 추천에서 끝나지 않고, **공공데이터와 방문 기록이 다시 연결되는 순환형 AI 여행플랜 서비스**입니다.

**서비스 작동 방식**:
1.  **출발**: 공공데이터를 기반으로 여행 후보를 생성합니다.
2.  **선택**: 사용자는 후보의 출처, 신뢰도, 확인 필요 항목을 확인하고 여행 플랜에 추가합니다.
3.  **순환**: 사용자의 실제 방문 기록은 내부 피드백 대시보드를 통해 분류, 검증된 후, 신뢰 가능한 신호만 서비스 판단 구조로 다시 연결됩니다.

**핵심 개념**: "**순환**"
*   추천으로 끝나는 것이 아니라, **방문 기록이 다음 여행자의 신뢰도 판단 근거로 활용**됩니다.
*   결론적으로, 제주를 담다는 공공데이터로 계획하고, **방문 기록으로 다시 좋아지는 AI 여행플랜 서비스**입니다.

## 데이터 수집 전략

**핵심 메시지**: 저희는 데이터를 많이 모으기보다, **여행자가 실제 판단에 쓸 수 있는 데이터만 선별**했습니다.

**데이터 수집 원칙**:
1.  **출처 명확**: 출처가 분명한 공공데이터를 우선 사용합니다.
2.  **필요성**: 사용자가 실제 일정에 포함할지 판단하는 데 필요한 데이터만 수집합니다.
3.  **구조화**: RAG 및 신뢰도 점수(Trust Score)에 바로 연결될 수 있도록 구조화합니다.

**여행자가 판단할 때 필요한 정보**:
*   장소의 위치
*   운영 여부 확인 가능성
*   이동 및 주차 가능 여부
*   날씨 영향 여부
*   정보 수정 이력 유무

**전략 결론**:
*   데이터의 양보다는 **'왜 이 후보를 추천했는가'를 설명할 수 있는 형태로 정리**하는 것을 중요하게 생각하여, 넓게 모으는 대신 정확하게 모으는 전략을 선택했습니다.

## 수집 데이터와 근거 구조

**핵심 메시지**: 저희는 추천 데이터가 아니라, **판단과 검증에 필요한 근거 데이터를 연결**했습니다.

**수집 및 구조화된 5가지 데이터**:
1.  **비짓제주 관광지 데이터**: 장소명, 주소, 좌표, 카테고리 등 여행 후보의 기본 정보.
2.  **비짓제주 수정요청 데이터**: 운영시간, 주소, 연락처, 가격 변경 등 **최신성 위험을 판단하는 신호**.
3.  **이동과 편의 데이터**: 주차장, 정류장, 접근성 정보 등 **실제 여행 가능성을 판단**하는 데 필요.
4.  **기상청 날씨 API**: 비, 바람, 강수확률, 단기예보 등 오름, 해변, 산책 코스의 **적합도를 변화시키는 정보**.
5.  **사용자 피드백과 검증 신호**: 방문 여부, 정보 다름, 만족도, 불만족 등 내부 대시보드에서 분류되는 피드백.

**데이터 흐름 및 목적**:
*   이 데이터들은 개별적으로 존재하지 않고, **원본 데이터 -> 검증 포인트 -> 서비스 반영**으로 이어지는 순환 흐름을 갖습니다.
*   데이터 구조의 목적은 "**많이 가지고 있다**"가 아니라 "**근거 있게 판단할 수 있다**"입니다.

## 공식 데이터도 수정된다는 증거

**핵심 메시지**: 공식 데이터도 바뀝니다. 그래서 여행 서비스는 추천보다 먼저 **최신성 확인 구조**를 가져야 합니다.

**핵심 발견**:
*   비짓제주 수정요청 데이터를 통해 공식 데이터에서도 **총 1,686건의 수정요청**이 발생했음을 확인했습니다.

**수정요청의 의미**:
*   이는 단순한 관리 기록이 아니라, 여행자 입장에서는 **실패 가능성을 알려주는 최신성 신호**입니다.
*   운영시간, 주소, 연락처, 가격, 메뉴, 사진, 상세 설명 등의 변경은 실제 여행 경험에 큰 영향을 미칩니다.

**결론**:
*   공공데이터가 믿을 수 없다는 뜻이 아니라, **데이터가 계속 업데이트된다는 사실까지 함께 반영**해야 한다는 의미입니다.
*   저희 서비스는 출처, 최신성, 수정 이력, 확인 필요 여부를 **함께 보면서** 현재 유효한 정보인지 판단합니다.

## 사용자 흐름

**핵심 메시지**: 사용자는 고르고, 근거를 확인하고, 플랜에 담고, 방문 기록을 남기며 **신뢰 순환에 참여**합니다.

**사용자 흐름 4단계**:
1.  **고르기**: 제주 지도에서 지역과 순간(예: 한림 지역, 혼자 여행, 힐링, 하루 일정)을 선택합니다.
2.  **근거 확인**: 후보 카드에서 단순 추천이 아닌 **신뢰도, 출처, 수정요청 이력, 확인 필요 항목**을 확인하고 납득하여 선택합니다.
3.  **플랜 담기**: 선택한 후보를 여행 플랜(여행팩)에 담습니다.
4.  **방문 기록**: 실제 방문 후 기록을 남기며, 이 기록은 내부 대시보드에서 검증되어 **다음 여행자의 신뢰 신호**가 됩니다.

**결론**: 제주를 담다의 사용자 흐름은 단순한 추천 소비가 아니라, **여행자가 판단하고 실행하고 기록을 남기면서 데이터 품질이 좋아지는 순환적인 흐름**입니다.

### Pack Your Jeju: 10분 발표 요약 (구어체로 다듬기)

---

#### 1. 문제 인식: 정보가 너무 많아도 문제, '신뢰할 수 있는 판단 기준'이 없어요. (1분)

*   **핵심**: 제주 여행 정보는 정말 많지만, **'지금, 나에게, 이 장소가 믿을 수 있는가?'** 에 대한 확실한 판단 기준이 부족하다는 게 문제입니다.
*   **이슈**: 정보가 **최신이 아니거나, 내 상황에 맞지 않는 경우(날씨, 동행자, 이동수단 등)** 여행이 망쳐질 수도 있겠죠.
*   **우리의 해석**: 좋은 모델을 찾는 것보다, **같은 모델(gpt-5-mini)이라도 믿을 만하게 작동하는 '시스템 구조'를 만드는 것**이 중요하다고 봤습니다.

#### 2. 서비스 핵심 정의: '제주를 담다', 계속 좋아지는 AI 여행 플랜 서비스 (1분 30초)

*   **핵심 메시지**: '제주를 담다'는 **공공데이터와 사용자 방문 기록이 서로 연결되어 계속 좋아지는 '순환형 AI 여행 플랜 서비스'**입니다.
*   **목표**: 단순히 유명한 곳을 알려주는 것을 넘어, **사용자들이 '믿고 실행할 수 있는 여행 판단'을 할 수 있도록 돕는 것**이 저희 목표입니다.
*   **작동 방식**: 공공데이터로 여행 후보를 추천하고 → 사용자는 신뢰도와 출처를 확인하고 플랜에 추가해요 → 다녀와서 방문 기록을 남기면 → 이 기록이 검증을 거쳐 다음 여행자를 위한 '신뢰 신호'로 다시 시스템에 반영되는 거죠. 계속 순환하면서 점점 더 좋아집니다.

#### 3. 데이터 전략: '판단과 검증'에 딱 필요한 데이터만 모았어요. (1분 30초)

*   **수집 원칙**: 데이터를 무작정 많이 모으기보다 **'여행자가 실제로 판단할 때 쓸 수 있는 데이터'만 골라서** 모았습니다.
*   **핵심 데이터**: 비짓제주 관광지 데이터, (정보 변경을 알려주는) 수정요청 데이터, 이동/편의 데이터, 기상청 날씨 API, 사용자 피드백 등이 핵심입니다.
*   **특징**: **공식 데이터도 사실 계속 바뀝니다.** (비짓제주에서만 1,686건의 수정요청을 발견했어요!) 그래서 저희 서비스는 출처, 최신성, 수정 이력, '확인 필요' 여부까지 **함께 보면서** 지금 이 정보가 유효한지 판단합니다.

#### 4. 사용자 흐름 (데모 요약): 판단하고, 담고, 기록하면서 신뢰를 만들어가요. (2분)

*   **지역 신뢰 대시보드**: 단순히 추천 목록만 보여주는 게 아니라, **'이 지역에서 무엇을 얼마나 믿고 고를 수 있는지'**를 먼저 보여줍니다. (데이터가 부족하면 억지로 추천하지 않고 '확인 필요'로 솔직하게 표시해요.)
*   **후보 카드와 여행팩**: 후보를 여행 계획에 추가하기 전에, **신뢰도 점수, 공공데이터 출처, 수정요청 이력, 이용/주차 여부, '확인 필요' 항목** 등 근거 정보를 먼저 꼼꼼히 확인합니다. 여행팩은 단순한 북마크가 아니라 '실제로 실행 가능한 계획'입니다.
*   **방문 기록과 신뢰도 업데이트**: 사용자가 다녀와서 남긴 방문 기록은 **즉시 반영되는 게 아니라, 내부 검증 과정을 거쳐서** 신뢰도 업데이트와 공공데이터 개선 신호로 연결됩니다. (악성 입력이나 단순 착오를 방지하기 위해서죠.)
*   **하루방 에이전트**: 무엇이든 다 아는 챗봇이 아닙니다. **조회된 근거 안에서만 정보를 조율하고 설명하는 '여행 판단 에이전트'**예요. (모르는 내용은 '확인 필요'로 명확히 알려줍니다.)

#### 5. 시스템 구조: RAG + Trust Engine + gpt-5-mini (1분 30초)

*   **핵심**: LLM이 먼저 대답하는 구조가 아니라, **데이터와 규칙(Trust Engine)이 먼저 판단하고, LLM(gpt-5-mini)은 마지막에 그걸 '설명'하는 구조**입니다.
*   **RAG**: 공공데이터와 외부 API에서 필요한 근거를 찾아옵니다.
*   **Trust Engine**: 찾아온 정보를 바탕으로 **신뢰도 점수, '확인 필요' 여부, 아예 배제할 조건** 등을 계산해서 후보를 걸러냅니다.
*   **gpt-5-mini**: 새로운 후보를 만들지 않아요. 이미 조회되고 계산된 근거를 바탕으로 사용자에게 **이해하기 쉬운 말로 '설명'해주고, 여행 계획 방향을 '조율'하는 역할**을 합니다.
*   **차별점**: 같은 `gpt-5-mini` 모델을 쓰더라도, 이런 구조 덕분에 **답변의 일관성과 검증 가능성을 훨씬 높일 수 있었습니다.**

#### 6. 신뢰성 검증 및 시뮬레이션 결과 (1분)

*   **신뢰도**: 저희가 말하는 신뢰도는 단순한 별점이 아닙니다. **공식성, 최신성, 출처, 조건 적합성, 피드백 신호** 등을 다 따로 분석해서 설명해 주는 구조예요.
*   **시뮬레이션**:
    *   처음에 0점이었던 '성산일출봉 (야외)' 플랜이 **단계적인 개선(정보 최신성 및 불일치 해결 (+5점), 부정적 피드백 해소 및 긍정 피드백 누적 (+30점), 공공데이터 품질 개선 (+20점), 날씨 적합성 해결 (+25점), 운영시간 일치 해결 (+20점))**을 통해 최종 100점에 도달하는 것을 증명했습니다.
    *   또, 65점이었던 '협재 해변' 플랜 역시 **공공데이터 품질 개선 및 날씨 조건 반영 (+35점)**을 통해 100점에 도달하는 것을 보여주면서, 저희의 신뢰도 향상 전략들이 실제로 효과가 있다는 것을 정량적으로 입증했습니다.
*   **결과**: 각 시뮬레이션 시나리오를 통해 신뢰도 향상 전략들이 정량적으로 효과적이라는 것을 입증했습니다.

#### 7. 결론: 더 많이 추천하는 게 아니라, '신뢰'를 드리는 여행 서비스 (1분)

*   **서비스 견고성**: 외부 데이터가 부족해도 **억지로 추천하지 않고 '확인 필요'로 솔직하게 표시**해서 정직함을 지킵니다. API 키는 Secrets로 관리해서 보안도 튼튼하게 하고요.
*   **최종 메시지**: '제주를 담다'는 더 많이 추천하는 서비스가 아니라, **공공데이터로 계획을 세우고, 사용자의 방문 기록으로 계속 좋아지는 '신뢰 기반' 여행 서비스**입니다.

## 🎬 '제주를 담다' 최종 발표 및 Q&A 통합 시나리오

### 1. 발표 도입: 문제의 재정의
*   **핵심**: 정보 과잉 시대, 우리에게 필요한 것은 '더 많은 정보'가 아니라 '믿고 실행할 판단 기준'입니다.
*   **멘트**: "제주 여행, 정보는 넘쳐나지만 과연 무엇을 믿어야 할까요? 저희는 GPT-5-mini를 활용해 단순히 답을 하는 챗봇이 아닌, 공공데이터와 사용자 피드백을 결합해 '신뢰할 수 있는 판단 기준'을 제공하는 서비스를 만들었습니다."

### 2. 기술적 핵심: RAG → Trust Engine → gpt-5-mini
*   **구조**: 데이터가 먼저 판단하고, AI는 설명만 합니다.
*   **멘트**: "저희 시스템의 심장은 'Trust Engine'입니다. RAG를 통해 수집된 팩트 데이터를 바탕으로 공식성, 최신성, 상황 적합성 등 6가지 지표를 계산합니다. gpt-5-mini는 이 검증된 범위 안에서만 사용자에게 친절한 가이드 역할을 수행합니다."

### 3. 신뢰의 증명: 시뮬레이션 및 정량적 수치
*   **내용**: 0점이었던 플랜이 어떻게 100점이 되는가?
*   **멘트**: "시뮬레이션 결과, 정보 불일치 해결과 피드백 누적만으로도 신뢰도 점수가 비약적으로 상승하는 것을 확인했습니다. 이는 저희의 신뢰도 루브릭이 실제 여행의 품질을 정량적으로 관리할 수 있음을 증명합니다."

### 4. Q&A 방어 전략 (심사위원 압박 질문 대응)

#### Q1. LLM의 환각(Hallucination) 문제를 어떻게 완벽히 차단하나요?
> **A**: "저희는 LLM에게 생성의 자유를 주지 않습니다. RAG를 통해 확보된 근거 문서와 Trust Engine이 계산한 신뢰도 지표(Metadata)가 답변의 가이드라인이 됩니다. 근거 범위를 벗어나는 질문에는 '확인 필요'라고 솔직하게 답하는 정직한 구조가 저희의 핵심 방어 기제입니다."

#### Q2. 대량의 허위 피드백(어뷰징)이 발생하면 시스템이 무너지지 않나요?
> **A**: "사용자 피드백은 즉시 반영되지 않고 '검증 레이어'를 거칩니다. GPS 기반 위치 인증, 사진 대조, 그리고 사용자별 신뢰 등급에 따른 가중치 차등 부여를 통해 악의적인 데이터 오염을 기술적으로 차단하고 있습니다."

#### Q3. MVP 이후 실제 상용화 가능성과 로드맵은?
> **A**: "현재 관광지 중심의 데이터를 숙소, 음식점, 교통 API로 확장할 준비가 되어 있습니다. 제주에서의 성공 모델을 바탕으로 타 지역 공공데이터와 결합하여 '신뢰 기반 여행 정보 플랫폼'의 표준이 되는 것이 저희의 마일스톤입니다."

### 5. 최종 메시지: 신뢰로 그리는 여행의 미래
*   **멘트**: "'제주를 담다'는 더 많이 추천하는 서비스가 아닙니다. 공공데이터로 계획하고, 여러분의 발자취로 다시 좋아지는 '신뢰 순환형 서비스'입니다. 설렘부터 근거 있게, 제주를 담다가 함께하겠습니다. 감사합니다."

## 🛡️ 심사위원 Q&A 대응 전략 및 모범 답안

### 1. 일반 질문 (Service & Logic)
| 질문 | 답변 포인트 |
| :--- | :--- |
| **기존 여행 추천 서비스와 무엇이 다른가요?** | 단순히 '유명한 곳'을 추천하는 것이 아니라, **'지금 실행 가능한가'**를 데이터로 검증한다는 점이 다릅니다. GPT의 창작 능력을 억제하고 Trust Engine의 검증 능력을 우선시한 구조가 핵심입니다. |
| **공공데이터의 오류는 어떻게 해결하나요?** | 공식 데이터도 변할 수 있음을 인지하고, **'수정 요청 데이터'**와 **'사용자 피드백'**을 실시간으로 결합합니다. 데이터가 상충할 경우 '확인 필요' 상태로 전환하여 정보의 투명성을 유지합니다. |
| **사용자 피드백이 악의적으로 입력되면요?** | 모든 피드백은 즉시 반영되지 않습니다. 내부 검증 레이어에서 GPS 인증, 패턴 분석을 거치며, 신뢰도가 검증된 사용자의 피드백에 더 높은 가중치를 부여하는 알고리즘을 사용합니다. |

### 2. 기술 압박 질문 (Technical Deep-Dive)

#### Q: GPT-5-mini의 환각(Hallucination) 현상을 기술적으로 어떻게 차단했습니까?
**A:** "저희는 LLM을 정보를 찾는 '엔진'이 아니라, 검증된 정보를 전달하는 **'설명자(Speaker)'**로만 활용합니다.
1. **RAG 구조**: LLM이 학습 데이터가 아닌, 저희가 제공한 공공데이터 내에서만 답하도록 Context를 제한합니다.
2. **Trust Engine**: LLM에 데이터를 넘기기 전, 메타데이터 가중치(최신성, 품질, 피드백)를 통해 부적합한 정보를 1차로 필터링합니다.
3. **가드레일**: 근거가 부족한 경우 '모름' 혹은 '확인 필요'로 답변하도록 프롬프트 엔지니어링과 출력 검증 로직을 구현했습니다."

#### Q: RAG의 검색 정확도를 높이기 위한 메타데이터 전략은 무엇인가요?
**A:** "단순 키워드 매칭의 한계를 극복하기 위해 **5가지 핵심 메타데이터 필드**에 가중치를 둡니다.
* `feedback_summary` (최우선): 실제 방문자의 최신 경험 반영
* `suitability_tags`: 날씨/운영시간 등 실시간 상황 적합성
* `public_data_quality`: 원본 데이터의 공식성 등급
이러한 필드들을 벡터 검색의 Re-ranking 과정에 반영하여 사용자의 상황에 가장 정직한 정보를 상단에 배치합니다."

### 3. 비즈니스 및 확장성 (Vision)
| 질문 | 답변 포인트 |
| :--- | :--- |
| **제주도 외 지역으로 확장이 가능한가요?** | 네, 본 시스템은 공공데이터의 '구조'를 활용하므로, 데이터 형식이 유사한 타 지자체의 공공데이터와 API만 연결하면 즉시 수평적 확장이 가능합니다. |
| **수익 모델이나 향후 로드맵은?** | 신뢰도가 검증된 로컬 상권과의 연계, 혹은 여행지 최신 정보를 필요로 하는 타 플랫폼에 '신뢰 데이터(Trust API)'를 제공하는 B2B 모델을 고려하고 있습니다. |

### 💡 발표 전 최종 체크리스트
- [ ] **수치 강조**: "1,686건의 데이터 수정 사례", "0점에서 100점의 신뢰도 상승" 수치를 언급했는가?
- [ ] **기술 용어**: RAG, Trust Engine, gpt-5-mini의 상관관계를 명확히 설명했는가?
- [ ] **태도**: "완벽한 AI"가 아닌 "모르는 것은 확인이 필요하다고 말하는 정직한 AI"임을 강조했는가?

이 구성으로 발표를 마무리하시면 기술적 완성도와 서비스의 진정성을 모두 전달하실 수 있을 것입니다. 건승을 빕니다!

## 🛡️ 심사위원 Q&A 대응 전략 및 모범 답안

### 1. 일반 질문 (Service & Logic)
| 질문 | 답변 포인트 |
| :--- | :--- |
| **기존 여행 추천 서비스와 무엇이 다른가요?** | 단순히 '유명한 곳'을 추천하는 것이 아니라, **'지금 실행 가능한가'**를 데이터로 검증한다는 점이 다릅니다. GPT의 창작 능력을 억제하고 Trust Engine의 검증 능력을 우선시한 구조가 핵심입니다. |
| **공공데이터의 오류는 어떻게 해결하나요?** | 공식 데이터도 변할 수 있음을 인지하고, **'수정 요청 데이터'**와 **'사용자 피드백'**을 실시간으로 결합합니다. 데이터가 상충할 경우 '확인 필요' 상태로 전환하여 정보의 투명성을 유지합니다. |
| **사용자 피드백이 악의적으로 입력되면요?** | 모든 피드백은 즉시 반영되지 않습니다. 내부 검증 레이어에서 GPS 인증, 패턴 분석을 거치며, 신뢰도가 검증된 사용자의 피드백에 더 높은 가중치를 부여하는 알고리즘을 사용합니다. |

### 2. 기술 압박 질문 (Technical Deep-Dive)

#### Q: GPT-5-mini의 환각(Hallucination) 현상을 기술적으로 어떻게 차단했습니까?
**A:** "저희는 LLM을 정보를 찾는 '엔진'이 아니라, 검증된 정보를 전달하는 **'설명자(Speaker)'**로만 활용합니다.
1. **RAG 구조**: LLM이 학습 데이터가 아닌, 저희가 제공한 공공데이터 내에서만 답하도록 Context를 제한합니다.
2. **Trust Engine**: LLM에 데이터를 넘기기 전, 메타데이터 가중치(최신성, 품질, 피드백)를 통해 부적합한 정보를 1차로 필터링합니다.
3. **가드레일**: 근거가 부족한 경우 '모름' 혹은 '확인 필요'로 답변하도록 프롬프트 엔지니어링과 출력 검증 로직을 구현했습니다."

#### Q: RAG의 검색 정확도를 높이기 위한 메타데이터 전략은 무엇인가요?
**A:** "단순 키워드 매칭의 한계를 극복하기 위해 **5가지 핵심 메타데이터 필드**에 가중치를 둡니다.
* `feedback_summary` (최우선): 실제 방문자의 최신 경험 반영
* `suitability_tags`: 날씨/운영시간 등 실시간 상황 적합성
* `public_data_quality`: 원본 데이터의 공식성 등급
이러한 필드들을 벡터 검색의 Re-ranking 과정에 반영하여 사용자의 상황에 가장 정직한 정보를 상단에 배치합니다."

### 3. 비즈니스 및 확장성 (Vision)
| 질문 | 답변 포인트 |
| :--- | :--- |
| **제주도 외 지역으로 확장이 가능한가요?** | 네, 본 시스템은 공공데이터의 '구조'를 활용하므로, 데이터 형식이 유사한 타 지자체의 공공데이터와 API만 연결하면 즉시 수평적 확장이 가능합니다. |
| **수익 모델이나 향후 로드맵은?** | 신뢰도가 검증된 로컬 상권과의 연계, 혹은 여행지 최신 정보를 필요로 하는 타 플랫폼에 '신뢰 데이터(Trust API)'를 제공하는 B2B 모델을 고려하고 있습니다. |

### 💡 발표 전 최종 체크리스트
- [ ] **수치 강조**: "1,686건의 데이터 수정 사례", "0점에서 100점의 신뢰도 상승" 수치를 언급했는가?
- [ ] **기술 용어**: RAG, Trust Engine, gpt-5-mini의 상관관계를 명확히 설명했는가?
- [ ] **태도**: "완벽한 AI"가 아닌 "모르는 것은 확인이 필요하다고 말하는 정직한 AI"임을 강조했는가?

이 구성으로 발표를 마무리하시면 기술적 완성도와 서비스의 진정성을 모두 전달하실 수 있을 것입니다. 건승을 빕니다!

## 📊 [발표 PPT 목차 구성안]

### 1. [도입] 제주 여행의 보이지 않는 장벽
*   **정보 과잉의 시대**: 정보가 없는 게 아니라, '믿을 수 있는 판단 기준'이 없습니다.
*   **우리가 정의한 문제**: 광고와 노후 데이터 사이에서 방황하는 여행자들.

### 2. [서비스 정의] 제주를 담다: 신뢰의 순환 구조
*   **핵심 컨셉**: 공공데이터로 계획하고, 방문 기록으로 계속 좋아지는 AI 플랜 서비스.
*   **브랜드 정체성**: '설렘부터 근거 있게' - 정직한 AI 여행 가이드.

### 3. [데이터 전략] 근거 중심의 데이터 큐레이션
*   **수집 데이터**: 관광지, 수정요청(1,686건), 기상청 API, 사용자 피드백.
*   **차별화**: 단순 장소 목록이 아닌 '최신성 위험 신호'를 데이터화.

### 4. [기술적 핵심] Trust Engine 아키텍처
*   **시스템 구조**: RAG → Trust Engine → gpt-5-mini의 순차적 결합.
*   **핵심 기술**: LLM의 환각을 방어하는 '테크니컬 배리어'와 6대 신뢰도 루브릭.

### 5. [신뢰성 검증] 정량적 시뮬레이션 결과
*   **Case Study**: 0점이었던 성산일출봉 플랜이 어떻게 100점으로 개선되는가.
*   **데이터의 힘**: 기술적 보정 조치가 신뢰도 점수에 미치는 정량적 영향 증명.

### 6. [데모 및 UX] 판단을 돕는 인터페이스
*   **주요 화면**: 지역 신뢰 대시보드, 신뢰 근거가 포함된 후보 카드.
*   **에이전트**: 모르는 것은 '확인 필요'로 솔직하게 말하는 하루방 에이전트.

### 7. [결론 및 Q&A] 신뢰로 그리는 여행의 미래
*   **마무리 메시지**: 더 많이 추천하기보다, 더 정직하게 가이드하겠습니다.
*   **Q&A 대응**: 기술적 환각 방어 및 사용자 피드백 검증 전략.

### 💡 PPT 제작을 위한 팁
*   **디자인 톤**: 제주의 청량함을 담은 블루/그린 톤과 데이터의 신뢰감을 주는 깔끔한 산세리프 폰트 조합을 권장합니다.
*   **키워드 배치**: 각 슬라이드 상단에 '신뢰도 100점', '1,686건의 수정요청' 등 핵심 수치를 강조하세요.
*   **비주얼 요소**: Trust Engine 구조는 복잡한 표보다 흐름도(Flowchart) 형태의 시각 자료를 활용하는 것이 효과적입니다.

## 🛡️ 심사위원 Q&A 대응 전략 및 모범 답안

### 1. 일반 질문 (Service & Logic)
| 질문 | 답변 포인트 |
| :--- | :--- |
| **기존 여행 추천 서비스와 무엇이 다른가요?** | 단순히 '유명한 곳'을 추천하는 것이 아니라, **'지금 실행 가능한가'**를 데이터로 검증한다는 점이 다릅니다. GPT의 창작 능력을 억제하고 Trust Engine의 검증 능력을 우선시한 구조가 핵심입니다. |
| **공공데이터의 오류는 어떻게 해결하나요?** | 공식 데이터도 변할 수 있음을 인지하고, **'수정 요청 데이터'**와 **'사용자 피드백'**을 실시간으로 결합합니다. 데이터가 상충할 경우 '확인 필요' 상태로 전환하여 정보의 투명성을 유지합니다. |
| **사용자 피드백이 악의적으로 입력되면요?** | 모든 피드백은 즉시 반영되지 않습니다. 내부 검증 레이어에서 GPS 인증, 패턴 분석을 거치며, 신뢰도가 검증된 사용자의 피드백에 더 높은 가중치를 부여하는 알고리즘을 사용합니다. |

### 2. 기술 압박 질문 (Technical Deep-Dive)

#### Q: GPT-5-mini의 환각(Hallucination) 현상을 기술적으로 어떻게 차단했습니까?
**A:** "저희는 LLM을 정보를 찾는 '엔진'이 아니라, 검증된 정보를 전달하는 **'설명자(Speaker)'**로만 활용합니다.
1. **RAG 구조**: LLM이 학습 데이터가 아닌, 저희가 제공한 공공데이터 내에서만 답하도록 Context를 제한합니다.
2. **Trust Engine**: LLM에 데이터를 넘기기 전, 메타데이터 가중치(최신성, 품질, 피드백)를 통해 부적합한 정보를 1차로 필터링합니다.
3. **가드레일**: 근거가 부족한 경우 '모름' 혹은 '확인 필요'로 답변하도록 프롬프트 엔지니어링과 출력 검증 로직을 구현했습니다."

#### Q: RAG의 검색 정확도를 높이기 위한 메타데이터 전략은 무엇인가요?
**A:** "단순 키워드 매칭의 한계를 극복하기 위해 **5가지 핵심 메타데이터 필드**에 가중치를 둡니다.
* `feedback_summary` (최우선): 실제 방문자의 최신 경험 반영
* `suitability_tags`: 날씨/운영시간 등 실시간 상황 적합성
* `public_data_quality`: 원본 데이터의 공식성 등급
이러한 필드들을 벡터 검색의 Re-ranking 과정에 반영하여 사용자의 상황에 가장 정직한 정보를 상단에 배치합니다."

### 3. 비즈니스 및 확장성 (Vision)
| 질문 | 답변 포인트 |
| :--- | :--- |
| **제주도 외 지역으로 확장이 가능한가요?** | 네, 본 시스템은 공공데이터의 '구조'를 활용하므로, 데이터 형식이 유사한 타 지자체의 공공데이터와 API만 연결하면 즉시 수평적 확장이 가능합니다. |
| **수익 모델이나 향후 로드맵은?** | 신뢰도가 검증된 로컬 상권과의 연계, 혹은 여행지 최신 정보를 필요로 하는 타 플랫폼에 '신뢰 데이터(Trust API)'를 제공하는 B2B 모델을 고려하고 있습니다. |

### 💡 발표 전 최종 체크리스트
- [ ] **수치 강조**: "1,686건의 데이터 수정 사례", "0점에서 100점의 신뢰도 상승" 수치를 언급했는가?
- [ ] **기술 용어**: RAG, Trust Engine, gpt-5-mini의 상관관계를 명확히 설명했는가?
- [ ] **태도**: "완벽한 AI"가 아닌 "모르는 것은 확인이 필요하다고 말하는 정직한 AI"임을 강조했는가?

## 📊 [발표 시각 자료 제안 1] Trust Engine 아키텍처 다이어그램

**슬라이드 목적**: 'gpt-5-mini' 모델을 사용하면서도 어떻게 높은 신뢰성을 확보했는지 기술적 구조로 증명

```mermaid
graph LR
    A[사용자 입력] --> B(구조화 및 필터링)
    B --> C{Trust Engine}
    subgraph "Technical Barrier"
    C --> D[공공데이터 RAG]
    C --> E[실시간 API 검증]
    C --> F[사용자 피드백 루프]
    end
    D & E & F --> G(신뢰도 점수 산정)
    G --> H[gpt-5-mini: 근거 기반 설명]
    H --> I[최종 추천/안내]
```

**시각화 포인트:**
*   **Technical Barrier**: LLM이 직접 답하기 전, 데이터와 규칙이 먼저 판단하는 '방어막' 구조 강조

### 1. [도입] 정보의 양보다 중요한 것은 '판단 기준'입니다.

**[시각 자료: 문제 정의 슬라이드]**

여러분, 제주 여행을 준비할 때 정보가 부족해서 힘드신가요? 아닙니다. 정보는 너무나 많습니다. 하지만 그 속에서 우리는 질문합니다. **"오늘, 지금, 이 장소를 정말 믿고 가도 될까?"**

저희 '제주를 담다'는 단순히 유명한 곳을 나열하는 서비스가 아닙니다. 정보 과잉 시대에 여행자가 **'믿고 실행할 수 있는 판단 기준'**을 제공하는 신뢰 기반 AI 여행 플랜 서비스입니다.

### 2. [기술 핵심] Trust Engine: 데이터가 판단하고 AI는 설명합니다.

**[시각 자료: 제안 1 - Trust Engine 아키텍처 다이어그램]**

저희 서비스의 핵심은 **'Trust Engine'**이라는 기술적 방어막입니다. 보통의 AI 서비스는 LLM이 먼저 답을 내놓지만, 저희는 다릅니다.

1. **RAG 엔진**이 공공데이터와 실시간 API에서 '팩트'를 먼저 수집합니다.
2. **Trust Engine**이 이 데이터를 바탕으로 공식성, 최신성, 상황 적합성 등 다차원 지표를 통해 신뢰도를 점수화합니다.
3. **gpt-5-mini**는 마지막 단계에서, 오직 검증된 근거 안에서만 사용자에게 친절하게 설명하고 조율하는 역할을 수행합니다.

### 3. [성과 입증] 0점이 100점이 되는 신뢰의 과학

**[시각 자료: 제안 2 - 신뢰도 개선 시뮬레이션 비교]**

이 구조가 실제로 작동하는지 수치로 증명했습니다. 초기 성산일출봉 플랜은 정보 불일치와 날씨 부적합으로 **0점**에 불과했습니다.

하지만 저희 시스템이 실시간 정보를 동기화하고, 사용자 피드백을 반영하며, 운영 시간을 보정하자 점수는 단계적으로 상승하여 **최종 100점**에 도달했습니다. 이는 저희 루브릭이 실제 여행의 품질을 정량적으로 관리할 수 있음을 보여주는 명확한 증거입니다.

### 4. [서비스 철학] 사용자의 발자취가 다음 여행자의 지도가 됩니다.

**[사용자 흐름 및 피드백 루프 슬라이드]**

저희는 '정직한 AI'를 지향합니다. 데이터가 부족하면 억지로 추천하지 않고 **'확인 필요'**라고 솔직하게 고지합니다. 그리고 사용자가 남긴 소중한 방문 기록은 내부 검증 레이어를 거쳐 다음 여행자를 위한 강력한 **'신뢰 신호'**로 재탄생합니다.

### 5. [맺음말] 설렘부터 근거 있게, 제주를 담다.

**[시각 자료: 제안 3 - 핵심 홍보 카피 슬라이드]**

'제주를 담다'는 더 많이 추천하는 서비스가 아닙니다. 공공데이터로 계획하고, 여러분의 경험으로 다시 좋아지는 **순환형 신뢰 서비스**입니다.

이제 제주 여행, 막연한 설렘을 넘어 확실한 근거를 가지고 떠나십시오. 저희 '제주를 담다'가 함께하겠습니다. 감사합니다.

## 심사위원 예상 질문에 대한 답변

### 1. 프로젝트의 문제 정의가 명확한가요?

**답변:** 네, 저희 '제주를 담다'는 제주 여행 정보가 많음에도 불구하고, 사용자들이 '오늘 이 장소를 믿고 선택해도 되는지'에 대한 명확한 근거, 즉 **'믿고 실행할 판단 기준의 부족'**을 핵심 문제로 정의했습니다.

단순히 유명한 장소 이름을 아는 것을 넘어, **실제 일정 안에서 움직일 수 있는지, 오늘 가도 괜찮은지, 후회하지 않을지**에 대한 실질적인 판단 근거가 부족하다는 점에 주목했습니다. 저희 서비스는 '제주에서 좋은 곳을 알려주자'가 아닌, **'사용자가 믿고 실행할 수 있는 여행 판단을 돕자'**는 명확한 목표를 가지고 있습니다. 이 문제 정의는 저희 프로젝트의 데이터 수집, RAG 구조, 신뢰도 계산, 방문 기록 설계 등 모든 요소에 연결되어 있습니다.

### 2. 공공데이터는 어떻게 활용되며, 신뢰성 확보 방안은 무엇인가요?

**답변:** 공공데이터는 '제주를 담다' 서비스의 가장 기본적인 신뢰 근거입니다. 저희는 공공데이터를 활용하여 여행 플랜을 생성하고, RAG 엔진을 통해 추천의 근거와 출처를 명확하게 제시합니다.

신뢰성 확보를 위해 다음과 같은 방안을 마련했습니다:

*   **RAG 엔진 활용**: RAG 엔진이 공공데이터를 기반으로 근거를 검색하고, 이를 LLM에 전달하여 정확하고 근거 있는 답변 및 추천을 생성합니다.
*   **데이터 선별 원칙**: 데이터를 많이 모으기보다 '여행자가 실제 판단에 쓸 수 있는' 출처가 명확한 공공데이터만을 선별하여 수집하고 구조화합니다.
*   **수정요청 데이터 반영**: 비짓제주 수정요청 데이터를 통해 공식 데이터도 바뀔 수 있다는 점을 인지하고, 정보의 최신성 위험을 판단하는 신호로 활용합니다. 출처, 최신성, 수정 이력, 확인 필요 여부 등을 함께 보며 현재 유효한 정보인지 판단합니다.
*   **사용자 피드백과의 연동**: 공공데이터의 빈틈을 사용자 피드백(방문 기록)으로 보완합니다. 사용자 피드백이 검증 과정을 거쳐 신뢰도 점수에 반영되고 공공데이터 개선 신호로 연결되는 선순환 구조를 통해 신뢰도를 지속적으로 개선합니다.

### 3. gpt-5-mini 활용 방식 및 역할 제한에 대해 설명해 주세요.

**답변:** 저희 '제주를 담다'는 gpt-5-mini를 '자유로운 추천 모델'로 두지 않고, 그 역할에 엄격한 제한을 두었습니다. 핵심은 **LLM이 모든 것을 결정하지 않도록 하는 것**입니다.

*   **'생성'이 아닌 '검증 기반 설명' 역할**: gpt-5-mini는 새로운 정보를 창작하거나 추천을 임의로 생성하지 않습니다. 오직 Trust Engine에서 '사전에 엄격히 검증되고 필터링된 근거 데이터'만을 입력받아 사용자에게 '설명'하고 '조율'하는 역할을 수행하도록 코드 레벨에서 명확히 제한했습니다.
*   **RAG 및 Trust Engine 선행**: 사용자 입력이 들어오면 RAG 엔진이 공공데이터와 외부 API에서 필요한 근거를 먼저 조회하고, Trust Engine이 이 정보를 바탕으로 신뢰도를 판단하며 후보를 걸러냅니다. gpt-5-mini는 이 과정 이후에 투입됩니다.
*   **불확실성 명시**: 모르는 내용은 그럴듯하게 채우지 않고 **'확인 필요'**로 명확히 분리하여, 사용자가 정보의 한계를 인지할 수 있도록 합니다.

### 4. RAG/Trust Engine은 어떤 역할을 하며, 어떻게 작동하나요?

**답변:** RAG와 Trust Engine은 '제주를 담다'의 핵심 기술이자 신뢰도 확보의 중추적인 역할을 합니다.

*   **RAG (Retrieval-Augmented Generation) 엔진**: 사용자 쿼리 및 여행 플랜 컨텍스트를 분석하여, 공공데이터와 외부 API에서 `후보 장소, 출처, 근거 문서, 이동 및 주차 정보, 수정 요청 이력, 방문 피드백 신호` 등 여행 판단에 필요한 다각적인 근거 데이터를 조회합니다. LLM이 '상상'이 아닌 '팩트'에 기반하여 작동하도록 하는 1차적인 안전장치 역할을 합니다.

*   **Trust Engine (신뢰 판단 엔진)**: RAG를 통해 조회된 정보를 기반으로 `신뢰도 점수, 확인 필요 여부, 배제 조건`을 정교하게 계산합니다. 이 과정에서 `공공데이터의 품질, 정보 최신성, 사용자 피드백(긍정/부정), 실시간 날씨 및 운영시간과의 적합성, 정보 불일치 여부` 등 다양한 신뢰도 루브릭 평가 항목들이 종합적으로 고려됩니다. Trust Engine은 **'가도 될 만한 장소'를 선별하고, '가서는 안 될 장소'를 미리 걸러내는 의사결정의 핵심**입니다. 최종적으로 gpt-5-mini에 전달될 정보를 필터링하고 신뢰도를 평가하는 역할을 합니다.

### 5. 서비스의 검증 가능성은 어떻게 확보되나요?

**답변:** 저희 서비스는 정보의 투명성과 판단 기준의 명확성을 통해 검증 가능성을 확보합니다.

*   **설명 가능한 신뢰도 점수**: 신뢰도 점수는 단순한 숫자가 아니라, **각 평가 항목별 점수(공공데이터 품질, 최신성, 피드백 신호, 날씨 적합성 등)를 통해 왜 그런 점수가 나왔는지 설명 가능한 구조**로 제공됩니다. 예를 들어, '성산일출봉'의 0점은 낮은 공공데이터 품질, 사용자 피드백 불일치, 날씨 부적합 등 구체적인 감점 요인으로 설명됩니다.
*   **'확인 필요' 명시**: 정보가 불확실하거나 부족할 경우, 억지로 추천하지 않고 **'확인 필요'**로 명확히 분리하여 사용자에게 정보의 한계를 투명하게 고지합니다.
*   **사용자 피드백 검증 프로세스**: 사용자의 방문 기록은 즉시 반영되지 않고, 내부 대시보드에서 '검증 완료', '검증 필요', '반영 제외'로 분류됩니다. AI 기반 자동 검증 및 운영자 수동 검토를 통해 정확성을 검증하며, **검증된 신호만** 신뢰도 점수 업데이트에 반영됩니다.

### 6. 데모의 견고성은 어느 정도로 구현되었나요?

**답변:** 저희 '제주를 담다'는 서비스의 견고성을 중요한 가치로 여기며 설계되었습니다. 특히, 외부 API나 데이터 부족 시에도 서비스 흐름이 끊기지 않도록 하는 데 중점을 둡니다.

*   **대체 UI 및 안내 문구**: 지도, 날씨 등 외부 호출이 실패하더라도 **대체 UI와 안내 문구로 흐름을 유지**합니다. 사용자 경험이 중단되지 않도록 하여 서비스의 안정성을 확보합니다.
*   **'확인 필요'로의 분리**: 정보가 부족한 후보는 억지로 추천하지 않고 **'확인 필요'로 명확히 분리하여 보여줍니다.** 이는 정보의 정직함을 지키면서도 사용자에게 합리적인 판단을 돕는 역할을 합니다.
*   **신뢰도 기반 정보 분리**: 지역별 신뢰 대시보드에서는 확인 가능한 후보와 확인이 필요한 후보를 명확하게 분리하여 보여줌으로써, 데이터가 부족한 조합에 대해 억지로 추천하지 않고 투명하게 정보를 제공합니다.

### 7. 개인 정보 및 데이터 보안은 어떻게 관리되고 있나요?

**답변:** 개인 정보 및 데이터 보안은 매우 중요하게 다루고 있으며, 다음과 같은 방식으로 관리됩니다.

*   **API 키 관리**: API 키는 코드에 직접 노출하지 않고 **환경변수와 Secrets Manager**를 통해 안전하게 관리합니다. 이는 무단 접근 및 오용을 방지합니다.
*   **민감 정보 최소화**: 사용자 방문 피드백 등에서 민감 정보는 최소화하여 수집하고, 수집된 정보는 **내부 신호로만 처리**됩니다. 외부로 유출되거나 오용될 가능성을 줄입니다.
*   **사용자 입력 구조화**: 사용자 입력은 구조화하여 안전하게 처리되며, 데이터 처리 과정에서 보안 취약점이 발생하지 않도록 설계되었습니다.
*   **클라우드 플랫폼 활용**: Google Cloud Platform(GCP)의 보안 기능을 활용하여 인프라 전반의 보안을 강화합니다. 예를 들어, Cloud SQL을 통한 데이터베이스 관리 등이 포함됩니다.

### 8. 향후 서비스 확장성 계획은 무엇인가요?

**답변:** 저희 '제주를 담다'는 MVP에서 핵심적인 신뢰 순환 구조를 증명하는 데 주력했으며, 향후 서비스 확장성 계획은 다음과 같습니다.

*   **단기 로드맵 (MVP 이후 3개월)**:
    *   **Trust Engine 고도화**: 더 많은 `suitability_tags` (동반자, 이동수단 등) 및 `discrepancy_status` (불일치 유형별)를 신뢰도 점수 산정에 정교하게 반영하고, 가중치를 동적으로 조절하는 시스템을 구축합니다.
    *   **하루방 에이전트 개인화**: 사용자 프로필 데이터(예: 이전 여행 스타일, 선호도)를 Trust Engine에 연동하여 하루방 에이전트의 설명을 더욱 개인화하고 맞춤형 대안을 제시할 수 있도록 발전시킵니다.
    *   **자동 검증 시스템 도입**: 머신러닝 기반의 자동 검증 모듈을 개발하여 대량의 사용자 피드백을 효율적으로 분류하고, 허위 피드백을 사전에 탐지하는 정확도를 높입니다.

*   **중장기 로드맵 (MVP 이후 6개월 이상)**:
    *   **데이터 확장**: 숙소, 음식점, 교통, 축제 등 제주 내 다른 카테고리의 공공데이터를 연동하고 Trust Engine에 통합합니다. 이를 통해 통합적인 여행 플랜을 제공할 수 있도록 확장합니다.
    *   **실시간 데이터 연동 강화**: 지도 API를 활용한 실시간 교통 정보, 이벤트 API를 통한 지역 축제 정보 등 실시간 변동성이 큰 데이터를 Trust Engine에 실시간으로 반영하여 '실행 가능성' 점수를 더욱 정교화합니다.
    *   **타 지역 확장**: 제주도에서의 성공적인 모델을 바탕으로 전국 단위의 공공데이터 및 지역 특성을 반영한 Trust Engine을 구축하여 서비스 지역을 확대합니다.

### 9. 신뢰 점수 산정의 실제 산식과 로그가 명확하게 설명될 수 있나요?

**답변:** 네, 저희 신뢰도 점수 산정은 단순히 추상적인 '별점'이 아닌, **명확한 평가 항목과 감점 로직**을 기반으로 하며, 그 과정이 상세하게 설명 가능합니다.

*   **초기 100점 시작**: 모든 플랜은 100점에서 시작합니다.
*   **주요 평가 항목**: 다음 항목들을 기준으로 감점이 이루어집니다.
    *   **공공데이터 기반 근거 제시율**: RAG 엔진을 통해 검색된 공공데이터의 양, 질, 관련성 (예: 품질 '저'에서 -20점)
    *   **사용자 피드백 신호 반영**: '정보 불일치' 피드백은 가장 큰 감점 요인 (-30점), '방문하지 않음'도 감점.
    *   **실행 가능성 조건 적합성**: 날씨 (-25점), 운영시간 (-20점), 이동수단/거리, 동행자 조건 등 플랜과 맞지 않을 경우 감점.
    *   **정보 불일치 탐지 및 경고**: 공공데이터와 실시간 정보 불일치 시 크게 감점 (-35점).
    *   **정보 최신성**: 정보가 '오래됨'으로 판단되면 감점 (-15점).
*   **상세 평가 및 로그**: 점수 변화와 그 이유가 `reasons` 리스트에 기록되어 최종 점수와 함께 제공됩니다. 예를 들어, '협재 해변'이 65점인 이유는 '공공데이터 기반 근거 제시율'에서 10점, '날씨 적합성'에서 25점이 감점되었기 때문입니다.

이처럼 각 항목별로 명확한 감점 기준이 존재하며, 이를 통해 각 여행 플랜의 신뢰도가 **왜 그런 점수가 나왔는지 투명하게 설명할 수 있습니다.**

### 10. 공공데이터에 오류가 발생했을 때의 보정 구조는 마련되어 있나요?

**답변:** 네, 공공데이터의 오류 가능성을 인지하고 있으며, 이를 보정하기 위한 다단계 구조를 마련했습니다.

*   **수정요청 데이터 활용**: '비짓제주 수정요청 데이터'와 같이 공공데이터 자체의 변경 이력을 추적합니다. 이는 단순한 관리 기록이 아니라, 정보의 최신성 위험을 판단하는 신호로 활용됩니다. RAG 엔진은 이 정보를 바탕으로 LLM에게 근거를 제시합니다.
*   **사용자 피드백 통한 감지**: 사용자의 실제 방문 기록 중 '정보 불일치' 피드백을 가장 중요한 오류 감지 신호로 활용합니다. 이 피드백은 신뢰도 점수를 크게 하락시키는 요인으로 작용합니다.
*   **내부 피드백 대시보드 및 검증**: '정보 불일치' 피드백은 즉시 반영되지 않고 내부 피드백 대시보드로 유입되어 '검증 필요'로 분류됩니다. 이후 AI 기반 자동 검증 시스템과 필요시 운영자의 수동 검토를 통해 **사진, GPS 위치 정보, 시간 스탬프** 등 다른 데이터 소스와 교차 대조하여 진위를 판단합니다.
*   **검증된 신호만 반영**: 검증이 완료된 신호에 한해서만 신뢰도 점수 업데이트 및 공공데이터 반영 큐(queue)로 연결됩니다. 이를 통해 공공데이터의 오류가 발견될 경우, **사용자 피드백과 시스템 검증을 통해 보정하고 다음 여행자를 위한 신뢰도를 높이는 순환 구조**를 가집니다.

## 추가적인 반박 질문에 대한 답변

### 1. GPT-5-mini의 역할 제한에도 불구하고 LLM의 본질적인 한계(예: 환각 현상)가 발생할 가능성은 없습니까? 이를 기술적으로 어떻게 완벽하게 방어하고 있으며, 구체적인 비상 대응 메커니즘은 무엇인가요?

**답변:** 저희는 GPT-5-mini의 본질적 한계를 누구보다 깊이 이해하고 있으며, 이를 완벽히 '방어'한다기보다는 **'근거 없는 답변 생성을 원천 차단하고, 모든 정보의 신뢰성을 기술적으로 검증하는 구조'**를 핵심 전략으로 삼았습니다. 이는 단순히 가능성을 최소화하는 것을 넘어, **공고한 아키텍처를 통해 환각 현상 자체가 발생할 여지를 차단**하는 데 주력합니다.

1.  **'생성'이 아닌 '검증 기반 설명' 역할의 하드코딩**: GPT-5-mini는 새로운 정보를 창작하거나 추천을 임의로 생성하도록 설계되지 않았습니다. 오직 **Trust Engine에서 '사전에 엄격히 검증되고 필터링된 근거 데이터'**만을 입력받아 사용자에게 '설명'하고 '조율'하는 역할을 수행하도록 **코드 레벨에서 그 역할이 명확히 제한**되어 있습니다. RAG 시스템은 `public_data_quality`, `last_verified_date`, `discrepancy_status` 등의 핵심 메타데이터를 활용해 신뢰성 높은 근거를 선제적으로 확보합니다.
2.  **RAG-Trust Engine 다단계 필터링 아키텍처**: 사용자의 쿼리는 즉시 LLM으로 전달되지 않고, **'하드 테크니컬 배리어(Hard Technical Barrier)'** 역할을 하는 **RAG 엔진**을 통해 신뢰할 수 있는 공공데이터와 외부 API 근거를 먼저 조회합니다. 이후 **Trust Engine이 이 조회된 근거들의 신뢰도를 다각도로 판단하고 필터링**합니다. LLM은 오직 이 '필터링되고 신뢰성이 입증된 근거'만을 최종 입력으로 받아 답변을 구성합니다. 근거가 부족하면 답변을 생성하지 않습니다. Trust Engine은 `public_data_quality`, `last_verified_date`, `feedback_summary`, `suitability_tags`, `discrepancy_status`와 같은 다차원적인 신뢰도 평가 지표들을 종합적으로 계산하여 후보 장소들을 정교하게 필터링합니다.
3.  **코드 레벨 가드레일 및 투명한 고지 프로토콜**: 만약 LLM이 할당된 근거 범위를 벗어나는 답변을 시도할 경우, 시스템 내부에 구현된 **'가드레일(Guardrails)'이 즉시 해당 답변 생성을 차단**합니다. 사용자에게는 '현재 정보로는 정확한 답변 제공이 어렵습니다'와 같은 **명확하고 중립적인 메시지를 제공**합니다. Trust Engine이 `discrepancy_status` 메타데이터를 통해 '확인 필요'로 분류한 정보는 LLM 또한 이를 답변에 명확히 명시하여 정보의 불확실성을 투명하게 전달하는 프로토콜을 따릅니다.

저희는 **Ragas 평가 지표**를 통한 지속적인 모니터링을 넘어, 이러한 **구조적, 코드 레벨의 방어 메커니즘**을 통해 LLM의 한계를 가장 견고하게 관리하고 있으며, 이는 **'제주를 담다'의 핵심 기술적 신뢰성**을 담보하는 핵심 요소입니다.

### 2. 사용자 피드백 검증 과정의 효율성과 공정성을 어떻게 담보할 수 있으며, 만약 대량의 의도적인 허위 피드백이 발생할 경우 이 '검증 레이어'가 얼마나 견고하게 작동할 수 있습니까?

**답변:** 사용자 피드백은 저희 서비스의 '신뢰 순환'에서 가장 중요한 축이지만, 동시에 데이터 오염의 위험이 가장 큰 영역임을 명확히 인지하고 있습니다. 저희는 **다단계 검증 프로세스**와 **이상 탐지 메커니즘**을 통해 효율성, 공정성, 그리고 허위 피드백에 대한 견고성을 확보합니다.

1.  **내부 피드백 대시보드와 교차 검증**: 모든 피드백은 '검증 완료', '검증 필요', '반영 제외'로 분류되는 내부 대시보드를 거칩니다. '검증 필요' 피드백은 AI 기반 자동 검증 시스템과 필요시 운영자의 수동 검토를 통해 **사진, GPS 위치 정보, 시간 스탬프** 등 다른 데이터 소스와 교차 대조하여 진위를 판단합니다. 특히 `discrepancy_status` 메타데이터와 피드백을 비교하여 정보 불일치 여부를 정밀하게 확인합니다.
2.  **허위 피드백 방어 시스템**:
    *   **패턴 분석 및 자동 차단**: 비정상적인 패턴의 피드백이 감지될 경우, 해당 계정이나 피드백 소스를 **자동으로 '반영 제외' 및 경고 조치**합니다.
    *   **가중치 차등 부여**: 피드백 제공자의 **활동 이력과 신뢰도**에 따라 피드백에 부여하는 가중치를 차등화합니다. 악의적 의도의 피드백은 RAG 검색 시 `feedback_summary` 메타데이터가 미치는 영향이 최소화되도록 설계되어 있습니다.

이러한 체계적인 과정을 통해 저희는 RAG 검색 시 `feedback_summary` 메타데이터가 가장 높은 가중치를 가질 수 있도록, **가장 신뢰할 수 있는 사용자 피드백만을 시스템에 반영하며 데이터 오염을 적극적으로 방어**합니다.

### 3. 제시된 아키텍처(RAG, Trust Engine, 하루방 에이전트 등)와 기능들 중 현재 MVP 단계에서 실질적으로 구현되어 작동하는 범위는 어디까지이며, 나머지 고도화될 기능들은 향후 어떤 구체적인 로드맵과 마일스톤으로 발전시켜 나갈 계획이신가요?

**답변:** 저희 'Pack Your Jeju'는 **'핵심 신뢰도 순환 구조'의 MVP 구현**에 집중했습니다. 현재 MVP 단계에서 실질적으로 작동하는 범위는 다음과 같습니다.

1.  **공공데이터 기반 RAG 엔진**: 비짓제주 관광지 데이터 등을 기반으로 `public_data_quality`, `last_verified_date` 메타데이터를 활용하여 LLM에 근거를 제공하는 RAG 엔진이 구현되어 있습니다.
2.  **Trust Engine (핵심 로직)**: 공공데이터 품질, 최신성, 사용자 피드백 신호(`feedback_summary`) 등을 반영하여 신뢰도를 계산하고 '확인 필요' 상태(`discrepancy_status`)를 분류하는 **기본 로직**이 구현되었습니다.
3.  **하루방 에이전트 (근거 기반 설명)**: Trust Engine에서 필터링된 정보를 바탕으로 사용자에게 '설명'하는 `gpt-5-mini` 기반의 기능이 구현되어, 근거 안에서만 조율하고 모르는 것은 '확인 필요'로 명시합니다.
4.  **사용자 피드백 수집 및 내부 대시보드 (수동 검증)**: 사용자의 방문 기록을 수집하고, 내부 대시보드를 통해 운영자가 수동으로 검토할 수 있는 초기 검증 프로세스가 작동합니다.

**저희의 고도화 로드맵은 다음과 같습니다.**

*   **단기 (MVP 이후 3개월)**:
    *   **Trust Engine 고도화**: `suitability_tags` (동반자, 이동수단 등) 및 `discrepancy_status` (불일치 유형별)를 신뢰도 점수 산정에 정교하게 반영하고, 가중치를 동적으로 조절하는 시스템을 구축할 것입니다.
    *   **자동 검증 시스템 도입**: 머신러닝 기반 모듈을 개발하여 허위 피드백 탐지 정확도를 높일 계획입니다.
*   **중장기 (MVP 이후 6개월 이상)**:
    *   **데이터 확장**: 숙소, 음식점, 교통 등 다른 카테고리의 공공데이터를 통합하고, **실시간 데이터 연동을 강화**하여 '실행 가능성'을 더욱 정교화할 것입니다.
    *   **타 지역 확장**: 제주도 모델을 기반으로 전국 단위 서비스 확장을 목표하고 있습니다.

저희는 MVP에서 **신뢰 순환 구조를 기술적으로 증명**하는 데 집중했으며, 이 로드맵을 통해 서비스의 기능과 신뢰도를 지속적으로 고도화하여 **가장 믿을 수 있는 AI 여행 플랜 서비스**로 발전시켜 나갈 것입니다.

### Pack Your Jeju: 'Trust Engine' 핵심 전략

'Pack Your Jeju' 프로젝트의 'Trust Engine'은 제주 여행 정보의 신뢰도를 극대화하기 위한 다음 다섯 가지 핵심 전략을 기반으로 작동합니다. 이는 서비스의 foundational principle이자, 모든 기술적 구현의 근간이 됩니다.

#### 1. 공공데이터 기반의 근거 제시 및 RAG 엔진 활용
*   **원칙**: AI가 추천하는 모든 장소 및 활동에 대한 **명확한 근거와 출처를 제시**하여, 'AI가 왜 이 장소를 추천했는지 알기 어려움'이라는 문제점을 해결합니다.
*   **메커니즘**: `RAG (Retrieval-Augmented Generation)` 엔진을 활용하여 벡터 데이터베이스에서 관련 공공데이터 문서를 검색하고, 이를 LLM에 전달하여 근거 기반 답변 및 추천을 생성합니다.

#### 2. 사용자 피드백 루프를 통한 신뢰도 업데이트
*   **원칙**: 사용자의 **실제 방문 기록(방문함, 방문하지 않음, 변경함, 정보 불일치)**을 '신뢰 신호'로 시스템에 반영하여 데이터의 신뢰도를 지속적으로 개선하고, '다녀온 뒤의 경험이 다음 추천에 반영되지 않음' 문제를 해결합니다.
*   **메커니즘**: '신뢰도 루브릭'의 핵심 평가 항목에 사용자 피드백 신호를 포함시켜, 특히 '정보 불일치'와 같은 부정적 피드백은 신뢰도를 크게 하락시키는 요인으로 작용하게 합니다.

#### 3. '하루방 에이전트'를 통한 개인화된 정보 해석 및 실행 가능성 반영
*   **원칙**: LLM (`Google Gemini API`) 기반의 '하루방 에이전트'가 사용자의 특정 조건(날씨, 동반자, 렌터카 등)에 맞춰 공공데이터를 해석하고, 개인화된 설명을 생성하여 '비·바람·운영시간·이동거리 때문에 일정 실패 가능성' 및 '공공데이터의 맥락 부족' 문제를 해결합니다.
*   **메커니즘**: 여행 플랜의 '실행 가능성 조건 적합성'을 평가 항목으로 두어, 날씨, 운영 시간, 이동 수단/거리, 동행자 조건 등이 플랜에 얼마나 적합한지를 점수에 반영합니다.

#### 4. 정보 불일치 탐지 및 경고
*   **원칙**: 공공데이터와 실시간 정보(날씨, 지도 API 등) 간에 잠재적인 불일치가 발생할 경우 이를 **탐지하고 사용자에게 경고**하여, 사용자에게 정확하지 않은 정보로 인한 불편을 최소화합니다.
*   **메커니즘**: '정보 불일치 탐지 및 경고'를 신뢰도 루브릭의 주요 감점 요인으로 설정하여, 정보의 정확성을 확보합니다.

#### 5. 정보 최신성 유지 관리
*   **원칙**: 제공되는 공공데이터 및 기타 정보의 **업데이트 주기와 최신 상태를 유지**하는 것을 중요한 신뢰도 요소로 관리하여 정보의 유효성을 보장합니다.
*   **메커니즘**: '정보 최신성'을 신뢰도 루브릭의 감점 요인으로 두어, 오래된 정보가 추천 신뢰도를 하락시키는 것을 방지합니다.

### [발표 방어] 05_evaluation_mapping.pdf 기반 심사위원 Q&A 스크립트

이 스크립트는 문서에서 정의된 **5개 핵심 평가 축**을 기준으로 심사위원의 압박 질문에 대응하기 위해 작성되었습니다.

#### 1. 출처 및 신뢰성 (Source & Reliability)
**질문:** "데이터의 신뢰성을 어떻게 보장하나요? 공공데이터도 오류가 있을 수 있는데 LLM이 이를 그대로 믿고 답변하는 것 아닌가요?"

**답변:**
> "저희는 데이터의 '공식성'을 최우선으로 합니다. 비짓제주(Visit Jeju)나 기상청과 같은 **검증된 공공 API**만을 1차 소스로 활용하며, LLM이 임의로 정보를 생성(Hallucination)하지 못하도록 **RAG(Retrieval-Augmented Generation) 구조**를 채택했습니다. 또한, 문서에 명시된 평가 기준에 따라 데이터 품질(Quality)을 '최상/상/중/하'로 분류하여, 신뢰도가 낮은 데이터는 사용자에게 추천 시 별도의 가중치를 낮추거나 경고를 표시하도록 설계했습니다."

#### 2. 최신성 및 이력 관리 (Recency & History)
**질문:** "제주도는 폐업이나 영업시간 변경이 잦은데, 데이터가 최신이라는 것을 어떻게 확신하나요?"

**답변:**
> "문서의 핵심 평가 항목인 **'Recency & History'** 전략을 적용했습니다. 단순히 현재 데이터를 보여주는 것에 그치지 않고, 해당 장소의 **'마지막 검증일(Last Verified Date)'**과 **'수정 요청 이력'**을 사용자에게 함께 공개합니다. 비짓제주의 수정 요청 데이터를 트래킹하여 변경이 잦은 장소는 신뢰도 점수를 감점하고, 사용자에게 '최신 정보 확인 필요'라는 시각적 신호를 즉각적으로 제공하여 정보의 유효성을 보장합니다."

#### 3. 상황 적합성 (Suitability)
**질문:** "기존 여행 서비스와 차별화된 '개인화'의 핵심은 무엇인가요?"

**답변:**
> "저희의 차별점은 **'실시간 실행 가능성'**입니다. 평가 지표의 'Suitability' 항목에 따라 사용자의 동반자 유형, 이동수단뿐만 아니라 **실시간 기상 조건(날씨)**을 결합합니다. 예를 들어, 강풍이 부는 날 야외 오름 일정을 추천하는 오류를 범하지 않도록 Trust Engine이 적합성을 판단합니다. 단순히 '좋은 곳'이 아니라 **'지금 당신이 가기에 가장 안전하고 적합한 곳'**을 추천하는 것이 저희의 목표입니다."

#### 4. 피드백 루프 및 투명성 (Feedback Loop & Transparency)
**질문:** "사용자 피드백이 악의적으로 입력되거나 왜곡될 가능성은 없나요?"

**답변:**
> "그 점을 방지하기 위해 **'투명성(Transparency)' 원칙**을 고수합니다. 모든 사용자 피드백은 즉시 반영되는 것이 아니라, 데이터 불일치(Discrepancy) 탐지 로직을 거칩니다. 다수의 사용자로부터 정보가 다르다는 리포트가 누적될 경우, 해당 정보는 즉시 **'Verification Needed(확인 필요)'** 상태로 전환됩니다. 모르는 것을 아는 척하지 않고, 불확실한 정보는 불확실하다고 솔직하게 표기하는 것이 저희 서비스가 신뢰를 얻는 방식입니다."

## Summary:

### Q&A

1.  **프로젝트의 문제 정의가 명확한가요?**
    네, '제주를 담다'는 제주 여행 정보의 과잉 속에서 사용자들이 '믿고 실행할 판단 기준의 부족'을 핵심 문제로 정의했습니다. 유명 장소 추천을 넘어, '사용자가 믿고 실행할 수 있는 여행 판단을 돕자'는 명확한 목표를 가지고 있습니다.

2.  **공공데이터는 어떻게 활용되며, 신뢰성 확보 방안은 무엇인가요?**
    공공데이터는 서비스의 가장 기본적인 신뢰 근거입니다. RAG 엔진을 통해 근거를 제시하고, '여행자가 실제 판단에 쓸 수 있는' 출처 명확한 데이터만 선별합니다. 특히 비짓제주 수정요청 데이터를 활용하여 정보의 최신성 위험을 판단하며, 사용자 피드백과 연동하여 데이터의 신뢰도를 지속적으로 개선하는 선순환 구조를 가집니다.

3.  **gpt-5-mini 활용 방식 및 역할 제한에 대해 설명해 주세요.**
    gpt-5-mini는 '생성'이 아닌 '검증 기반 설명' 역할에 엄격히 제한됩니다. Trust Engine에서 사전에 검증된 근거 데이터만을 입력받아 설명하고 조율하며, 근거가 부족하면 '확인 필요'로 명확히 분리하여 불확실성을 투명하게 알립니다.

4.  **RAG/Trust Engine은 어떤 역할을 하며, 어떻게 작동하나요?**
    RAG 엔진은 공공데이터와 외부 API에서 `후보 장소, 출처, 근거 문서, 이동 및 주차 정보, 수정 요청 이력, 방문 피드백 신호` 등 다각적인 근거 데이터를 조회하여 LLM이 '팩트'에 기반하도록 합니다. Trust Engine은 조회된 정보를 바탕으로 `신뢰도 점수, 확인 필요 여부, 배제 조건`을 정교하게 계산하여 '가도 될 만한 장소'를 선별하고 '가서는 안 될 장소'를 미리 걸러내는 의사결정의 핵심 역할을 합니다.

5.  **서비스의 검증 가능성은 어떻게 확보되나요?**
    신뢰도 점수는 단순히 숫자가 아니라, 각 평가 항목별 점수를 통해 '왜 그런 점수가 나왔는지' 설명 가능한 구조입니다. 정보가 불확실할 경우 '확인 필요'로 명확히 명시하며, 사용자 피드백은 내부 대시보드에서 AI 기반 자동 검증 및 운영자 수동 검토를 거쳐 검증된 신호만 반영됩니다.

6.  **데모의 견고성은 어느 정도로 구현되었나요?**
    외부 API 실패 시 대체 UI와 안내 문구로 흐름을 유지하고, 정보 부족 시 '확인 필요'로 명확히 분리하여 보여줍니다. 신뢰도 기반으로 정보를 분리하여 억지 추천을 하지 않음으로써 서비스의 안정성과 정직함을 확보합니다.

7.  **개인 정보 및 데이터 보안은 어떻게 관리되고 있나요?**
    API 키는 환경변수와 Secrets Manager로 관리하며, 사용자 피드백에서 민감 정보는 최소화하여 내부 신호로만 처리합니다. 사용자 입력은 구조화하여 안전하게 처리하고, GCP의 보안 기능을 활용하여 인프라 전반의 보안을 강화합니다.

8.  **향후 서비스 확장성 계획은 무엇인가요?**
    단기적으로 Trust Engine 고도화(suitability_tags, discrepancy_status), 하루방 에이전트 개인화, 자동 검증 시스템 도입을 계획하고 있습니다. 중장기적으로는 숙소, 음식점, 교통 등 다른 카테고리 데이터 확장 및 실시간 정보 연동 강화, 그리고 타 지역으로의 서비스 확장을 목표로 합니다.

    네, 모든 플랜은 100점에서 시작하며, 공공데이터 기반 근거 제시율, 사용자 피드백, 날씨 적합성, 운영시간 일치, 정보 불일치 탐지 및 경고, 정보 최신성 등 6가지 주요 평가 항목에 따라 명확한 감점 로직이 적용됩니다. 각 항목별 감점 내용과 이유가 상세히 기록되어 투명하게 설명 가능합니다.

10. **공공데이터에 오류가 발생했을 때의 보정 구조는 마련되어 있나요?**

### Pack Your Jeju: 'Trust Engine' 핵심 전략

'Pack Your Jeju' 프로젝트의 'Trust Engine'은 제주 여행 정보의 신뢰도를 극대화하기 위한 다음 다섯 가지 핵심 전략을 기반으로 작동합니다. 이는 서비스의 foundational principle이자, 모든 기술적 구현의 근간이 됩니다.

#### 1. 공공데이터 기반의 근거 제시 및 RAG 엔진 활용
*   **원칙**: AI가 추천하는 모든 장소 및 활동에 대한 **명확한 근거와 출처를 제시**하여, 'AI가 왜 이 장소를 추천했는지 알기 어려움'이라는 문제점을 해결합니다.
*   **메커니즘**: `RAG (Retrieval-Augmented Generation)` 엔진을 활용하여 벡터 데이터베이스에서 관련 공공데이터 문서를 검색하고, 이를 LLM에 전달하여 근거 기반 답변 및 추천을 생성합니다.

#### 2. 사용자 피드백 루프를 통한 신뢰도 업데이트
*   **원칙**: 사용자의 **실제 방문 기록(방문함, 방문하지 않음, 변경함, 정보 불일치)**을 '신뢰 신호'로 시스템에 반영하여 데이터의 신뢰도를 지속적으로 개선하고, '다녀온 뒤의 경험이 다음 추천에 반영되지 않음' 문제를 해결합니다.
*   **메커니즘**: '신뢰도 루브릭'의 핵심 평가 항목에 사용자 피드백 신호를 포함시켜, 특히 '정보 불일치'와 같은 부정적 피드백은 신뢰도를 크게 하락시키는 요인으로 작용하게 합니다.

#### 3. '하루방 에이전트'를 통한 개인화된 정보 해석 및 실행 가능성 반영
*   **원칙**: LLM (`Google Gemini API`) 기반의 '하루방 에이전트'가 사용자의 특정 조건(날씨, 동반자, 렌터카 등)에 맞춰 공공데이터를 해석하고, 개인화된 설명을 생성하여 '비·바람·운영시간·이동거리 때문에 일정 실패 가능성' 및 '공공데이터의 맥락 부족' 문제를 해결합니다.
*   **메커니즘**: 여행 플랜의 '실행 가능성 조건 적합성'을 평가 항목으로 두어, 날씨, 운영 시간, 이동 수단/거리, 동행자 조건 등이 플랜에 얼마나 적합한지를 점수에 반영합니다.

#### 4. 정보 불일치 탐지 및 경고
*   **원칙**: 공공데이터와 실시간 정보(날씨, 지도 API 등) 간에 잠재적인 불일치가 발생할 경우 이를 **탐지하고 사용자에게 경고**하여, 사용자에게 정확하지 않은 정보로 인한 불편을 최소화합니다.
*   **메커니즘**: '정보 불일치 탐지 및 경고'를 신뢰도 루브릭의 주요 감점 요인으로 설정하여, 정보의 정확성을 확보합니다.

#### 5. 정보 최신성 유지 관리
*   **원칙**: 제공되는 공공데이터 및 기타 정보의 **업데이트 주기와 최신 상태를 유지**하는 것을 중요한 신뢰도 요소로 관리하여 정보의 유효성을 보장합니다.
*   **메커니즘**: '정보 최신성'을 신뢰도 루브릭의 감점 요인으로 두어, 오래된 정보가 추천 신뢰도를 하락시키는 것을 방지합니다.
