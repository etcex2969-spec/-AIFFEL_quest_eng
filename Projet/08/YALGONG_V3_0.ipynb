## 🏗️ 제안된 시스템 모듈 구조 (Architecture Proposal)

이 구조는 각 기능(인증, 분석, 학습)을 분리하여 유지보수성을 높이고, 가이드에서 언급한 **Robustness**와 **Modularization**을 실현합니다.

````python
import os
import logging
import json
import heapq
from typing import List, Dict, Any
from openai import OpenAI
from google.colab import userdata

# 로깅 설정
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger("OrchestratorV3")

class FinanceOrchestrator:
    """
    고도화된 지속 학습(Continual Learning) 및 우선순위 경험 재생이 탑재된 금융 오케스트레이터
    """
    def __init__(self, model: str = "gpt-3.5-turbo", buffer_capacity: int = 10):
        self.model = model
        self.client = self._init_client()
        # 핵심 기술: Prioritized Experience Replay Buffer (우선순위 큐 활용)
        # (priority, data) 형태로 저장하여 고득점 피드백 우선 추출
        self.knowledge_buffer = []
        self.buffer_capacity = buffer_capacity
        self.important_weights_metadata = {}

    def _init_client(self) -> OpenAI:
        try:
            key = userdata.get('OPENAI_API_KEY').strip()
            return OpenAI(api_key=key)
        except Exception as e:
            logger.error(f"인증 실패: {e}")
            raise

    def get_investment_advice(self, market_data: Dict[str, Any]) -> str:
        """
        우선순위가 높은 과거 지식을 동적으로 추출하여 RAG 기반 프롬프트 생성
        """
        # 버퍼에서 가장 점수가 높은 최상위 경험 3개 추출
        top_experiences = heapq.nlargest(3, self.knowledge_buffer, key=lambda x: x[0])
        past_knowledge_str = "\n".join([f"- [Score {exp[0]}] {exp[1]['feedback']}" for exp in top_experiences])

        prompt = f"""
        [실시간 시장 지표]: {market_data}
        [핵심 학습 지식 (Prioritized)]:
        {past_knowledge_str if past_knowledge_str else '축적된 지식이 없습니다. 기초 분석을 수행하세요.'}

        위 지식을 참조하여 현재 시장 상황에 대한 초정밀 투자 전략을 수립하세요.
        """

        try:
            response = self.client.chat.completions.create(
                model=self.model,
                messages=[{"role": "user", "content": prompt}]
            )
            return response.choices[0].message.content
        except Exception as e:
            logger.error(f"분석 오류: {e}")
            return "분석 불가"

    def apply_continual_learning(self, feedback_data: Dict[str, Any]):
        """
        지속 학습 엔진: 피드백 점수를 기반으로 우선순위 버퍼 업데이트 및 EWC 시뮬레이션
        """
        score = feedback_data.get('score', 0)

        # 1. 우선순위 기반 저장 (점수가 높을수록 중요하게 취급)
        if len(self.knowledge_buffer) >= self.buffer_capacity:
            # 가장 낮은 점수의 지식을 제거하고 새 지식 삽입 (Heap 관리)
            heapq.heappushpop(self.knowledge_buffer, (score, feedback_data))
        else:
            heapq.heappush(self.knowledge_buffer, (score, feedback_data))

        logger.info(f"✅ 지식 버퍼 갱신 완료 (현재 크기: {len(self.knowledge_buffer)}, 최신 점수: {score})")

        # 2. EWC 스타일의 중요 파라미터 메타데이터 관리
        task_id = feedback_data.get('task_type', 'general')
        self.important_weights_metadata[task_id] = max(self.important_weights_metadata.get(task_id, 0), score)
        logger.info(f"🔄 태스크 [{task_id}] 가중치 중요도 업데이트 완료")
````

### 🧪 고도화된 지속 학습(Continual Learning) 기능 테스트
아래 코드는 `Experience Replay` 버퍼에 지식이 축적되고, 이것이 다음 분석에 어떻게 반영되는지 시뮬레이션합니다.

````python
def simulate_advanced_learning():
    orchestrator = FinanceOrchestrator()

    # 1. 초기 상태 확인
    print(f"[초기 상태] 지식 버퍼 크기: {len(orchestrator.knowledge_buffer)}")

    # 2. 고품질 피드백 학습 (Experience Replay 데이터 축적)
    print("\n--- [지식 습득 단계] ---")
    learning_data = {
        "task_type": "sector_analysis",
        "feedback": "반도체 사이클 상승 국면에서는 시가총액 상위주 위주의 보수적 접근이 유효함",
        "score": 0.95
    }
    orchestrator.apply_continual_learning(learning_data)

    # 3. 학습된 지식이 반영된 두 번째 분석 요청
    print("\n--- [학습 반영 분석 단계] ---")
    market_context = {"Current_Event": "반도체 수출 지표 발표", "Trend": "Upward"}

    # get_investment_advice 내부에서 knowledge_buffer를 활용하여 프롬프트를 재구성함
    advice = orchestrator.get_investment_advice(market_context)

    print(f"분석 요청 결과: {advice[:100]}...")
    if "분석 불가" not in advice:
        print("\n✅ 성공: 과거 학습된 지식이 프롬프트에 주입되어 분석이 수행되었습니다.")
    else:
        print("\n⚠️ 참고: API 키 미설정 상태이므로 실제 응답 대신 로그를 확인하세요.")
        print(f"버퍼에 저장된 지식: {orchestrator.knowledge_buffer[0]['feedback']}")

if __name__ == "__main__":
    simulate_advanced_learning()
````

### 🧠 지속적 학습 엔진 (Continual Learning Engine)

이 모듈은 수집된 피드백을 관리하고, 특정 임계값에 도달했을 때 모델의 재학습(Fine-tuning) 또는 지식 업데이트를 트리거하는 역할을 수행합니다.

````python
class LearningEngine:
    """
    모델의 진화 및 피드백 관리를 담당하는 엔진 (Orchestrator의 우선순위 지식과 연동 가능)
    """
    def __init__(self, feedback_threshold: int = 5):
        self.feedback_buffer = []
        self.threshold = feedback_threshold
        self.version = 1.0

    def collect_feedback(self, orchestrator: FinanceOrchestrator, task_type: str, content: str, performance_score: float):
        """피드백 수집 및 Orchestrator의 지식 버퍼와 동기화"""
        feedback_entry = {
            "task_type": task_type,
            "feedback": content,
            "score": performance_score
        }

        # Orchestrator의 Prioritized Replay Buffer에 학습 데이터 전달
        orchestrator.apply_continual_learning(feedback_entry)

        # 엔진 자체의 파인튜닝 트리거용 버퍼 관리
        self.feedback_buffer.append(feedback_entry)
        logger.info(f"피드백 엔진 수집 완료. 현재 누적: {len(self.feedback_buffer)}/{self.threshold}")

        if len(self.feedback_buffer) >= self.threshold:
            self._trigger_evolution()

    def _trigger_evolution(self):
        """자동 파인튜닝 트리거 시뮬레이션"""
        logger.info(f"🚀 고가치 지식 기반 모델 진화 시작 (V{self.version} -> V{round(self.version + 0.1, 1)})")
        self.version = round(self.version + 0.1, 1)
        self.feedback_buffer.clear()
        logger.info("✅ 가중치 보존(EWC) 및 파인튜닝 완료")
````

````python
# LearningEngine의 feedback_buffer 데이터 구조 확인
test_engine = LearningEngine(feedback_threshold=5)
test_orchestrator = FinanceOrchestrator()

# 샘플 데이터 주입
test_engine.collect_feedback(
    orchestrator=test_orchestrator,
    task_type="macro_analysis",
    content="환율 급등에 따른 수출주 중심 포트폴리오 강화 권고",
    performance_score=0.92
)

# 버퍼의 첫 번째 항목 구조 출력
print("--- feedback_buffer 데이터 구조 ---")
if test_engine.feedback_buffer:
    import json
    sample_entry = test_engine.feedback_buffer[0]
    print(json.dumps(sample_entry, indent=4, ensure_ascii=False))
````

````python
import json

def prepare_finetuning_dataset(engine: LearningEngine, output_file: str = "finetuning_data.jsonl"):
    """
    LearningEngine의 피드BACK 버퍼를 OpenAI Fine-tuning용 JSONL 파일로 변환합니다.
    """
    dataset = []

    for entry in engine.feedback_buffer:
        # OpenAI Chat Completion 파인튜닝 포맷 구성
        message_format = {
            "messages": [
                {"role": "system", "content": "당신은 자율 진화형 금융 투자 어드바이저입니다."},
                {"role": "user", "content": f"[{entry['task_type']}] 관련 시장 상황에 대한 분석 피드백을 반영하세요."},
                {"role": "assistant", "content": entry['feedback']}
            ]
        }
        dataset.append(message_format)

    # JSONL 파일로 저장
    with open(output_file, "w", encoding="utf-8") as f:
        for item in dataset:
            f.write(json.dumps(item, ensure_ascii=False) + "\n")

    print(f"✅ 총 {len(dataset)}개의 데이터를 {output_file}로 변환 완료했습니다.")
    return dataset

# 변환 실행 및 미리보기
converted_data = prepare_finetuning_dataset(test_engine)

print("\n--- JSONL 데이터 샘플 미리보기 ---")
if converted_data:
    print(json.dumps(converted_data[0], indent=2, ensure_ascii=False))
````

### 🔗 시스템 통합 실행 샘플 (Integrated Workflow)

이 섹션에서는 분석 엔진과 학습 엔진을 결합하여, 실시간 분석과 피드백 기반의 자율 진화가 동시에 이루어지는 과정을 시뮬레이션합니다.

````python
import heapq

def run_integrated_orchestrator():
    # 1. 시스템 컴포넌트 초기화 (버퍼 용량 5개로 제한)
    orchestrator = FinanceOrchestrator(model="gpt-3.5-turbo", buffer_capacity=5)
    learning_engine = LearningEngine(feedback_threshold=2)

    # 2. 고품질 및 저품질 피드백 데이터 주입 (우선순위 테스트용)
    print("--- [Step 1: 다차원 피드백 주입 및 지식 우선순위화] ---")
    feedbacks = [
        ("strategy", "상승장에서는 시가총액 상위주 위주 포트폴리오가 안정적임", 0.98), # 최우선 지식
        ("general", "단기 변동성에 일희일비하지 말 것", 0.45),               # 낮은 우선순위
        ("analysis", "환율 1350원 돌파 시 수출주 비중 확대 전략 유효", 0.88)       # 중간 우선순위
    ]

    for task, content, score in feedbacks:
        learning_engine.collect_feedback(orchestrator, task, content, score)

    # 3. 지능적 시장 분석 시뮬레이션
    print("\n--- [Step 2: 우선순위 지식 기반 시장 분석 요청] ---")
    current_market = {"KOSPI": 2700, "Exchange_Rate": 1360}
    advice = orchestrator.get_investment_advice(current_market)

    # API 실패 시에도 버퍼에서 선택된 지식 확인 (디버깅용)
    if advice == "분석 불가":
        top_3 = heapq.nlargest(3, orchestrator.knowledge_buffer, key=lambda x: x[0])
        print("\n[System Notice] API 인증 오류가 발생했으나, 내부 버퍼에서 다음의 우선순위 지식을 추출했습니다:")
        for i, (score, data) in enumerate(top_3):
            print(f"{i+1}. [점수: {score}] {data['feedback']}")
    else:
        print(f"AI 투자 전략: {advice[:100]}...")

    print(f"\n최종 상태 - 모델 버전: V{learning_engine.version}, 버퍼 지식 수: {len(orchestrator.knowledge_buffer)}")

if __name__ == "__main__":
    run_integrated_orchestrator()
````

````python
import yfinance as yf
from datetime import datetime

def fetch_real_market_data():
    """
    야후 파이낸스를 통해 주요 시장 지표를 수집합니다.
    """
    print("📡 실시간 시장 데이터 수집 중...")

    # 티커 설정 (^KS11: KOSPI, ^GSPC: S&P 500, KRW=X: 환율)
    tickers = {
        "KOSPI": "^KS11",
        "S&P500": "^GSPC",
        "USD_KRW": "KRW=X"
    }

    market_data = {}

    for name, ticker in tickers.items():
        try:
            data = yf.Ticker(ticker).history(period="1d")
            if not data.empty:
                current_price = data['Close'].iloc[-1]
                change = ((current_price - data['Open'].iloc[-1]) / data['Open'].iloc[-1]) * 100
                market_data[name] = f"{current_price:.2f} ({change:+.2f}%)"
            else:
                market_data[name] = "데이터 없음"
        except Exception as e:
            market_data[name] = f"오류: {e}"

    market_data["Timestamp"] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    return market_data

def run_real_data_analysis():
    # 1. 오케스트레이터 초기화
    orchestrator = FinanceOrchestrator()

    # 2. 실제 데이터 가져오기
    real_data = fetch_real_market_data()
    print(f"📊 수집된 지표: {real_data}")

    # 3. 분석 수행
    print("\n--- [실제 데이터 기반 AI 분석] ---")
    advice = orchestrator.get_investment_advice(real_data)

    if advice == "분석 불가":
        print("\n⚠️ API 연결 실패: API 키와 잔액을 확인해주세요. 하지만 수집된 데이터는 정상입니다.")
    else:
        print(f"\n[AI 투자 전략]:\n{advice}")

if __name__ == "__main__":
    run_real_data_analysis()
````

## 📘 핵심 클래스 구조 및 동작 원리 상세 설명

이 시스템은 **지능(Intelligence)**을 담당하는 분석 엔진과 **성장(Growth)**을 담당하는 학습 엔진이 상호작용하며 자율적으로 진화합니다.

### 1. FinanceOrchestrator (시장 분석 엔진)
이 클래스는 투자 전략의 브레인 역할을 수행하며, OpenAI API와의 안정적인 인터페이스를 제공합니다.

*   **역할**: 시장 데이터를 입력받아 LLM(GPT)이 이해할 수 있는 형태의 프롬프트로 변환하고, 최적화된 투자 조언을 생성합니다.
*   **주요 매커니즘**:
    *   **초기화 (`_init_client`)**: API 키의 무결성을 검사하고 공백을 제거하여 런타임 오류를 최소화합니다.
    *   **분석 로직 (`get_investment_advice`)**: 시장 지표를 동적으로 프롬프트에 삽입합니다. API 장애나 쿼터 초과 시 프로그램이 중단되지 않도록 예외 처리가 되어 있으며, 모든 과정은 `logging` 모듈을 통해 추적됩니다.

### 2. LearningEngine (자율 진화 엔진)
이 클래스는 '지속적 학습(Continual Learning)'의 파이프라인을 관리하여 시간이 지날수록 모델이 정교해지도록 돕습니다.

*   **역할**: 생성된 조언에 대한 피드백(수익률, 정확도 등)을 수집하고, 학습이 필요한 시점을 스스로 판단하여 모델 업데이트를 수행합니다.
*   **주요 매커니즘**:
    *   **피드백 버퍼링**: 모든 피드백마다 학습을 수행하는 대신, `feedback_buffer`를 활용하여 데이터를 효율적으로 누적합니다.
    *   **자동 트리거 (`_trigger_evolution`)**: 누적된 데이터가 설정된 `threshold`(임계값)를 넘어서면 자동으로 파인튜닝 로직을 실행하고 시스템 버전을 업데이트합니다.
    *   **품질 검증**: 피드백 수집 시 `performance_score`를 함께 저장하여, 추후 양질의 데이터로만 모델을 고도화할 수 있는 기반을 마련합니다.

### 3. 협업 워크플로우 (Synergy)
1.  **분석**: `FinanceOrchestrator`가 현재 시장을 분석하여 사용자에게 조언을 제공합니다.
2.  **수집**: 해당 조언에 따른 실제 시장의 반응이나 수익률 데이터가 `LearningEngine`에 피드백으로 전달됩니다.
3.  **진화**: 피드백이 일정 수준 쌓이면 `LearningEngine`이 모델을 업데이트하여, 다음 분석 시에는 이전의 실수나 성공 사례를 반영하도록 합니다.

## 🚀 실시간 및 지속적 학습(Online/Continual Learning) 상세 가이드

데이터 입출력 시점에 AI 모델을 튜닝하여 성능을 지속적으로 개선하는 메커니즘에 대한 심층 가이드입니다.

### 1. 개념: 실시간/지속적 튜닝이란?
*   **정의**: AI 모델이 새로운 데이터가 들어올 때마다(입력) 또는 결과를 생성할 때마다(출력) 모델 파라미터를 업데이트하여 성능을 개선하는 학습 방식입니다. 기존 배치(batch) 학습과 달리, 데이터가 순차적으로 도착할 때마다 즉각 반영하는 특징이 있습니다.
*   **목적**:
    *   환경 변화에 빠르게 적응
    *   사용자 피드백 즉시 반영
    *   모델 성능 저하 방지 및 지속적 개선

### 2. 구체적 동작 과정
1.  **데이터 수집 및 전처리**
    *   새로운 입력 데이터(사용자 질의, 센서 데이터, 피드백 등)를 수집
    *   노이즈 제거, 정규화, 특징 추출 등 전처리 수행
2.  **모델 업데이트 트리거**
    *   데이터가 일정량 누적되거나, 특정 이벤트(피드백 수신 등) 발생 시 업데이트 시작
    *   즉시 업데이트(Online Learning) 또는 미니배치(Mini-batch) 단위로 주기적 업데이트 선택
3.  **파라미터 튜닝**
    *   **미니배치 SGD**: 새로운 데이터로 손실 함수 계산 후 경사하강법으로 파라미터 조정
    *   **재학습(Fine-tuning)**: 기존 모델에 새 데이터로 추가 학습
    *   **메타러닝**: 빠른 적응을 위한 초기 파라미터 조정
    *   **강화학습**: 보상 신호 기반 정책 업데이트
4.  **검증 및 모니터링**
    *   업데이트 후 검증 데이터로 성능 평가
    *   성능 저하 시 롤백 또는 추가 조정
    *   로그 및 지표 모니터링
5.  **모델 배포**
    *   업데이트된 모델을 운영 환경에 반영
    *   A/B 테스트, Canary 배포 등으로 안정성 확보

### 3. 기술적 고려사항
| 항목 | 설명 및 대응 방안 |
| :--- | :--- |
| **망각 문제 (Catastrophic Forgetting)** | 새로운 데이터 학습 시 기존 지식 손실 방지 (EWC, Replay Buffer 활용) |
| **계산 비용** | 실시간 업데이트 부담 완화 위해 미니배치, 경량화 모델 사용 |

| **보안 및 개인정보** | 민감 데이터 처리 시 암호화 및 접근 제어 강화 |
