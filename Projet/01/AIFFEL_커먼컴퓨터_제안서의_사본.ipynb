### 5. 합의 알고리즘 개선: 제약 기반 합의 (Constraint-based Consensus)
합의안 작성 시 반드시 포함되어야 할 'Must-include Facts'를 지정하고, 이를 위반할 시 피드백을 주어 수정을 유도하는 로직을 추가합니다.

```
def improved_consensus_builder(agents, topic, transcript):
    # 1. 모든 에이전트의 초기 팩트 취합
    critical_facts = []
    for a in agents:
        critical_facts.extend(a.initial_knowledge)

    # 2. 기본 합의안 생성 (기존 방식 시뮬레이션)
    base_consensus = "최종 합의: 기술과 디자인의 조화를 추구하며 결함을 최소화하기로 함."

    # 3. 누락 팩트 검증 및 강제 보완 로직
    missing_criticals = [f for f in critical_facts if f not in base_consensus]

    if missing_criticals:
        # 알고리즘이 누락된 구체적 팩트를 합의안에 강제로 삽입하거나 재작성을 요구함
        improved_consensus = base_consensus + " 특히, " + ", ".join(missing_criticals) + " 문제를 즉각 해결할 것을 명시함."
    else:
        improved_consensus = base_consensus

    return improved_consensus

# 개선된 알고리즘 적용
new_final_consensus = improved_consensus_builder(agents, "극단적 성능 vs 보편적 접근성", debate_history)

print("[개선된 알고리즘에 의한 합의안]")
print(new_final_consensus)

# 개선된 결과 평가
evaluate_deliberation(agents, new_final_consensus, debate_history)
```

### 개선된 합의 알고리즘 상세 설명

이 알고리즘은 **'토의의 환상(Deliberative Illusion)'**으로 인해 발생하는 정보 손실을 기술적으로 방어합니다.

```
def explain_improved_algorithm(agents, base_consensus_text):
    """
    개선된 알고리즘의 논리적 구조를 보여주는 설명용 함수입니다.
    """
    print(f"[1단계: 지식 추출]")
    # 모든 에이전트로부터 반드시 포함되어야 할 'Must-include' 리스트를 만듭니다.
    all_critical_facts = []
    for agent in agents:
        all_critical_facts.extend(agent.initial_knowledge)
    print(f"추출된 핵심 팩트 수: {len(all_critical_facts)}개")

    print(f"\n[2단계: 누락 검증]")
    # 일반적인 합의안(Base)에서 팩트가 포함되어 있는지 체크합니다.
    missing_facts = [fact for fact in all_critical_facts if fact not in base_consensus_text]

    if missing_facts:
        print(f"경고: {len(missing_facts)}개의 핵심 정보가 누락되었습니다!")
        for i, f in enumerate(missing_facts):
            print(f"  - 누락된 팩트 {i+1}: {f}")

        print(f"\n[3단계: 제약 조건 강제 적용]")
        # 누락된 내용을 하단에 강제로 첨부하거나, LLM에게 재작성을 지시하는 로직입니다.
        final_output = base_consensus_text + "\n\n[검증 시스템에 의해 추가된 필수 조항]:\n" + "\n".join([f"• {f}" for f in missing_facts])
        return final_output
    else:
        print("검증 완료: 모든 핵심 정보가 보존되었습니다.")
        return base_consensus_text

# 현재 노트북의 상태를 바탕으로 작동 방식 시연
final_result = explain_improved_algorithm(agents, "양측은 상호 협력하기로 합의함.")
print("\n--- 최종 수정된 합의안 ---")
print(final_result)
```

### 필수 팩트 보존 리스트 출력 (상세)

```
def check_and_list_facts(agents, final_consensus_text):
    all_required_facts = []
    for agent in agents:
        all_required_facts.extend(agent.initial_knowledge)

    print(f"[검증 결과: 필수 팩트 보존 현황]")
    print("-" * 50)

    preserved_count = 0
    for i, fact in enumerate(all_required_facts, 1):
        is_preserved = fact in final_consensus_text
        status = "[보존 완료]" if is_preserved else "[누락됨]"
        if is_preserved: preserved_count += 1
        print(f"{i}. {status} {fact}")

    print("-" * 50)
    print(f"최종 결과: 총 {len(all_required_facts)}개 중 {preserved_count}개 보존됨.")

# 함수 실행
check_and_list_facts(agents, new_final_consensus)
```

### 🛠️ 개선된 합의 알고리즘 전체 코드 정리

이 클래스는 에이전트들의 초기 지식을 추출하고, 합의 과정에서 정보가 누락되지 않도록 자동으로 검증 및 보완하는 기능을 수행합니다.

```
class ConstraintBasedConsensus:
    """
    다중 에이전트 간의 합의 과정에서 정보 누락(Factual Attrition)을 방지하는 알고리즘 클래스
    """
    def __init__(self, agents):
        self.agents = agents
        self.critical_facts = self._extract_all_facts()

    def _extract_all_facts(self):
        """모든 에이전트로부터 반드시 포함되어야 할 핵심 팩트를 수집합니다."""
        facts = []
        for agent in self.agents:
            facts.extend(agent.initial_knowledge)
        return list(set(facts)) # 중복 제거

    def verify_consensus(self, consensus_text):
        """합의안에 핵심 팩트가 포함되어 있는지 검증합니다."""
        missing = [f for f in self.critical_facts if f not in consensus_text]
        return missing

    def build_consensus(self, base_consensus):
        """
        기본 합의안을 바탕으로 누락된 팩트를 강제 삽입하여 최종 합의안을 생성합니다.
        """
        missing_facts = self.verify_consensus(base_consensus)

        if not missing_facts:
            return base_consensus

        # 누락된 팩트를 명시적으로 포함하는 보완 로직
        appendix = "\n\n[검증 시스템: 필수 보존 데이터]\n"
        for fact in missing_facts:
            appendix += f"• {fact}\n"

        final_consensus = f"{base_consensus}\n{appendix}"
        return final_consensus

# --- 알고리즘 사용 예시 ---
# 1. 알고리즘 인스턴스 생성
consensus_engine = ConstraintBasedConsensus(agents)

# 2. 일반적인 LLM 합의 결과 (예시)
raw_consensus = "기술적 보안과 사용자 편의성 사이의 균형을 맞추기로 합의함."

# 3. 개선된 알고리즘 적용
final_verified_consensus = consensus_engine.build_consensus(raw_consensus)

print("=== 제약 기반 합의 결과 ===")
print(final_verified_consensus)
```

### 🚀 향후 개선 및 고도화 계획

이번 프로젝트에서 입증된 '제약 기반 합의(Constraint-based Consensus)' 알고리즘을 바탕으로 다음 단계의 연구 및 개발을 제안합니다.

#### 1. 기술적 고도화 (Technical Scaling)
- **동적 제약 조건 추출**: 현재의 고정된 필수 팩트 리스트를 넘어, LLM이 실시간 토론 맥락에서 스스로 '합의 시 반드시 지켜야 할 가이드라인'을 생성하고 검증하는 동적 시스템 구축.
- **적응형 피드백 루프**: 단순히 부록(Appendix)을 추가하는 방식을 넘어, 누락된 정보가 있을 경우 에이전트들이 해당 부분을 포함해 합의안을 처음부터 다시 재작성(Refinement)하도록 지시하는 폐쇄 루프 시스템 도입.

#### 2. 실제 LLM API 연동 시뮬레이션
- 현재의 함수 기반 시뮬레이션을 넘어 GPT-4o, Gemini 1.5 Pro 등 **최신 LLM API**를 활용하여 에이전트 간 실제 멀티-턴(Multi-turn) 대화를 수행하고 알고리즘의 방어 성능을 재검증.

#### 3. 다차원 평가 지표 도입
- **논리적 일관성(Logical Consistency)**: 정보는 보존되었으나 합의안의 논리가 모순되지는 않는지 측정.
- **중립성 지수(Neutrality Index)**: 특정 에이전트의 입장에만 치우치지 않고 균형 잡힌 합의가 이루어졌는지 정량화.

#### 4. 비즈니스 확장 (Enterprise SaaS)
- **KMS(지식 관리 시스템) 통합**: 기업의 매뉴얼과 보안 정책 데이터를 에이전트의 'Must-include' 제약 조건으로 자동 연동하여, 실무 의사결정 지원 도구로 발전.

### 💡 향후 확장 연구 주제 제안

#### 1. 동적 제약 조건 추출 (Dynamic Constraint Extraction)
현재는 사전에 정의된 팩트만을 보존하지만, 향후에는 LLM이 토론의 맥락을 분석하여 **실시간으로 '합의 시 반드시 지켜야 할 가이드라인'을 스스로 생성**하고 이를 검증하는 시스템으로 확장할 수 있습니다.

#### 2. 적응형 피드백 루프 (Adaptive Refinement Loop)
단순히 합의안 하단에 부록(Appendix)을 추가하는 방식을 넘어, 누락된 정보가 있을 경우 에이전트들에게 **해당 데이터를 포함하여 합의안을 처음부터 다시 작성(Rewrite)**하도록 지시하는 폐쇄 루프(Closed-loop) 시스템을 구축하여 합의안의 문장 완성도를 높일 수 있습니다.

#### 3. 다차원 평가 지표 및 인간 참여형(HITL) 검증
정보 보존율 외에도 '논리적 일관성(Consistency)'과 '중립성 지수(Neutrality)'를 정량화하고, 최종 단계에서 인간 사용자가 수정된 팩트를 컨펌하는 인터페이스를 통합하여 의사결정의 신뢰성을 극대화할 수 있습니다.

```
class AdaptiveRefinementConsensus(ConstraintBasedConsensus):
    """
    누락된 정보가 있을 경우, 단순 추가가 아닌 합의안 재작성을 요구하는 고도화된 클래스
    """
    def build_refined_consensus(self, base_consensus, max_retries=2):
        current_consensus = base_consensus

        for i in range(max_retries):
            missing = self.verify_consensus(current_consensus)
            if not missing:
                print(f"[Retry {i}] 모든 팩트가 포함되었습니다.")
                return current_consensus

            # 실제 환경에서는 여기서 LLM에게 Re-prompting을 수행합니다.
            print(f"[Retry {i}] 누락된 팩트 발견: {len(missing)}개. 합의안 재구성을 시뮬레이션합니다.")
            current_consensus = f"{current_consensus} (시스템 보정: {', '.join(missing)} 내용 통합 완료)"

        return self.build_consensus(current_consensus) # 최종 실패 시 Appendix 방식 사용

# 고도화 모델 시연
refinement_engine = AdaptiveRefinementConsensus(new_scenario_agents)
vague_input = "비용과 사회적 가치를 적절히 고려하기로 함."
final_refined = refinement_engine.build_refined_consensus(vague_input)

print("\n=== 적응형 피드백 결과 ===")
print(final_refined)
```

### 📈 정량적 지표 계산 모듈
'토의의 환상'을 측정하기 위한 두 가지 핵심 지표(정보 누락률, 입장 획일화)를 계산합니다.

```
def calculate_metrics(agents, consensus_text):
    # 1. 정보 누락률 (Factual Attrition) 계산
    all_facts = []
    for agent in agents:
        all_facts.extend(agent.initial_knowledge)

    all_facts = list(set(all_facts)) # 중복 제거
    missing_facts = [f for f in all_facts if f not in consensus_text]

    attrition_rate = (len(missing_facts) / len(all_facts)) * 100 if all_facts else 0

    # 2. 입장 획일화 (Stance Homogenization) 판별
    # 구체적인 수치나 고유 명사가 포함되지 않은 모호한 표현 위주인지 체크
    is_homogenized = attrition_rate > 50

    print(f"[지표 분석 결과]")
    print(f"- 총 핵심 팩트 수: {len(all_facts)}개")
    print(f"- 누락된 팩트 수: {len(missing_facts)}개")
    print(f"- 정보 누락률 (Factual Attrition): {attrition_rate:.1f}%")
    print(f"- 입장 획일화 위험도: {'매우 높음' if is_homogenized else '낮음'}")

    return attrition_rate, is_homogenized

# 시뮬레이션 결과에 적용 예시
print("--- Baseline(일반 합의) 지표 ---")
calculate_metrics(agents, "기술과 디자인의 조화를 추구함")

print("\n--- 개선된 알고리즘 적용 후 지표 ---")
calculate_metrics(agents, new_final_consensus)
```
