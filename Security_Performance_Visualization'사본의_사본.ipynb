```
from pptx import Presentation
from pptx.util import Pt
from pptx.dml.color import RGBColor

# 파일 로드
file_path = '/content/Autonomous_Financial_Security_Center_Styled.pptx'
prs = Presentation(file_path)

# 스타일 설정 값
THEME_FONT = 'Arial'
THEME_COLOR = RGBColor(0, 51, 102)  # Dark Blue

for slide in prs.slides:
    for shape in slide.shapes:
        if not hasattr(shape, "text_frame"):
            continue

        for paragraph in shape.text_frame.paragraphs:
            for run in paragraph.runs:
                run.font.name = THEME_FONT
                run.font.color.rgb = THEME_COLOR

                # 제목(Title) 타입의 도형인 경우 더 크게 설정
                if shape == slide.shapes.title:
                    run.font.size = Pt(40)
                    run.font.bold = True
                else:
                    run.font.size = Pt(18)

# 전체 테마가 적용된 파일 저장
full_theme_path = '/content/Autonomous_Financial_Security_Center_FullTheme.pptx'
prs.save(full_theme_path)
print(f"전체 슬라이드에 테마가 적용되었습니다: {full_theme_path}")
```

```
from pptx import Presentation
from IPython.display import display, Markdown
import os

# 수정 대상 및 결과 경로 설정
input_path = '/content/Autonomous_Financial_Security_Center_Styled.pptx'
output_path = '/content/Autonomous_Financial_Security_Center_Final_v2.pptx'

if os.path.exists(input_path):
    prs = Presentation(input_path)
    # 슬라이드 14 (인덱스 13) 제목 수정
    if len(prs.slides) >= 14:
        slide_14 = prs.slides[13]
        if slide_14.shapes.title:
            slide_14.shapes.title.text = "Security Performance Visualization"
        else:
            # 제목 틀이 없는 경우 새로 생성
            from pptx.util import Inches
            left, top, width, height = Inches(0.5), Inches(0.3), Inches(9), Inches(1)
            title_shape = slide_14.shapes.add_textbox(left, top, width, height)
            title_shape.text = "Security Performance Visualization"

        prs.save(output_path)

        # Markdown 형식으로 결과 출력
        display(Markdown(f"### Slide 14 Verification\n**Title:** {slide_14.shapes.title.text if slide_14.shapes.title else 'Security Performance Visualization'}"))
    else:
        print(f"슬라이드 개수 부족: {len(prs.slides)}개")
else:
    print(f"입력 파일을 찾을 수 없습니다: {input_path}")
```

```
from pptx import Presentation
import os

file_path = '/content/Autonomous_Financial_Security_Center_Final_v2.pptx'

if os.path.exists(file_path):
    prs = Presentation(file_path)
    script_context = []

    print("--- 슬라이드 내용 분석 및 대본 생성 준비 ---\n")
    for i, slide in enumerate(prs.slides):
        content = []
        for shape in slide.shapes:
            if hasattr(shape, 'text') and shape.text.strip():
                content.append(shape.text.strip())

        slide_text = " / ".join(content)
        script_context.append(f"Slide {i+1}: {slide_text}")
        print(f"Slide {i+1} 분석 완료")

    # 분석된 내용을 바탕으로 대본 초안 작성 (Markdown)
    print("\n--- 분석 완료! 이제 이 내용을 바탕으로 대본을 작성합니다. ---")
else:
    print(f"오류: {file_path} 파일을 찾을 수 없습니다.")
```
