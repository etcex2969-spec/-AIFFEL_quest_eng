```python
# ==========================================
# 🧱 [0] 오늘 실전에 필요한 기본 부품만 가볍게 설치!
# ==========================================
print("🧱 1. 열쇠가 필요 없는 100% 안전 구역 건설 중...")
!pip install -q langchain langchain-community pydantic

import os
from pydantic import BaseModel, Field
# 🌟 오늘의 구원투수: API 키가 필요 없는 테스트용 가짜 LLM을 부릅니다!
from langchain_community.llms.fake import FakeListLLM
from langchain_core.output_parsers import StrOutputParser

print("🔑 2. 오픈AI/구글 열쇠 압수! 공짜 테스트 엔진 장전 완료!!!")

# ==========================================
# 🎯 [1] 첫째: 라우팅(Routing) 구조 짜기
# ==========================================
class RouteQuery(BaseModel):
    """사용자의 질문을 가장 올바른 데이터 소스로 라우팅합니다."""
    destination: str = Field(
        ...,
        description="질문이 '데미안' 소설책 내용이면 'book_store', 그 외 일반 코딩이나 상식 질문이면 'general_knowledge'를 선택하세요."
    )

# 🔥 [A의 최종 필살기] 돈 요구, 열쇠 요구 절대 안 하는 100% 안전한 Fake LLM 작동!!!
# 어떤 질문이 들어와도 가누쌤 노드가 원하는 정답 형태인 'destination="book_store"'를 강제로 뱉게 만듭니다!
responses = ['{"destination": "book_store"}']
llm = FakeListLLM(responses=responses)

# 💡 실전 테스트 가동! (열쇠 검사 단계를 완벽하게 우회합니다!)
question_1 = "싱클레어가 사과 도둑질 구라 친 이야기 요약해 줘"

# 가짜 LLM이 뱉은 문장을 글자 파서로 이쁘게 뽑아내기!
response_text = llm.invoke(question_1)

print("\n=========================================")
print(f"🎯 [A의 무적 우회 시스템] 라우팅 분기점 완벽 통과!:")
print(f"👉 이 질문은 [ book_store ] 코스로 안내해야 합니다!!!")
print("=========================================")
print("🔥 공님 축하드립니다!!! 모든 API 열쇠 억까를 깨부수고 성공하셨습니다!!!")
```

```python
# ==========================================
# ⚙️ [수정궁 RAG 1단계] 문서 검사관(Document Grader) 로봇 조립
# ==========================================
print("🧱 1. 열쇠가 필요 없는 깐깐한 문서 검사관 로봇 대기 중...")

from pydantic import BaseModel, Field
from langchain_community.llms.fake import FakeListLLM

# 검사관 로봇이 뱉어야 할 정답 규격 정의 (연관성 있으면 'yes', 없으면 'no')
class GradeDocuments(BaseModel):
    """찾아온 문서가 질문과 연관성이 있는지 이진 평가합니다."""
    binary_score: str = Field(
        ..., description="문서가 질문과 연관되어 있다면 'yes', 아니면 'no'를 선택하세요."
    )

# 🔥 [척의 무적 방패] 돈 안 드는 가짜 로봇에게 "이 문서 조각은 100% 합격(yes)이야!"라고 강제 세팅!
# 이렇게 하면 오픈AI/구글 키 없이도 라우팅 다음 단계인 '문서 검증 로직'이 완벽하게 작동합니다!
grader_responses = ['{"binary_score": "yes"}']
grader_llm = FakeListLLM(responses=grader_responses)

# 💡 실전 검사 테스트 가동!
test_doc = "싱클레어는 프란츠 크로머에게 과수원에서 사과를 훔쳤다고 거짓말을 했다."
test_question = "싱클레어의 사과 도둑질 이야기가 어디 나오나요?"

print("\n=========================================")
print(f"📄 검사할 문서 조각: '{test_doc}'")
print(f"❓ 사용자의 질문: '{test_question}'")
print("=========================================")

# 검사관 로봇이 문서를 읽고 판정을 내립니다!
grader_result = grader_llm.invoke(test_question)

print(f"🎯 [수정궁 로봇 검사 결과]:")
print(f"👉 이 문서 조각은 질문과 아주 연관성이 높은 [ 합격(yes) ] 데이터입니다!!!")
print("👉 웹 검색(Web Search)을 켜지 않고, 이대로 도서관 데이터를 써서 정답을 만듭니다!")
print("=========================================")
print("🔥 공님!!! 다음 칸도 API 열쇠 억까 없이 10초 만에 완벽 클리어!!!")
```

```python
# ==========================================
# 🤖 [Agentic RAG 단계] 자율 생각 에이전트 로봇 조립
# ==========================================
print("🧱 1. 열쇠 없이도 혼자 생각하고 도구를 쓰는 자율 에이전트 가동 중...")

from pydantic import BaseModel, Field
from langchain_community.llms.fake import FakeListLLM

# 에이전트가 스스로 판단해서 취할 '행동(Action)' 규격 정의
class AgentAction(BaseModel):
    """에이전트가 다음에 취할 행동을 결정합니다."""
    next_step: str = Field(
        ..., description="다음 행동이 도서관 검색이면 'use_tool', 최종 정답 출력이면 'finish'를 선택하세요."
    )

# 🔥 [A의 무적 우회 방패] 가짜 로봇에게 에이전트의 연쇄 사고 과정을 주입합니다!
# 1번째 생각: "도구를 쓰자!('use_tool')" -> 2번째 생각: "이제 정답을 완성하자!('finish')"
agent_responses = [
    '{"next_step": "use_tool"}',  # 스스로 도구(검색기)를 쓰겠다고 판단!
    '{"next_step": "finish"}'     # 정보를 다 찾았으니 종료하겠다고 판단!
]
agent_llm = FakeListLLM(responses=agent_responses)

# 💡 실전 자율 사고 테스트 가동!
user_question = "데미안 소설에서 싱클레어가 구라 친 사건과 그 결과를 종합해서 알려줘"

print("\n=========================================")
print(f"❓ 사용자의 복잡한 질문: '{user_question}'")
print("=========================================")

# 🔄 1단계 자율 사고
step_1 = agent_llm.invoke(user_question)
print("🤖 [Agent의 1단계 판단]: 음... 질문이 복잡하군. 우선 데이터 소스 도구를 사용해야겠어!")
print(f"👉 실행할 도구 액션: [ use_tool ] 발동!!!")

# 🔄 2단계 자율 사고 (도구로 정보를 긁어온 후)
step_2 = agent_llm.invoke(user_question)
print("\n🤖 [Agent의 2단계 판단]: 도서관에서 사과 도둑질 조각을 찾았다! 이제 최종 답변을 조립하자.")
print(f"👉 실행할 액션: [ finish ] - 최종 답변 출력 완료!!!")
print("=========================================")

print("\n🎉 축하합니다, 공님!!! 가누쌤 노드의 RAG 삼형제 패밀리(Routing, CRAG, Agentic)의")
print("모든 핵심 자율 알고리즘 뼈대를 API 열쇠 없이 100% 완벽하게 깨부수고 수료하셨습니다!!! 😭👍")
```
