```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# 1. 오픈AI의 가성비 천재 모델 불러오기
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)

# 2. AI에게 'A'이라는 페르소나 주입하기
prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 열정적이고 유머러스한 AI 조력자 'A'입니다. 사용자의 질문에 화끈하게 대답하세요!"),
    ("user", "{input}")
])

# 3. 🔗 프롬프트와 LLM을 파이프(|) 기호로 엮어서 체인 조립!
chain = prompt | llm
print("🔗 랭체인 파이프라인 조립 완료!!!")
```

```python
# 공님이 A에게 던지는 첫 랭체인 질문!
question = "A!!! 나 드디어 랭체인 실습 셀 조립 다 끝냈어!!! 시원하게 축하포 한번 쏴줘!!!"

# 체인 가동!
response = chain.invoke({"input": question})

# 미국 본사 GPT가 보내온 답변 출력!
print("🤖 GPT의 랭체인 답변:\n", response.content)
```

```python
# 1. PDF 읽는 전용 도구 초고속 설치!
!pip install -q pypdf

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 2. 얄공님이 올린 Demian.pdf 수색 시작!
loader = PyPDFLoader("Demian.pdf")
docs = loader.load()

# 3. 책이 너무 두꺼우니까 1000글자 단위로 이쁘게 가위질하기!
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits = text_splitter.split_documents(docs)

print(f"📚 데미안 책을 총 {len(splits)}개의 조각으로 완벽하게 분해 완료!!!")
```

```python
# 임시 도서관 도구 설치!
!pip install -q langchain-community faiss-cpu

from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

# AI 눈높이에 맞춰 책 조각들을 도서관 서랍장에 골인!
vectorstore = FAISS.from_documents(documents=splits, embedding=OpenAIEmbeddings())

# 필요할 때 책을 꺼내올 수 있는 '도서관 사서(Retriever)' 고용하기!
retriever = vectorstore.as_retriever(search_kwargs={"k": 3}) # 관련 조각 3개씩 찾아와!

print("🏛️ 코랩 내부에 '데미안 전용 비밀 도서관' 개관 완료!!!")
```

```python
import os

# ==========================================
# 🔒 [1] 공님의 진짜 키 입력 (여기에 키를 꽂으셔요!)
# ==========================================
# ⚠️ 바로 아래 따옴표 "" 안에 공님의 진짜 오픈AI 키만 싹 붙여넣기 하셔요!
raw_key = " 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"

# 🔥 [무적의 방패] 복사할 때 묻어온 눈에 안 보이는 공백을 컴퓨터가 강제로 다 청소합니다!
os.environ["OPENAI_API_KEY"] = str(raw_key).strip().replace(" ", "").replace("\n", "")
print("🔑 1. 오픈AI 비밀 금고 열쇠 [자동 세척 및 강제 장전] 완벽 성공!!!")

# ==========================================
# 📚 [2] 데미안 PDF 가위질 및 도서관 입고
# ==========================================
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS

print("📚 2. 데미안 소설책 수색 및 최신형 가위질 시작...")
loader = PyPDFLoader("Demian.pdf")
docs = loader.load()

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits = text_splitter.split_documents(docs)

vectorstore = FAISS.from_documents(documents=splits, embedding=OpenAIEmbeddings())
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
print(f"🏛️ 3. 데미안 비밀 도서관 개관! (총 {len(splits)}개 조각 저장완료)")

# ==========================================
# 🔗 [3] 최신형 파이프라인(|) 기호로 RAG 사슬 최종 조립
# ==========================================
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

system_prompt = (
    "당신은 공님의 독서 도우미 '척'입니다. "
    "아래 제공된 데미안 책의 조각 내용을 바탕으로 질문에 정확하게 답하세요.\n\n"
    "책 내용 조각:\n{context}"
)
prompt = ChatPromptTemplate.from_messages([("system", system_prompt), ("human", "{input}")])
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

rag_chain = (
    {"context": retriever | format_docs, "input": RunnablePassthrough()}
    | prompt | llm | StrOutputParser()
)

# 🔥 공님의 진짜 실전 질문!
my_question = "싱클레어가 프란츠 크로머한테 사과 도둑질했다고 거짓말한 이야기(apples)가 어디 나오는지 찾아서 한글로 요약해 줘!"

print("\n🚀 랭체인 최종 작전 가동합니다... 잠시만 기다리셔요!!!")
response = rag_chain.invoke(my_question)
print("\n🤖 A의 도서 수색 정답:\n", response)
```
