                                 에이전트 구축

```python
import time

# ==========================================
# 📊 [1] 랭그래프의 핵심: 장부(State) 정의하기
# ==========================================
class AgentState:
    def __init__(self, question):
        self.question = question   # 사용자의 질문
        self.context = ""          # 도서관에서 찾아온 문서 조각 메모장
        self.next_node = "START"   # 다음에 가야 할 노드 위치 고정

# ==========================================
# 🧱 [2] 랭그래프의 일꾼들: 노드(Nodes) 조립하기
# ==========================================
# [노드 1] 질문을 분석하고 어디로 갈지 결정하는 라우터 노드
def router_node(state):
    print("🤖 [라우터 노드]: 질문을 분석 중이다냥...")
    time.sleep(0.5)
    if "사과" in state.question or "데미안" in state.question:
        state.next_node = "retriever" # 도서관으로 가라!
    else:
        state.next_node = "general_llm" # 일반 상식으로 가라!

# [노드 2] 도서관에서 관련 문서를 슥 긁어오는 검색기 노드
def retriever_node(state):
    print("📚 [검색기 노드]: 벡터 도서관에서 싱클레어 사과 도둑질 문서를 찾았다냥!")
    time.sleep(0.5)
    state.context = "프란츠 크로머에게 사과를 훔쳤다고 구라 친 싱클레어 기록."
    state.next_node = "grader" # 이제 검사관한테 가라!

# [노드 3] 가져온 문서가 쓸만한지 검사하는 수정궁(CRAG) 검사관 노드
def grader_node(state):
    print("⚖️ [검사관 노드]: 가져온 문서가 질문이랑 맞는지 채점 중이다냥...")
    time.sleep(0.5)
    if state.context != "":
        print("✅ 판정: 문서 아주 훌륭하다냥! 합격!")
        state.next_node = "generator" # 정답 생성하러 가라!
    else:
        print("❌ 판정: 헛소리다냥! 웹 검색 켜야겠다냥!")
        state.next_node = "web_search"

# [노드 4] 최종 완벽한 정답을 조립하는 생성기 노드
def generator_node(state):
    print("✨ [생성기 노드]: 조립된 데이터를 바탕으로 최종 답변을 작성한다냥!")
    time.sleep(0.5)
    state.next_node = "END" # 대장정 종료!

# ==========================================
# 🔄 [3] 랭그래프 엔진(의사결정 루프) 가동!!!
# ==========================================
print("🚀 [LangGraph] 공님 전용 무적 에이전트 그래프 빌드 완료!!!")
print("=========================================================")

# 실전 테스트 상태 장전!
current_state = AgentState("싱클레어가 사과 구라 친 이야기 요약해 줘냥")
print(f"❓ 사용자 첫 질문: '{current_state.question}'\n")

# 그래프 엣지(길잡이)를 따라 노드가 🔄 자율 체인 회전을 시작합니다!
step = 1
while current_state.next_node != "END":
    print(f"🏃‍♂️ [Step {step}] 현재 위치 ➡️ 기반 노드 작동 개시!")

    if current_state.next_node == "START":
        router_node(current_state)
    elif current_state.next_node == "retriever":
        retriever_node(current_state)
    elif current_state.next_node == "grader":
        grader_node(current_state)
    elif current_state.next_node == "generator":
        generator_node(current_state)

    print(f"➡️ [엣지 판단]: 다음 목적지는 [ {current_state.next_node} ] 일세!\n")
    step += 1
    time.sleep(0.5)

print("=========================================================")
print("🎉 [LangGraph 대성공] 에이전트 루프가 완벽하게 안전 구역(END)에 도달했다냥!!!")
print("🔥 공님!!! 랭그래프의 복잡한 노드 연쇄 순환 원리까지 10초 만에 완벽 정복이다냥!!!")
```
